# Common Interview Questions

### Q1. App Service or Container Apps?

- **App Service** for traditional web applications and simple APIs.
- **Container Apps** for Dockerized AI applications and microservices.

---

### Q2. Container Apps or AKS?

- **Container Apps** for simpler deployments with less operational overhead.
- **AKS** when full Kubernetes capabilities and advanced orchestration are required.

---

### Q3. Azure OpenAI or Azure AI Search?

- **Azure OpenAI** provides LLM inference.
- **Azure AI Search** provides keyword, semantic, and vector retrieval.

Both are used together in RAG.

---

### Q4. Blob Storage or AI Search?

- **Blob Storage** stores the original documents.
- **Azure AI Search** stores searchable indexes and vectors.

---

### Q5. Azure Functions or Container Apps?

- **Azure Functions** for event-driven, short-lived tasks (e.g., PDF ingestion).
- **Container Apps** for long-running FastAPI services.

# EPAM Senior Answer (3–4 Minutes)

> "For an enterprise AI application on Azure, I package the FastAPI and LangGraph application into a Docker image and store it in Azure Container Registry. For most AI APIs, I deploy it on Azure Container Apps because it provides serverless container execution with minimal operational overhead. For larger Kubernetes-based platforms, I use Azure Kubernetes Service (AKS). User traffic flows through Azure DNS, Front Door, API Management, and Application Gateway before reaching the application. Enterprise documents are stored in Azure Blob Storage, and Azure Functions handle event-driven ingestion such as generating embeddings and indexing them into Azure AI Search. Azure OpenAI provides LLM inference, while Azure Database for PostgreSQL stores application data and Azure Cache for Redis improves performance. Secrets are managed securely using Azure Key Vault, and monitoring is handled through Application Insights together with LangSmith for AI workflow tracing. CI/CD is implemented using Azure DevOps or GitHub Actions, enabling automated testing, Docker image creation, deployment, and rolling updates. This architecture is scalable, secure, and suitable for production-grade enterprise AI applications."
