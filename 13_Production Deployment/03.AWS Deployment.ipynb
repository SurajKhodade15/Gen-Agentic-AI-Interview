# 1. Amazon ECS (Elastic Container Service)

## What is ECS?

ECS is a managed container orchestration service for running Docker containers.

Think of it as:

```text
Docker Container

↓

Amazon ECS

↓

Running AI Application
```

---

## Why ECS?

Instead of

```text
Laptop

↓

docker run
```

Deploy

```text
Docker Image

↓

Amazon ECS

↓

Production
```

---

## ECS Architecture

```text
Docker Image

↓

Amazon ECR

↓

Amazon ECS

↓

Running Containers
```

---

## Use Cases

- FastAPI
- LangGraph
- AI APIs
- Backend Services

---

## Interview Answer

> ECS is a managed container orchestration service used to deploy Dockerized applications without managing virtual machines.

---

# ECS vs EC2

| ECS | EC2 |
|------|------|
| Managed Containers | Virtual Machine |
| Easy Deployment | Manual Server Management |
| Auto Scaling | Manual Scaling |

---

# Fargate

Most enterprise AI applications use

```text
Amazon ECS

+

Fargate
```

because

No server management.

---

# ECS vs EKS

| ECS | EKS |
|------|------|
| Easier | More Flexible |
| AWS Native | Kubernetes Standard |
| Simpler | More Complex |
| Best for small/medium | Best for large enterprises |

---

# When to Choose?

Use **ECS** when:

- Small or medium AI applications
- Simpler operations
- Faster deployment

Use **EKS** when:

- Many microservices
- Kubernetes expertise
- Multi-team platform
- Advanced orchestration needs

# Common Interview Questions

### Q1. ECS or EKS?

- **ECS** for simpler container deployments and smaller teams.
- **EKS** when Kubernetes capabilities or a multi-team platform are required.

---

### Q2. Why ECR?

To securely store and version Docker images used by ECS and EKS.

---

### Q3. Why S3?

To store documents, images, and other unstructured data used by AI applications.

---

### Q4. Why CloudWatch?

To monitor application health, collect logs, create dashboards, and configure alerts.

---

### Q5. Why IAM instead of access keys?

IAM roles provide temporary credentials and eliminate hardcoded secrets, improving security.

---

### Q6. Lambda or ECS?

Use **Lambda** for event-driven tasks (e.g., PDF ingestion, embedding generation triggers) and **ECS** for long-running APIs like FastAPI.

---

### Q7. Why Bedrock?

It provides managed access to foundation models and embedding models without managing GPU infrastructure.

# EPAM Senior Answer (3–4 Minutes)

> "For a production AI application, I containerize the FastAPI and LangGraph service using Docker and push the image to Amazon ECR. The application is deployed on Amazon ECS with Fargate for serverless container execution, while larger Kubernetes-based platforms can use Amazon EKS. User requests are routed through Route53, CloudFront, API Gateway, and an Application Load Balancer before reaching the FastAPI service. Documents are stored in Amazon S3, and event-driven ingestion is handled by AWS Lambda, which generates embeddings and indexes them into Amazon OpenSearch or Qdrant. The application interacts with Amazon Bedrock for LLM inference and Titan Embeddings, uses Amazon RDS for relational data, ElastiCache for Redis-based caching, and AWS Secrets Manager for credentials. IAM roles provide secure service-to-service access without hardcoded keys. CloudWatch monitors infrastructure, logs, and application metrics, while LangSmith traces AI workflows. CI/CD is implemented using GitHub Actions or AWS CodePipeline to automate testing, Docker image creation, deployment to ECR, and rolling updates on ECS. This architecture provides a secure, scalable, and production-ready deployment for enterprise AI applications."
