# 10. FastAPI JWT Example

Install

```bash
pip install python-jose passlib
```

Generate Token

```python
from jose import jwt

SECRET = "secret"

token = jwt.encode(
    {"user":"Suraj"},
    SECRET,
    algorithm="HS256"
)

print(token)
```

---

Decode

```python
payload = jwt.decode(
    token,
    SECRET,
    algorithms=["HS256"]
)

print(payload)
```

# 22. EPAM Senior Answer (2–3 Minutes)

> "In a production AI application, authentication and authorization are handled separately. Authentication verifies the user's identity, typically through Microsoft Entra ID or Amazon Cognito, which issue JWT access tokens after successful login. Each request includes the JWT in the Authorization header, and FastAPI validates the token before processing the request. After authentication, authorization is enforced using Role-Based Access Control (RBAC), ensuring users can only access resources permitted for their role. Access tokens are short-lived, while refresh tokens are used to obtain new access tokens without requiring users to log in again. For service-to-service communication, such as FastAPI calling AWS Bedrock or Azure OpenAI, I use IAM Roles on AWS or Managed Identities on Azure instead of JWTs. Secrets are securely stored in AWS Secrets Manager or Azure Key Vault, and all communication occurs over HTTPS. This approach provides a secure, scalable, and enterprise-ready authentication architecture."
