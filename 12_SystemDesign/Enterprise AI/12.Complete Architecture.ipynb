# 11. EPAM Senior Answer (5 Minutes)

> "The application starts with Route53 and CloudFront on AWS or Azure Front Door for DNS resolution, global routing, and CDN. Requests are routed through Amazon API Gateway or Azure API Management, which enforces authentication, rate limiting, and API policies. Traffic is then distributed by an Application Load Balancer or Azure Application Gateway to stateless FastAPI containers running on ECS Fargate or Azure Container Apps. FastAPI validates JWT tokens, applies RBAC, and checks Redis for cached responses. On a cache miss, it invokes a LangGraph workflow that orchestrates retrieval, tool execution, and reasoning. The retriever queries Qdrant, Amazon OpenSearch, or Azure AI Search to fetch the most relevant document chunks, which are sent to AWS Bedrock or Azure OpenAI for response generation. Chat history and metadata are stored in PostgreSQL, while uploaded files remain in Amazon S3 or Azure Blob Storage. Long-running ingestion tasks such as OCR, chunking, and embedding generation are handled asynchronously using Amazon SQS or Azure Service Bus with dedicated worker containers. Secrets are managed through AWS Secrets Manager or Azure Key Vault. Infrastructure is monitored with CloudWatch or Azure Monitor, and AI workflows are traced with LangSmith. The architecture is fully containerized with Docker, deployed through GitHub Actions to Amazon ECR and ECS or Azure Container Registry and Container Apps, enabling a secure, scalable, highly available, and production-ready enterprise AI platform."

## One Interview Tip

For **every system design interview**, finish your explanation by mentioning these **five production pillars**:

1. **Security** – JWT, RBAC, Secrets Manager/Key Vault, HTTPS
2. **Scalability** – Stateless APIs, Load Balancer, Auto Scaling
3. **Reliability** – Health Checks, Retries, Dead Letter Queue
4. **Observability** – CloudWatch/Azure Monitor + LangSmith
5. **Cost Optimization** – Redis Cache, Async Processing, Appropriate LLM Selection

Mentioning these consistently demonstrates that you're thinking like a **Senior Software Engineer** rather than just an AI application developer.
