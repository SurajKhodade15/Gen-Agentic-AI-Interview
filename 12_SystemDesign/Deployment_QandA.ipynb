## 171. **Docker architecture for AI applications?**

> **Answer:** “I containerize the AI application and keep the image immutable, while externalizing configuration, secrets, persistent state and managed AI services.”

```text
Docker Container
├── FastAPI
├── LangGraph
├── LangChain
├── Application Code
└── Dependencies
       │
       ├── Azure OpenAI
       ├── Azure AI Search
       ├── PostgreSQL
       └── Key Vault
```

### Best practices

- Multi-stage Docker build
- Small base image
- Non-root user
- Pinned dependencies
- Environment-specific configuration outside image
- No secrets inside image
- Health/readiness endpoints
- Vulnerability scanning

**One-liner:**

> **“The container should be stateless and immutable; state, secrets and configuration should live outside the container.”**

## 178. How do you perform **safe model upgrades**?

```text
Current Model
     │
     ├── 95% Traffic
     │
New Model
     └── 5% Traffic
          ↓
     Compare Metrics
          ↓
       Better?
      ┌────┴────┐
     Yes       No
      ↓          ↓
 Increase     Rollback
 Traffic
```

### Compare

- Accuracy
- Faithfulness
- Answer relevancy
- Tool selection
- Latency
- Token usage
- Cost
- Error rate

**One-liner:**

> **“I use offline golden-set evaluation followed by a controlled canary rollout and compare both technical and AI-quality metrics before full promotion.”**

## 179. How do you implement **blue-green/canary deployment**?

### Blue-Green

> **Answer:** “I maintain two production environments. Blue runs the current version and Green runs the new version. After validation, I switch traffic to Green.”

```text
             Load Balancer
                  ↓
          ┌───────┴───────┐
          ↓               ↓
        BLUE             GREEN
       Current             New
          │                 │
          └──────┬──────────┘
                 ↓
              Switch
```

**Benefit:** Fast rollback.

### Canary

> **Answer:** “I initially route a small percentage of traffic to the new version and gradually increase it if the metrics remain healthy.”

```text
100% → Blue

90% Blue + 10% Green
        ↓
75% + 25%
        ↓
50% + 50%
        ↓
0% + 100%
```

**One-liner:**

> **“Blue-green gives me fast environment-level switching, while canary gives me gradual exposure and real-production validation.”**

## 180. How do you roll back a **bad model/prompt release**?

> **Answer:** “I keep the previous model/prompt version immutable and make the active version configurable. If quality or operational metrics breach thresholds, I immediately switch traffic back to the previous known-good version.”

```text
Production
    ↓
Quality Degradation
    ↓
Alert
    ↓
Rollback
    ↓
Previous Model + Prompt
    ↓
Validate
```

### Example

```text
Current:
Model v2 + Prompt v7

Rollback:
Model v1 + Prompt v6
```

### Rollback triggers

- Faithfulness degradation
- Answer relevancy degradation
- Tool-selection errors
- Increased latency
- Increased error rate
- Cost spike
- Safety violations

**One-liner:**

> **“I make model and prompt versions independently deployable and reversible, so rollback is a configuration change rather than a code rebuild.”**

### 🔑 Strong deployment mental model

```text
CODE
 ↓
Docker
 ↓
CI/CD
 ↓
Unit + Security Tests
 ↓
RAGAS / DeepEval
 ↓
UAT
 ↓
Canary / Blue-Green
 ↓
Production
 ↓
Monitor
 ↓
Rollback if Quality Degrades
```

**Key interview statement:**

> **“For GenAI, deployment safety is broader than application deployment. I version the code, model, prompts, embeddings and evaluation dataset, validate them together, release progressively, monitor both infrastructure and AI-quality metrics, and maintain an immediate rollback path.”**
