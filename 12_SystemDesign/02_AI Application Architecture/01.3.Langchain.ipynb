```python
from langchain.tools import Tool
from langchain.agents import initialize_agent, AgentType
from langchain_openai import ChatOpenAI

# -----------------------------
# Tools
# -----------------------------
def web_search(query):
    return f"Web result for: {query}"

def pubmed_search(query):
    return f"PubMed result for: {query}"

web_tool = Tool(
    name="WebSearch",
    func=web_search,
    description="Use for weather and news questions."
)

pubmed_tool = Tool(
    name="PubMedSearch",
    func=pubmed_search,
    description="Use for disease, treatment and drug related questions."
)

# -----------------------------
# LLM
# -----------------------------
llm = ChatOpenAI(
    model="gpt-4o-mini",
    temperature=0
)

# -----------------------------
# Agent
# -----------------------------
agent = initialize_agent(
    tools=[web_tool, pubmed_tool],
    llm=llm,
    agent=AgentType.ZERO_SHOT_REACT_DESCRIPTION,
    verbose=True
)

# -----------------------------
# Conversation History
# -----------------------------
history = []

while True:

    question = input("You: ")

    if question.lower() == "exit":
        break

    answer = agent.invoke(question)

    history.append({
        "question": question,
        "answer": answer["output"]
    })

    print("Agent:", answer["output"])

print("\nConversation History")
print(history)
```
