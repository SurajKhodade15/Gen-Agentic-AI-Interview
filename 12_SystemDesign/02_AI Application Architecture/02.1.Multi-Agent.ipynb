# Multi-Agent Architecture (Enterprise AI System Design)

This is one of the hottest topics in **EPAM, Microsoft, Amazon, Deloitte, and Accenture** interviews for Agentic AI.

A common interview question is:

> **"When would you choose a Multi-Agent system instead of a Single Agent?"**

---

# 1. What is a Multi-Agent System?

## Definition

A **Multi-Agent System (MAS)** consists of **multiple specialized AI agents**, where each agent has a specific responsibility. Instead of one agent doing everything, work is divided among expert agents that collaborate to solve a task.

---

## Interview Answer

> "A Multi-Agent architecture consists of multiple specialized AI agents that collaborate to solve complex tasks. Each agent is responsible for a specific capability such as planning, retrieval, coding, validation, or reporting, improving modularity, scalability, and maintainability."

---

# 2. Single Agent vs Multi-Agent

| Single Agent | Multi-Agent |
|--------------|-------------|
| One agent handles everything | Multiple specialized agents |
| Simple workflows | Complex workflows |
| Easier to build | More complex |
| Lower latency | Slightly higher latency |
| Lower cost | Higher cost |
| Limited specialization | Domain experts |

# 15. Best Practices

✅ Keep each agent focused on one responsibility.

✅ Share state through LangGraph.

✅ Use a Supervisor for orchestration.

✅ Prevent infinite loops.

✅ Add timeouts and retry policies.

# 16. Common Mistakes

❌ Creating an agent for every tiny task.

❌ Agents calling each other endlessly.

❌ Large shared state.

❌ No supervision.

# 17. Interview Questions

### Q1. When would you use Multi-Agent?

When a problem requires multiple specialized domains, independent reasoning, or coordination across different systems.

---

### Q2. What is the role of the Supervisor?

To analyze the request, decide which agents to invoke, coordinate execution, and combine the results.

---

### Q3. Can agents call tools?

Yes.

Each agent can have its own tools, prompts, and memory.

---

### Q4. Can agents run in parallel?

Yes.

Independent tasks such as HR lookup and Payroll lookup can execute in parallel, reducing overall latency.

---

### Q5. How do agents communicate?

Through shared **LangGraph state** or structured messages passed between nodes.

---

### Q6. Does every agent need its own LLM?

Not necessarily.

A common production pattern is to use **one shared Bedrock/OpenAI model** with different prompts for each agent. Different models are only used when specific capabilities or cost optimizations justify them.

# 20. EPAM Senior Answer (2–3 Minutes)

> "I choose a Multi-Agent architecture when the problem spans multiple domains or requires specialized reasoning. Instead of one agent handling every task, I introduce a Supervisor Agent that analyzes the user's request and delegates work to domain-specific agents such as an HR Agent, Payroll Agent, Retrieval Agent, or SQL Agent. Each agent has its own prompt, tools, and responsibilities but shares state through LangGraph. The agents may execute sequentially or in parallel depending on dependencies. Their outputs are aggregated by the Supervisor and passed to AWS Bedrock or Azure OpenAI to generate a coherent final response. The application runs behind FastAPI with JWT authentication, uses Redis for caching, PostgreSQL for business data, Qdrant or OpenSearch for retrieval, CloudWatch or Azure Monitor for infrastructure monitoring, and LangSmith for AI observability. This architecture improves modularity, scalability, and maintainability while keeping each agent focused on a single responsibility."
