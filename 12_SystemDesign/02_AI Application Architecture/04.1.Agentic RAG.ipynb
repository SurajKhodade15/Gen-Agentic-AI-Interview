# 1. What is Agentic RAG?

## Definition

**Agentic RAG** combines **RAG + AI Agents**.

Instead of following a fixed pipeline, an **LLM Agent** reasons about the user's request and decides:

- Do I need retrieval?
- Which tool should I call?
- Do I need SQL?
- Do I need an API?
- Should I ask a follow-up question?

---

## Interview Answer

> "Agentic RAG extends traditional RAG by introducing an intelligent agent that can reason, plan, decide whether retrieval is necessary, select appropriate tools, and orchestrate multiple steps before generating the final response."

# 15. Interview Questions

### Q1. Why Agentic RAG?

To enable intelligent decision-making instead of forcing every request through the same retrieval pipeline.

---

### Q2. Why LangGraph?

Because workflows involve conditional routing, state management, retries, loops, and tool orchestration.

---

### Q3. Does every question need retrieval?

No.

Greetings, calculations, or user-specific database queries may not require vector retrieval.

---

### Q4. Can Agentic RAG use SQL?

Yes.

The agent can decide to call SQL instead of a vector database.

---

### Q5. Can Agentic RAG call multiple tools?

Yes.

One request can invoke a retriever, SQL database, REST API, and calculator before generating the final response.

> **Production Note:** In a real LangGraph implementation, this routing is handled using **tool calling** (`bind_tools()`, `ToolNode`, `tools_condition`) rather than manual `if-elif` statements.

# 20. EPAM Senior Answer (2–3 Minutes)

> "Traditional RAG always follows a fixed sequence of retrieval followed by generation. In contrast, Agentic RAG introduces an intelligent agent—typically implemented with LangGraph—that first analyzes the user's intent and decides whether retrieval is actually required. Depending on the request, the agent may retrieve documents from Qdrant, OpenSearch, or Azure AI Search, execute SQL queries, call enterprise APIs, use calculators, or answer directly without any retrieval. The retrieved information or tool outputs are then combined and passed to AWS Bedrock or Azure OpenAI for response generation. This dynamic orchestration reduces unnecessary vector searches, lowers cost, improves response quality, and enables enterprise AI assistants to interact with multiple data sources and business systems. The solution is secured with JWT authentication, uses Redis for caching, PostgreSQL for business data, CloudWatch or Azure Monitor for infrastructure monitoring, and LangSmith for AI observability."
