# 1. What is RAG?

## Definition

**Retrieval-Augmented Generation (RAG)** is a technique where an LLM retrieves relevant information from an external knowledge base before generating a response.

Instead of relying only on its training data, the LLM uses your organization's documents.

---

## Interview Answer

> "RAG enhances an LLM by retrieving relevant information from an external knowledge base such as PDFs, SharePoint documents, or a vector database and providing that context to the LLM before generating the final response."

# 8. Why Chunking?

Suppose PDF

```text
500 Pages
```

LLM cannot process the whole document.

```text
Chunk 1

Chunk 2

Chunk 3
```

# 9. Why Embeddings?

LLMs don't search using keywords.

They search using **semantic meaning**.

Example

Question

```text
Leave Policy
```

Document

```text
Annual Vacation
```

Keyword search

↓

Misses it.

Embedding search

↓

Finds it.

# 18. Interview Questions

### Q1. Why RAG?

To provide the LLM with up-to-date, organization-specific knowledge and reduce hallucinations.

---

### Q2. Why Vector DB?

For semantic similarity search.

---

### Q3. Why Embeddings?

To represent text as vectors so semantically similar content can be retrieved.

---

### Q4. Why Chunking?

Because documents exceed the LLM context window and retrieval is more accurate with smaller sections.

---

### Q5. Why Retriever?

To retrieve the most relevant chunks before calling the LLM.

---

### Q6. Why S3?

To store original documents.

---

# 19. Scenario-Based Question

### Interviewer

> User uploads a 500-page HR policy PDF. Explain the complete flow.

Expected Answer

1. User uploads the PDF.
2. Store it in Amazon S3 or Azure Blob Storage.
3. Extract text (Textract or Document Intelligence if needed).
4. Split the text into chunks.
5. Generate embeddings.
6. Store embeddings in Qdrant, OpenSearch, or Azure AI Search.
7. During a query, embed the user's question.
8. Retrieve the most relevant chunks.
9. Send the question plus retrieved context to Bedrock or Azure OpenAI.
10. Generate the final answer.
11. Cache the response in Redis if appropriate.

# 21. EPAM Senior Answer (2–3 Minutes)

> "In a production RAG architecture, documents are uploaded to Amazon S3 or Azure Blob Storage. If the documents are scanned, OCR is performed using Amazon Textract or Azure AI Document Intelligence. The extracted text is divided into manageable chunks, converted into embeddings using an embedding model, and indexed in a vector database such as Qdrant, Amazon OpenSearch, or Azure AI Search. When a user submits a query, the same embedding model converts the question into a vector. The retriever performs a semantic search and retrieves the most relevant chunks, which are passed as context to AWS Bedrock or Azure OpenAI. The LLM generates the final answer using both the user's question and the retrieved context. Chat history is stored in PostgreSQL, frequently requested responses are cached in Redis, and CloudWatch or Azure Monitor together with LangSmith provide infrastructure and AI observability. This architecture delivers accurate, scalable, and enterprise-ready question answering while significantly reducing hallucinations."
