# API Gateway & Load Balancer (Enterprise AI System Design)

This is one of the most commonly asked **Senior Software Engineer** interview topics because almost every production application uses both.

---

# 1. What is API Gateway?

## Definition

An **API Gateway** is the **single entry point** for all client requests.

Instead of clients directly calling your FastAPI service, requests first go through the API Gateway.

---

## Interview Answer

> "An API Gateway acts as the front door of an application. It handles authentication, authorization, rate limiting, request routing, API versioning, logging, and security before forwarding requests to backend services."

✅ API Gateway should handle:

- JWT
- OAuth
- API Keys
- Rate Limiting
- API Versioning
- Request Validation (basic)
- Request/Response Transformation
- Logging

---

✅ Load Balancer should handle:

- SSL Termination
- Health Checks
- Traffic Distribution
- Failover
- Auto Scaling Integration

---

# Common Mistakes

❌ Thinking API Gateway and Load Balancer are the same.

❌ Performing JWT validation only in FastAPI (the gateway can validate or enforce auth too, but **the application should still validate authorization and trust boundaries**).

❌ No health check endpoint.

❌ Stateful FastAPI instances (making horizontal scaling difficult).

❌ Storing user sessions in application memory instead of Redis or another shared store.

# Follow-up Questions

### Q1. Which comes first?

**Answer**

```
User

↓

CloudFront / Front Door

↓

API Gateway

↓

Load Balancer

↓

FastAPI
```

---

### Q2. Why both API Gateway and Load Balancer?

**Answer**

The API Gateway manages **API concerns** (authentication, throttling, routing, versioning), while the Load Balancer manages **traffic concerns** (distribution, health checks, failover, and scaling).

---

### Q3. Can ALB replace API Gateway?

**Answer**

Not completely.

An ALB can route traffic and terminate SSL, but it doesn't provide full API management features such as usage plans, API keys, request transformation, and advanced throttling like API Gateway.

---

### Q4. Can API Gateway replace ALB?

**Answer**

Not for high-scale container workloads.

API Gateway manages APIs but doesn't replace a dedicated load balancer for distributing traffic across multiple container instances.

# EPAM Senior Answer (2–3 Minutes)

> "In an enterprise AI application, the request first reaches CloudFront or Azure Front Door for global routing and caching. It then goes to the API Gateway, such as Amazon API Gateway or Azure API Management, which acts as the front door for APIs by handling authentication, rate limiting, routing, API versioning, and logging. After that, the request is forwarded to an Application Load Balancer, which distributes traffic across multiple FastAPI instances running on ECS Fargate, EKS, Azure Container Apps, or AKS. The Load Balancer continuously performs health checks and routes traffic only to healthy instances, enabling high availability and horizontal scaling. FastAPI then orchestrates the LangGraph workflow, retrieves relevant context from the vector database, invokes the LLM through AWS Bedrock or Azure OpenAI, and returns the response. Using both API Gateway and Load Balancer separates API management concerns from traffic management, resulting in a secure, scalable, and production-ready architecture."
