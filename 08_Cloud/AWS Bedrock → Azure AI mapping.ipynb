# 10. Step Functions → Durable Functions / Logic Apps

This mapping requires nuance.

AWS:

```text
Step Functions
 ↓
Workflow orchestration
```

Azure has multiple options.

### Code-centric workflow

```text
AWS Step Functions
       ↓
Azure Durable Functions
```

### Integration/business workflow

```text
AWS Step Functions
       ↓
Azure Logic Apps
```

For AI agent workflows, you may instead use:

```text
LangGraph
```

because LangGraph handles **LLM/agent workflow orchestration**, whereas Durable Functions handles **distributed application workflow orchestration**.

# 23. Important Interview Question

### Q: "You have built an AI solution using AWS Bedrock. How would you migrate it to Azure?"

### Strong answer:

> "I would first separate the application into capability layers rather than trying to perform a direct service-by-service migration. Bedrock model invocation would map to Azure OpenAI or appropriate Azure AI Foundry model endpoints. S3 would map to Blob Storage, Bedrock Knowledge Bases to an Azure AI Search-based RAG architecture, Lambda to Azure Functions, API Gateway to API Management, DynamoDB to Cosmos DB, EventBridge to Event Grid, Secrets Manager to Key Vault, and CloudWatch to Azure Monitor with Application Insights.
>
> For the agent orchestration layer, I could use LangGraph, Azure AI Foundry Agent Service or Copilot Studio depending on the application's requirements. I would also redesign identity using Microsoft Entra ID, Managed Identity and Azure RBAC rather than simply translating AWS IAM policies."
