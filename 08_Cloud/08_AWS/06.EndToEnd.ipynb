# 5-Minute EPAM Answer

> "I start by gathering functional and non-functional requirements, such as expected traffic, latency, security, and compliance needs. Based on those, I design a stateless microservice architecture using FastAPI, LangGraph for agent orchestration, PostgreSQL for structured data, Qdrant for vector search, Redis for caching, and Celery for background processing. I package the application in Docker containers and manage the source code with Git. Every code push triggers a GitHub Actions CI pipeline that runs tests, performs code quality checks, builds a Docker image, and pushes it to Amazon ECR. The application is deployed on AWS EC2 behind an Application Load Balancer, with Route 53 handling DNS. For larger-scale deployments, I would use ECS or EKS for orchestration. Security is implemented using JWT authentication, RBAC authorization, HTTPS, AWS Secrets Manager for credentials, and LLM guardrails to prevent prompt injection and unauthorized SQL execution. The application is horizontally scalable using Auto Scaling Groups, Redis caching, asynchronous FastAPI endpoints, and connection pooling. Monitoring is handled with Prometheus, Grafana, LangSmith, and CloudWatch. For production resilience, I implement retries, fallback models, health checks, graceful degradation, and cost guardrails such as caching, token limits, and routing simple requests to smaller LLMs. This architecture is scalable, secure, observable, and suitable for enterprise-grade GenAI applications."
