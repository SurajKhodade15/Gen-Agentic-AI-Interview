# 1. What is Amazon Bedrock?

### Answer

> Amazon Bedrock is a fully managed AWS service that provides API access to foundation models from providers such as Anthropic, Meta, Amazon, and Cohere without managing infrastructure. It supports text generation, embeddings, RAG, tool calling, and agentic AI applications.

---

# 2. Why Bedrock instead of OpenAI API?

### Answer

> Bedrock provides enterprise security, IAM integration, VPC support, multiple foundation models through a single API, and native integration with AWS services like Lambda, S3, CloudWatch, and Knowledge Bases. It is well suited for organizations already operating in AWS.

---

# 3. Which models have you used?

### Answer

> I have primarily worked with Claude Sonnet on Amazon Bedrock for conversational AI and RAG applications. For embeddings, I have used Amazon Titan Embeddings.

# 5. What is Claude?

### Answer

> Claude is Anthropic's family of LLMs available through Amazon Bedrock. It is optimized for reasoning, long-context processing, coding assistance, and enterprise conversational AI.

# 7. What are Bedrock Knowledge Bases?

### Answer

> Bedrock Knowledge Bases is a managed RAG service that automatically ingests documents, chunks them, generates embeddings, stores vectors in a supported vector store, and retrieves relevant context before invoking the LLM.

---

# 8. Knowledge Base vs Custom RAG

| Knowledge Base | Custom RAG |
|----------------|------------|
| AWS managed | Fully customizable |
| Less code | More control |
| Automatic ingestion | Manual pipeline |
| Faster setup | Flexible architecture |

# 13. What is Step Functions?
