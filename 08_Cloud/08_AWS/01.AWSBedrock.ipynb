# ☁️ Amazon Bedrock - Quick Interview Cheat Sheet

| Topic | Definition | Why It Matters | AWS Service/API | Small Example / Use Case |
|-------|------------|----------------|-----------------|--------------------------|
| **1. What is Amazon Bedrock?** ⭐⭐⭐⭐⭐ | Fully managed AWS service to build GenAI applications using foundation models without managing infrastructure. | Simplifies building GenAI apps. | Amazon Bedrock | Build chatbot using Claude without hosting the model. |
| **2. Bedrock Architecture** | Bedrock sits between your application and foundation models, handling authentication, scaling, and inference. | Understand request flow. | Bedrock | User → Lambda → Bedrock → Claude → Response |
| **3. Foundation Models (FMs)** ⭐⭐⭐⭐⭐ | Pre-trained models provided by AWS partners and Amazon. | Choose the right model for the task. | Claude, Llama, Nova, Titan, Mistral, Cohere | Claude for chat, Titan for embeddings. |
| **4. InvokeModel API** | Low-level API to send prompts directly to a model. | Used for single model inference. | `InvokeModel` | Call Claude with a prompt and receive text. |
| **5. Converse API** ⭐⭐⭐⭐⭐ | Unified conversational API supporting multiple Bedrock models. | Recommended API for chat applications. | `Converse` | Multi-turn chatbot using Claude or Nova. |
| **6. Embedding Models** | Convert text into vectors for semantic search. | Required for RAG and vector search. | Titan Embeddings, Cohere Embed | Store document vectors in OpenSearch/Qdrant. |
| **7. Guardrails** ⭐⭐⭐⭐⭐ | Apply safety policies to model input/output. | Prevent unsafe or restricted responses. | Bedrock Guardrails | Block PII, hate speech, jailbreak attempts. |
| **8. Prompt Management** | Store, version, and reuse prompts centrally. | Improves consistency and maintainability. | Prompt Management | Reuse a prompt across multiple applications. |
| **9. Knowledge Bases** ⭐⭐⭐⭐⭐ | Managed RAG solution that handles ingestion, embeddings, retrieval, and context injection. | Simplifies enterprise RAG. | Knowledge Bases for Bedrock | Upload PDFs → Ask questions about documents. |
| **10. Agents for Bedrock** ⭐⭐⭐⭐⭐ | AI agents that plan tasks, call APIs, and use Knowledge Bases automatically. | Build autonomous workflows. | Bedrock Agents | Travel booking agent calling weather and hotel APIs. |
| **11. Model Evaluation** | Compare multiple foundation models using quality metrics. | Select the best model. | Bedrock Evaluation | Compare Claude vs Llama for summarization. |
| **12. Fine-Tuning / Custom Models** | Adapt supported foundation models using your own data. | Improve domain-specific performance. | Bedrock Custom Models | Fine-tune for healthcare or finance. |
| **13. Pricing & Best Practices** | Pricing based on tokens, model, inference type, and storage. | Optimize production cost. | Bedrock Pricing | Use prompt caching, smaller models, streaming. |

# Bedrock Request Flow

```text
User
   │
   ▼
Application
   │
   ▼
AWS Lambda
   │
   ▼
Amazon Bedrock
   │
   ▼
Foundation Model
(Claude/Llama/Nova)
   │
   ▼
Response
```

---

# RAG with Knowledge Base

```text
User Question
      │
      ▼
Knowledge Base
      │
      ▼
Retrieve Documents
      │
      ▼
Foundation Model
      │
      ▼
Grounded Response
```

---

# Bedrock Agent Flow

```text
User
   │
   ▼
Bedrock Agent
   │
   ├────────► Knowledge Base
   │
   ├────────► Lambda
   │
   ├────────► REST API
   │
   └────────► Database
            │
            ▼
      Final Response
```

# InvokeModel vs Converse API

| Feature | InvokeModel | Converse API |
|---------|-------------|--------------|
| Type | Low-level inference | Conversational API |
| Chat History | Manage manually | Built-in message format |
| Multi-turn Chat | Limited | Excellent |
| Recommended | Single prompt inference | Chatbots & AI Assistants |

---

# Knowledge Base vs Bedrock Agent

| Knowledge Base | Bedrock Agent |
|----------------|---------------|
| Retrieves information | Retrieves + Performs actions |
| RAG only | Agentic AI |
| Reads documents | Reads documents + Calls APIs |
| No planning | Planning & Tool Calling |

# Common Interview Questions

| Question | Short Answer |
|----------|--------------|
| What is Amazon Bedrock? | Managed service to access foundation models without managing infrastructure. |
| Why Bedrock instead of SageMaker? | Bedrock is for GenAI inference; SageMaker is for building and training ML models. |
| What is Converse API? | Unified API for conversational interactions with Bedrock models. |
| What is a Knowledge Base? | Managed RAG solution that retrieves relevant context before generation. |
| What are Bedrock Agents? | AI agents that orchestrate tools, APIs, and knowledge bases to complete tasks. |
| What are Guardrails? | Safety controls that filter harmful input/output and enforce policies. |
| What are Embedding Models used for? | Generate vectors for semantic search and RAG. |
| How do you reduce Bedrock cost? | Use smaller models, prompt caching, streaming, optimize prompts, and minimize token usage. |

# 30-Second Interview Answer

> **Amazon Bedrock is a fully managed AWS service for building Generative AI applications using foundation models such as Claude, Llama, Nova, and Titan. It provides APIs for inference, managed RAG through Knowledge Bases, autonomous workflows with Bedrock Agents, safety through Guardrails, and prompt management, allowing developers to build scalable GenAI applications without managing model infrastructure.**
