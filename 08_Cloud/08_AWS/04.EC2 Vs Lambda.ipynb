# AWS EC2 vs AWS Lambda (Interview Cheat Sheet)

| Feature | AWS EC2 | AWS Lambda |
|---------|----------|------------|
| Service Type | Virtual Machine (IaaS) | Serverless Compute (FaaS) |
| Server Management | Required | Not Required |
| Infrastructure | You manage OS, patches, scaling | AWS manages everything |
| Scaling | Manual / Auto Scaling Groups | Automatic |
| Billing | Pay while instance is running | Pay only for execution time |
| Startup Time | Instance launch takes time | Milliseconds (Warm Start), slower on Cold Start |
| Execution Time | Unlimited | Limited (up to 15 minutes per invocation) |
| State | Long-running | Stateless (per invocation) |
| Best For | Long-running applications | Event-driven applications |
| Traffic Pattern | Continuous traffic | Intermittent/Burst traffic |
| Maintenance | High | Very Low |
| Deployment | SSH, Docker, CI/CD | Upload function/code package |
| Resource Control | Full CPU/RAM/Disk control | Memory & timeout configuration only |
| Cost | Better for continuously running workloads | Better for infrequent or event-driven workloads |
| Availability | You configure HA | Built-in High Availability |
| Scaling Speed | Slower | Automatic and rapid |
| Operating System Access | Yes | No |
| Persistent Local Storage | Yes (EBS/EFS) | Temporary `/tmp` storage only |
| Common Languages | Any | Python, Java, Node.js, .NET, Go, etc. |
| Typical GenAI Use | Host vLLM, Ollama, FastAPI, GPU workloads | Backend APIs, Bedrock invocation, preprocessing |

# GenAI Architecture Comparison

## Using EC2

```text
User

↓

Load Balancer

↓

EC2

↓

FastAPI

↓

Bedrock / Ollama / vLLM

↓

Response
```

Suitable for:

- Self-hosted LLMs
- vLLM
- Ollama
- GPU inference
- Long-running APIs

---

## Using Lambda

```text
User

↓

API Gateway

↓

Lambda

↓

Amazon Bedrock

↓

Response
```

Suitable for:

- Chatbots
- RAG APIs
- Bedrock integration
- Event-driven AI workflows
- Serverless applications

# Advantages

| EC2 | Lambda |
|------|---------|
| Full server control | No server management |
| Can host your own models | Automatic scaling |
| Supports GPU instances | Pay per execution |
| No execution time limit | Easy integration with AWS services |
| Suitable for continuous workloads | Ideal for event-driven workloads |

---

# Limitations

| EC2 | Lambda |
|------|---------|
| Higher operational overhead | 15-minute execution limit |
| Pay even when idle | Cold start latency |
| Manual scaling | Limited runtime customization |
| Requires OS maintenance | Not suitable for long-running jobs |

# Which One Should You Choose?

| Scenario | Recommended Service |
|----------|---------------------|
| Amazon Bedrock Chatbot API | ✅ Lambda |
| Document Summarization API | ✅ Lambda |
| Event-driven AI Processing | ✅ Lambda |
| Run Ollama Locally | ✅ EC2 |
| Host vLLM | ✅ EC2 |
| GPU-based Inference | ✅ EC2 |
| Continuous FastAPI Service | ✅ EC2 |
| High-throughput Self-hosted LLM | ✅ EC2 |
| Scheduled AI Jobs | ✅ Lambda (or EventBridge + Lambda) |

# Interview Questions

### Q1. When would you choose Lambda over EC2?

**Answer:**

Choose Lambda for serverless, event-driven applications such as Bedrock APIs, chatbots, document processing, and RAG pipelines. It automatically scales and you pay only for execution time.

---

### Q2. When would you choose EC2 over Lambda?

**Answer:**

Choose EC2 when you need full control over the operating system, want to host your own LLMs (such as vLLM or Ollama), require GPU instances, or run long-lived services.

---

### Q3. Can Lambda host an LLM like Llama 3?

**Answer:**

Not practically. Large open-source models require significant memory, GPU resources, and long-running processes. Lambda is typically used to **invoke** hosted models (such as Amazon Bedrock), not to host them.
