# 1. Azure AI Overview / Azure AI Services

For your **Altimetrik AI Consultant** interview, first understand the **Azure AI ecosystem**, not individual services in isolation.

## 1. What is Azure AI?

> **Azure AI is Microsoft's portfolio of cloud services and tools for building, deploying, and operating AI applications, including Generative AI, computer vision, speech, language processing, document intelligence, search, and AI agents.**

Think of it as an **AI service ecosystem**, similar to how AWS provides multiple AI services around Bedrock.

```text
                    Azure AI
                       │
       ┌───────────────┼────────────────┐
       │               │                │
       ▼               ▼                ▼
  Generative AI    AI Services      AI Search
       │               │                │
 Azure OpenAI      Vision/Speech    RAG/Search
 AI Foundry        Language         Vector Search
       │           Document AI
       ▼
   AI Agents
```

---

# 2. What are Azure AI Services?

**Azure AI Services** are pre-built, managed AI APIs that allow developers to add AI capabilities without training models from scratch.

Major categories include:

| Category | Azure Service / Capability | Usage |
|---|---|---|
| **Generative AI** | Azure OpenAI | GPT-based applications, agents, RAG |
| **AI Development** | Azure AI Foundry | Build/evaluate/deploy AI applications |
| **Search** | Azure AI Search | Enterprise search, RAG, vector/hybrid search |
| **Vision** | Azure AI Vision | Image analysis, OCR, object detection |
| **Speech** | Azure AI Speech | Speech-to-text, text-to-speech |
| **Language** | Azure AI Language | NLP, sentiment, entities, summarization |
| **Documents** | Azure AI Document Intelligence | Extract data from PDFs/forms/documents |
| **Content Safety** | Azure AI Content Safety | Detect harmful/unsafe content |
| **Agents** | Azure AI Foundry Agent Service | Build/manage AI agents |

---

# 3. Very Important Distinction

Don't say:

> "Azure AI is one service."

Instead:

> **"Azure AI is an umbrella ecosystem containing multiple managed AI capabilities and platforms."**

For example:

```text
Azure AI
│
├── Azure OpenAI
├── Azure AI Search
├── Azure AI Foundry
├── Azure AI Vision
├── Azure AI Speech
├── Azure AI Language
├── Azure AI Document Intelligence
└── Azure AI Content Safety
```

# 4. Azure AI vs Azure OpenAI

This is highly likely to be asked.

| Azure AI | Azure OpenAI |
|---|---|
| Broad AI ecosystem | Specific generative AI service |
| Multiple AI capabilities | OpenAI models |
| Vision | GPT models |
| Speech | Embeddings |
| Language | Generative AI |
| Document Intelligence | LLM applications |
| Content Safety | Chat/completions/agent scenarios |
| AI Search | — |
| AI Foundry | — |

### Simple analogy

> **Azure AI = umbrella**

> **Azure OpenAI = one major service under that ecosystem**

# 5. Azure AI Services vs Building Your Own Model

Suppose you need OCR.

### Custom approach

```text
Collect Images
     ↓
Label Data
     ↓
Train Model
     ↓
Deploy Model
     ↓
Maintain Model
```

### Azure AI approach

```text
Document
   ↓
Azure AI Service
   ↓
Extracted Information
```

This reduces development and operational overhead.

# 7. Example — HR AI Assistant

Suppose Altimetrik asks you to build an enterprise HR assistant.

### User asks:

> "What is the maternity leave policy?"

Architecture:

```text
User
 ↓
Teams
 ↓
AI Application
 ↓
Azure AI Search
 ↓
HR Documents
 ↓
Azure OpenAI
 ↓
Grounded Answer
```

Here:

- **Azure AI Search** → retrieves relevant HR information
- **Azure OpenAI** → generates the response
- **Azure Blob Storage / SharePoint** → can hold source documents
- **AI application** → orchestrates the workflow

# 11. AWS Comparison — Important for You

Because your production experience is heavily AWS-oriented, use this mental mapping:

| AWS | Azure |
|---|---|
| **Amazon Bedrock** | **Azure OpenAI / Azure AI Foundry ecosystem** |
| Claude | GPT models through Azure OpenAI |
| Amazon Textract | Azure AI Document Intelligence |
| Amazon Transcribe | Azure AI Speech |
| Amazon Comprehend | Azure AI Language |
| Amazon Rekognition | Azure AI Vision |
| Bedrock Knowledge Bases | Azure AI Search + RAG |
| Bedrock Guardrails | Azure AI Content Safety |
| SageMaker | Azure Machine Learning / AI development ecosystem |
| S3 | Azure Blob Storage |
| Lambda | Azure Functions |
| CloudWatch | Azure Monitor |

**Important interview nuance:** these aren't all exact one-to-one equivalents. For example, **Azure AI Foundry is broader than simply being an equivalent of Bedrock**; it is an AI development platform covering model access, development, evaluation, and deployment workflows.
