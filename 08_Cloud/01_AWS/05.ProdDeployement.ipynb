# EPAM Interview Answer (2 Minutes)

> "I would package the FastAPI application into a Docker container and push the image to Amazon ECR. For deployment, I typically prefer ECS Fargate because it is serverless and eliminates infrastructure management. An Application Load Balancer distributes incoming traffic across multiple FastAPI containers and supports SSL termination and health checks. User-uploaded documents are stored in Amazon S3, optionally processed with Textract for OCR, chunked, embedded, and indexed into a vector database such as Pinecone or Qdrant. The FastAPI service orchestrates retrieval using LangChain or LangGraph and invokes an LLM through AWS Bedrock or OpenAI. Secrets such as API keys are managed with AWS Secrets Manager, while IAM roles enforce least-privilege access. Monitoring and centralized logging are handled by CloudWatch, and Auto Scaling adjusts the number of running containers based on traffic. For CI/CD, I use GitHub Actions to build the Docker image, push it to ECR, and trigger a new ECS deployment."

## Be Ready for These Follow-up Questions

1. **Why ECS Fargate instead of EC2?**
   - No server management, automatic scaling, pay for usage, simpler operations.

2. **Why ECR?**
   - It's AWS's managed Docker image registry, tightly integrated with ECS/EKS.

3. **Where do you store OpenAI or Bedrock credentials?**
   - AWS Secrets Manager (or IAM roles for AWS-native services like Bedrock).

4. **How do you scale the application?**
   - ECS Service Auto Scaling behind an Application Load Balancer.

5. **How do you monitor production issues?**
   - CloudWatch logs, metrics, alarms, and application-level observability (e.g., LangSmith for LLM tracing).

This level of explanation is generally what senior interviewers expect when discussing deployment architecture.
