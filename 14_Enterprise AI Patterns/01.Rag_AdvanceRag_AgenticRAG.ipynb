# 1. Traditional RAG

## Definition

Traditional RAG retrieves relevant documents from a vector database and sends them to the LLM.

It follows a **fixed pipeline**.

---

## Architecture

```text
                User
                  │
                  ▼
              FastAPI
                  │
                  ▼
             Retriever
                  │
                  ▼
     OpenSearch / Qdrant
                  │
                  ▼
            Top-K Chunks
                  │
                  ▼
       AWS Bedrock / Azure OpenAI
                  │
                  ▼
              Response
```

---

## Flow

```text
Question

↓

Embedding

↓

Retriever

↓

Vector DB

↓

Top 5 Chunks

↓

LLM

↓

Answer
```

---

## AWS Components

| Component | AWS |
|------------|------|
| API | FastAPI |
| LLM | Bedrock |
| Embedding | Titan Embedding |
| Vector DB | OpenSearch |
| Storage | S3 |

---

## Azure Components

| Component | Azure |
|------------|--------|
| API | FastAPI |

| Vector DB | Azure AI Search |

---

## Advantages

- Easy to build
- Good accuracy
- Low cost

---

## Limitations

Every request always goes through retrieval.

Even

```text
Hello
```

still performs

```text
Retriever

↓

Vector DB

↓

LLM
```

Wasteful.

# 2. Advanced RAG

Advanced RAG improves retrieval quality.

Instead of

```text
Question

↓

Retriever

↓

LLM
```

it introduces additional optimization steps.

---

## Architecture

```text
                     User
                       │
                       ▼
                  Query Rewrite
                       │
                       ▼
              Hybrid Search
                       │
         ┌─────────────┴──────────────┐
         ▼                            ▼
 Keyword Search               Vector Search
         │                            │
         └─────────────┬──────────────┘
                       ▼
                   Reranker
                       │
                       ▼
              Context Compression
                       │
                       ▼
         Bedrock / Azure OpenAI
                       │
                       ▼
                    Response
```

---

## Components

Instead of one retriever

Advanced RAG uses

- Query Rewriting
- Hybrid Search
- Metadata Filtering
- Parent-Child Retrieval
- Reranking
- Context Compression

---

## Flow

```text
Question

↓

Rewrite Question

↓

Hybrid Search

↓

Retrieve 20 Chunks

↓

Rerank

↓

Best 5 Chunks

↓

Compress Context

↓

LLM

↓

Answer
```

---

## Why?

Suppose user asks

```text
PTO policy
```

Document contains

```text
Annual Leave Policy
```

Query rewriting converts

```text
PTO

↓

Paid Time Off

↓

Annual Leave
```

Now retrieval improves.

---

## Advantages

- Better retrieval
- Better accuracy
- Fewer hallucinations
- Better ranking

---

## Limitations

- Higher latency
- More infrastructure
- More expensive

---

# 3. Agentic RAG

This is the latest generation.

Instead of always retrieving,

an **AI Agent decides** what to do.

---

## Architecture

```text
                             User
                               │
                               ▼
                      LangGraph Agent
                               │
             ┌─────────────────┼─────────────────┐
             ▼                 ▼                 ▼
         Retriever         SQL Tool         REST API
             │
             ▼
     OpenSearch / Qdrant
             │
             ▼
       Bedrock / Azure OpenAI
             │
             ▼
          Final Answer
```

---

## Flow

```text
Question

↓

Reason

↓

Need RAG?

↓

Need SQL?

↓

Need API?

↓

Need Calculator?

↓

Execute Tool

↓

LLM

↓

Answer
```

# Common Interview Questions

### Q1. Why not always use Agentic RAG?

Because it introduces additional complexity, more LLM calls, higher latency, and higher cost. If your application only needs document-based question answering, Traditional or Advanced RAG is often sufficient.

---

### Q2. When would you use Advanced RAG?

When retrieval quality is the primary challenge—for example, large document repositories, ambiguous user queries, or a need for semantic + keyword search.

---

### Q3. When would you use Agentic RAG?

When users need more than document retrieval, such as querying databases, calling enterprise APIs, performing calculations, or orchestrating multiple business systems.

---

# EPAM Senior Answer (3–4 Minutes)

> "Traditional RAG follows a fixed pipeline where every query is embedded, relevant documents are retrieved from a vector database, and the retrieved context is sent to an LLM. Advanced RAG improves this pipeline by introducing techniques such as query rewriting, hybrid search, metadata filtering, reranking, and context compression to improve retrieval accuracy and reduce irrelevant context. Agentic RAG goes one step further by introducing an LLM-based agent, typically orchestrated with LangGraph, that reasons about the user's intent before deciding whether retrieval is necessary at all. The agent can dynamically choose between document retrieval, SQL queries, REST APIs, calculators, or direct reasoning, and combine multiple tool outputs before generating the final response. In enterprise systems, I typically use Amazon Bedrock or Azure OpenAI for inference, OpenSearch or Azure AI Search for retrieval, ECS/EKS or Azure Container Apps/AKS for deployment, Redis for caching, PostgreSQL for business data, and CloudWatch or Application Insights together with LangSmith for observability. This layered evolution from Traditional RAG to Advanced RAG to Agentic RAG provides increasing flexibility, accuracy, and business capability."
