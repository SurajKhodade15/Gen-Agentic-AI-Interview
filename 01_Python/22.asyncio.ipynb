# 8. `asyncio` (Python)

## Interview Answer

`asyncio` is Python's built-in library for writing **asynchronous and concurrent** programs. It uses an **event loop** to run multiple I/O-bound tasks efficiently without creating multiple threads.

---

## Simple Example

```python
import asyncio

async def fetch():
    print("Fetching...")
    await asyncio.sleep(2)
    print("Done")

await fetch()      # Jupyter/Colab
# asyncio.run(fetch())   # Python script
```

**Output**

```text
Fetching...
(2 sec wait)
Done
```

---

## Real-World Example

Calling multiple LLM APIs concurrently:

```python
results = await asyncio.gather(
    call_openai(),
    call_claude(),
    call_gemini()
)
```

Instead of waiting for each API call one by one, all requests are sent concurrently, reducing total response time.

## Important Components

### `async`

Declares an asynchronous function.

```python
async def fetch():
    ...
```

---

### `await`

Pauses the current coroutine until the awaited task completes **without blocking** the event loop.

```python
await asyncio.sleep(2)
```

---

### `asyncio.gather()`

Runs multiple async tasks concurrently.

```python
await asyncio.gather(task1(), task2(), task3())
```

## Common Follow-up Questions

### Q. What is an Event Loop?

**Answer:**

The **event loop** is the core of `asyncio`. It schedules and manages asynchronous tasks, switching between them whenever one is waiting for an I/O operation.

---

### Q. Difference between `async` and `await`?

- `async` → Defines an asynchronous function.
- `await` → Waits for an async operation to complete without blocking other tasks.

---

### Q. What is `asyncio.gather()`?

**Answer:**

It executes multiple asynchronous tasks concurrently and returns all their results.
