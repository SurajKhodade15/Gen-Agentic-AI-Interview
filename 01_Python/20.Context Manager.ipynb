# 6. Context Managers

## Interview Answer

A **context manager** is used to **manage resources automatically**, such as opening and closing files or database connections. It is commonly used with the `with` statement.

---

## Simple Example

```python
with open("test.txt", "r") as file:
    data = file.read()

print(data)
```

Here, the file is **automatically closed**, even if an exception occurs.

---

## Without Context Manager

```python
file = open("test.txt", "r")
data = file.read()
file.close()
```

The developer must remember to call `close()`, which is error-prone.

---

## How It Works

A context manager implements two methods:

- `__enter__()` → Acquires the resource.
- `__exit__()` → Releases the resource (cleanup).

---

## Real-World Example

Database connection:

```python
with db_connection() as conn:
    conn.execute("SELECT * FROM users")
```

The connection is automatically closed after the block finishes.

---

## One-Line Interview Answer

> "A context manager automatically manages resources by performing setup and cleanup operations using the `with` statement."

## Common Follow-up Questions

### Q. Why use a context manager?

- Prevents resource leaks
- Handles exceptions safely
- Makes code cleaner and more readable

---

### Q. Where is it used?

- File handling
- Database connections
- Network sockets
- Thread locks
- Transactions

---

### Q. Which methods does a context manager implement?

- `__enter__()`
- `__exit__()`
