# 9. Multithreading vs Multiprocessing

## Interview Answer

- **Multithreading** uses **multiple threads within the same process**. Threads share the same memory and are best for **I/O-bound tasks**.
- **Multiprocessing** uses **multiple processes**, each with its own memory. It is best for **CPU-bound tasks**.

---

## Simple Example

### Multithreading

```python
from threading import Thread
import time

def task(name):
    time.sleep(2)
    print(name)

t1 = Thread(target=task, args=("A",))
t2 = Thread(target=task, args=("B",))

t1.start()
t2.start()

t1.join()
t2.join()
```

**Use Case:** Downloading files, API calls, database queries.

---

### Multiprocessing

```python
from multiprocessing import Process

def task():
    print("Running")

p1 = Process(target=task)
p2 = Process(target=task)

p1.start()
p2.start()

p1.join()
p2.join()
```

**Use Case:** Image processing, ML training, heavy computations.

---

## Difference

| Multithreading | Multiprocessing |
|---------------|-----------------|
| Multiple threads | Multiple processes |
| Shared memory | Separate memory |
| Best for I/O-bound tasks | Best for CPU-bound tasks |
| Affected by Python GIL | Not affected by GIL |

---

## Real-World Example

### Multithreading

- Calling multiple LLM APIs
- Reading multiple files
- Database operations
- Web scraping

### Multiprocessing

- Training ML models
- Processing large images/videos
- Feature engineering on large datasets

## One-Line Interview Answer

> "I use multithreading for I/O-bound tasks like API or database calls, and multiprocessing for CPU-bound tasks like model training or data processing."

---

## Common Follow-up Questions

### Q. Why doesn't multithreading improve CPU-bound performance in Python?

**Answer:**

Because of the **Global Interpreter Lock (GIL)**, only one thread can execute Python bytecode at a time.

---

### Q. Which would you use for OpenAI API calls?

**Answer:**

**Multithreading** (or preferably `asyncio` in modern Python) because API calls are I/O-bound.

---

### Q. Which would you use for model training?

**Answer:**

**Multiprocessing**, because training is CPU-intensive and benefits from multiple CPU cores.
