# 5. Closures

## Interview Answer

A **closure** is a function that **remembers and can access variables from its outer function even after the outer function has finished executing**.

---

## Simple Example

```python
def outer(msg):
    def inner():
        print(msg)
    return inner

greet = outer("Hello")
greet()
```

**Output**

```text
Hello
```

Even though `outer()` has finished executing, `inner()` still remembers the value of `msg`.

---

## Why are Closures Used?

- Data hiding (private variables)
- Maintaining state
- Building decorators
- Factory functions

> **Important:** Decorators are implemented using **closures**.

---

## Real-World Example

A counter that remembers its value:

```python
def counter():
    count = 0

    def increment():
        nonlocal count
        count += 1
        return count

    return increment

c = counter()

print(c())   # 1
print(c())   # 2
print(c())   # 3
```

The `count` variable is preserved between calls.

---

## One-Line Interview Answer

> "A closure is an inner function that retains access to variables from its enclosing scope even after the outer function has completed execution."

---

## Common Follow-up Questions

### Q. What is `nonlocal`?

**Answer:**
`nonlocal` allows an inner function to **modify** a variable from the enclosing (outer) function.

---

### Q. Where are closures used in Python?

- Decorators
- Callbacks
- Factory functions
- State management

---

### Q. Difference between Closure and Decorator?

| Closure | Decorator |
|---------|-----------|
| Remembers variables from the outer scope | Uses a closure to modify another function's behavior |
| Can be used independently | A specific use case built on closures |
