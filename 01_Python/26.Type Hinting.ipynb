# 12. Type Hinting

## Interview Answer

**Type hinting** is used to specify the expected data types of variables, function parameters, and return values. It improves **code readability**, **IDE support**, and **static type checking**.

---

## Simple Example

```python
def add(a: int, b: int) -> int:
    return a + b

print(add(10, 20))
```

Here:
- `a: int` → `a` should be an integer.
- `b: int` → `b` should be an integer.
- `-> int` → Function returns an integer.

---

## More Examples

### List

```python
def get_names() -> list[str]:
    return ["Suraj", "Rahul"]
```

---

### Dictionary

```python
def get_user() -> dict[str, int]:
    return {"age": 30}
```

---

### Optional

```python
from typing import Optional

def get_name() -> Optional[str]:
    return None
```

Means the function can return either a `str` or `None`.

---

## Why Use Type Hinting?

- Better readability
- IDE autocomplete
- Static type checking (`mypy`)
- Easier code maintenance
- Helps in large projects

## Common Follow-up Questions

### Q. Does Python enforce type hints at runtime?

**Answer:**
**No.** Type hints are **not enforced** at runtime. They are mainly used by IDEs and static analysis tools like `mypy`.

Example:

```python
def add(a: int, b: int) -> int:
    return a + b

print(add("10", "20"))
```

This **runs successfully** and prints:

```text
1020
```

Python doesn't stop you.

---

### Q. Which tool checks type hints?

**Answer:**
`mypy`

---

### Q. Why are type hints important in production?

- Better collaboration
- Easier debugging
- Fewer type-related bugs
- Improved developer experience with autocomplete and static analysis
