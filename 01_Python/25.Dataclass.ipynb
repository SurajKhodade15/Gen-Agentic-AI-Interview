# 11. Dataclass

## Interview Answer

A **dataclass** is a Python feature that automatically generates boilerplate methods like `__init__()`, `__repr__()`, and `__eq__()` for classes, making the code cleaner and easier to maintain.

Import:

```python
from dataclasses import dataclass
```

---

## Simple Example

Without `dataclass`:

```python
class Employee:
    def __init__(self, name, age):
        self.name = name
        self.age = age

emp = Employee("Suraj", 30)
```

With `dataclass`:

```python
from dataclasses import dataclass

@dataclass
class Employee:
    name: str
    age: int

emp = Employee("Suraj", 30)
print(emp)
```

**Output**

```text
Employee(name='Suraj', age=30)
```

Notice that you didn't write the `__init__()` method.

---

## Why Use Dataclass?

- Less boilerplate code
- Automatically creates `__init__()`
- Automatically creates `__repr__()`
- Automatically creates `__eq__()`
- Improves readability

---

## Real-World Example

Representing an LLM request:

```python
from dataclasses import dataclass

@dataclass
class ChatRequest:
    question: str
    model: str
    temperature: float
```

Instead of manually writing constructors, you can directly create objects:

```python
req = ChatRequest(
    "What is AI?",
    "gpt-4",
    0.7
)
```

## Common Follow-up Questions

### Q. What methods does `@dataclass` generate automatically?

- `__init__()`
- `__repr__()`
- `__eq__()`

---

### Q. When should you use a dataclass?

Use it for **data-holding objects** such as:
- API request/response models
- Configuration objects
- Employee/User models
- LLM request payloads

---

### Q. Dataclass vs Normal Class

| Dataclass | Normal Class |
|-----------|--------------|
| Less boilerplate | Manual `__init__()` required |
| Auto-generated methods | Write methods yourself |
| Best for storing data | Best for complex business logic |
