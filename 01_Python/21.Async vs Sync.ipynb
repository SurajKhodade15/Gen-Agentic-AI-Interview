# 7. Async vs Sync

## Interview Answer

- **Synchronous (Sync):** Tasks execute **one after another**. The next task waits until the current one finishes.
- **Asynchronous (Async):** A task can **pause while waiting** (e.g., for an API or database response), allowing other tasks to run in the meantime.

---

## Simple Example

### Sync

```python
import time

def task():
    time.sleep(2)
    print("Done")

task()
print("Finished")
```

**Output (after 2 seconds)**

```text
Done
Finished
```

---

### Async

```python
import asyncio

async def task():
    await asyncio.sleep(2)
    print("Done")

asyncio.run(task())
```

Here, `await` pauses the task without blocking the entire program.

## One-Line Interview Answer

> "Synchronous execution blocks until a task completes, whereas asynchronous execution allows other tasks to run while waiting for I/O operations, improving throughput."

---

## Common Follow-up Questions

### Q. When should you use Async?

Use it for **I/O-bound operations**, such as:
- API calls
- Database queries
- File operations
- LLM requests

---

### Q. When should you use Sync?

Use it for **CPU-bound tasks**, such as:
- Data processing
- Machine learning computations
- Image processing

---

### Q. What does `await` do?

**Answer:**
`await` pauses the current async function until the awaited operation completes, without blocking the event loop.
