# 10. GIL (Global Interpreter Lock)

## Interview Answer

The **Global Interpreter Lock (GIL)** is a mechanism in CPython that allows **only one thread to execute Python bytecode at a time**, even on a multi-core CPU.

---

## Simple Example

```python
from threading import Thread

def task():
    for _ in range(10000000):
        pass

t1 = Thread(target=task)
t2 = Thread(target=task)

t1.start()
t2.start()

t1.join()
t2.join()
```

Although two threads are created, **only one thread executes Python bytecode at a time** because of the GIL.

---

## Why Does Python Have a GIL?

To make **memory management and reference counting thread-safe**, simplifying the CPython implementation.

---

## Effect of GIL

✅ **I/O-bound tasks:** Minimal impact because threads release the GIL while waiting.

❌ **CPU-bound tasks:** Threads cannot execute Python bytecode in parallel, limiting performance.

---

## Real-World Example

### Good Use of Threads

- OpenAI API calls
- Database queries
- Reading files
- Network requests

### Bad Use of Threads

- Image processing
- Model training
- Matrix multiplication
- Large numerical computations

For CPU-intensive work, use **multiprocessing** instead.

---

## One-Line Interview Answer

> "The GIL allows only one thread to execute Python bytecode at a time, which limits multithreading for CPU-bound tasks but has little impact on I/O-bound tasks."

---

## Common Follow-up Questions

### Q. Does the GIL affect multiprocessing?

**Answer:**
**No.** Each process has its own Python interpreter and its own GIL, so processes can run truly in parallel.

---

### Q. Does the GIL affect async programming?

**Answer:**
**No.** `asyncio` is designed for **I/O-bound concurrency** and doesn't rely on multiple threads executing Python bytecode simultaneously.

---

### Q. When would you use multithreading despite the GIL?

For I/O-bound tasks such as:
- API calls
- Database operations
- File I/O
- Web scraping
