# 1. What do you understand by Evaluation (Eval) in Generative AI?

## Interview Answer (2–3 Minutes)

> Evaluation in Generative AI is the process of measuring the quality, accuracy, reliability, and usefulness of an AI model's output. Unlike traditional machine learning, where evaluation is based on metrics such as accuracy or F1-score, Generative AI produces open-ended responses, making evaluation more complex.
>
> In enterprise applications, we evaluate not only the final response but also the entire AI pipeline, including document retrieval, prompt quality, tool selection, reasoning, latency, token usage, and overall cost.
>
> For example, in a RAG application, evaluation includes checking whether the retriever returned the correct documents, whether the LLM remained faithful to the retrieved context without hallucinating, and whether the answer is relevant to the user's query.
>
> We typically use **LangSmith** for tracing and observability, **RAGAS** for RAG-specific evaluation, and **DeepEval** for automated testing of AI applications. Common metrics include Faithfulness, Hallucination, Answer Relevancy, Precision@K, Recall@K, MRR, NDCG, Latency, Cost, and Token Usage.

# Part 1: Evaluation + RAG (Interview Revision Notes)

## 1. What do you understand by evaluation (eval) in Generative AI?

Evaluation measures the quality, accuracy, relevance, and safety of LLM responses using predefined metrics.
It helps validate model performance before deployment and monitor it in production.

---

## 2. Why is evaluation important?

Evaluation identifies hallucinations, regressions, and poor responses before users experience them.
It ensures consistent, reliable, and high-quality AI outputs.

---

## 3. What types of evaluation are used in GenAI?

Common types are **automated evaluation**, **human evaluation**, and **LLM-as-a-Judge**.
They assess correctness, relevance, safety, and user satisfaction.

---

## 4. What is automated evaluation?

Automated evaluation uses predefined metrics or scripts to compare LLM outputs against expected results.
It enables fast, repeatable testing without manual review.

---

## 5. What is human evaluation?

Human evaluators assess response quality, usefulness, fluency, and correctness.
It is especially valuable for subjective tasks where automated metrics are insufficient.

---

## 6. What is LLM-as-a-Judge?

An LLM evaluates another model's response using scoring criteria or comparison prompts.
It reduces manual effort but should be validated against human judgment.

---

## 7. What metrics do you use for RAG evaluation?

Common metrics include **Precision@K**, **Recall@K**, **MRR**, **NDCG**, **Faithfulness**, and **Answer Relevance**.
These evaluate both retrieval quality and final answer quality.

---

## 8. What is Precision@K?

Precision@K measures how many of the top K retrieved documents are relevant.
Higher Precision@K means fewer irrelevant documents are retrieved.

---

## 9. What is Recall@K?

Recall@K measures how many relevant documents are retrieved from all available relevant documents.
Higher Recall@K reduces the chance of missing important context.

---

## 10. What is MRR (Mean Reciprocal Rank)?

MRR measures how early the first relevant result appears in the ranked list.
A higher MRR indicates users find relevant information faster.

---

## 11. What is NDCG?

NDCG evaluates ranking quality by rewarding highly relevant documents appearing near the top.
It is useful when documents have different relevance scores.

---

## 12. What is faithfulness?

Faithfulness measures whether the LLM's answer is supported by the retrieved context.
A faithful response avoids making unsupported claims.

---

## 13. What is answer relevance?

Answer relevance measures how well the generated answer addresses the user's question.
A relevant answer directly satisfies the user's intent.

---

## 14. What is hallucination?

Hallucination occurs when an LLM generates incorrect or fabricated information not supported by evidence.
It commonly results from missing context, poor retrieval, or model limitations.

---

## 15. How do you reduce hallucinations?

Use RAG, Hybrid Search, reranking, strong prompts, and instruct the model to answer only from retrieved context.
Return "I don't know" if sufficient evidence is unavailable.

---

## 16. What is RAG?

Retrieval-Augmented Generation combines information retrieval with an LLM to generate grounded answers.
The LLM answers using retrieved documents instead of relying only on its training.

---

## 17. Why do we use RAG?

RAG enables LLMs to answer questions using up-to-date enterprise data.
It reduces hallucinations and avoids retraining the model.

---

## 18. Explain the RAG workflow.

Documents are chunked, embedded, and stored in a vector database.
At query time, relevant chunks are retrieved and sent to the LLM with the user query.

---

## 19. What are the components of a RAG pipeline?

A RAG pipeline includes document loader, chunking, embedding model, vector database, retriever, prompt template, and LLM.
Each component contributes to accurate and efficient retrieval.

---

## 20. What is an embedding?

An embedding is a dense numerical vector representing the semantic meaning of text.
Similar texts have embeddings that are close in vector space.

---

## 21. Why are embeddings required?

Embeddings enable semantic search instead of keyword matching.
They help retrieve contextually similar documents.

---

## 22. What is semantic search?

Semantic search retrieves documents based on meaning rather than exact keyword matches.
It uses embeddings and similarity search to find relevant content.

---

## 23. What is vector similarity search?

Vector similarity search finds embeddings closest to the query embedding using distance metrics.
It powers semantic retrieval in RAG systems.

---

## 24. Which similarity metrics have you used?

Common metrics include **Cosine Similarity**, **Euclidean Distance**, and **Dot Product**.
Cosine similarity is the most widely used for text embeddings.

---

## 25. What is cosine similarity?

Cosine similarity measures the angle between two vectors to determine semantic similarity.
A value closer to 1 indicates greater similarity.

---

## 26. What is a vector database?

A vector database stores embeddings and performs efficient similarity search.
Examples include Qdrant, Milvus, Pinecone, and pgvector.

---

## 27. What happens after retrieval?

Retrieved chunks are combined with the user query in a prompt.
The LLM uses this context to generate the final response.

---

## 28. What is Top-K retrieval?

Top-K retrieval returns the K most relevant chunks based on similarity scores.
Typical values range from 3 to 10 depending on the application.

---

## 29. How do you choose Top-K?

Choose a value that provides enough context without adding irrelevant information.
Too small may miss information, while too large may increase noise and token cost.

---

## 30. What is reranking?

Reranking reorders retrieved documents using a more accurate model after initial retrieval.
It improves retrieval precision before sending context to the LLM.

---

## 31. Why use reranking?

Vector search retrieves candidate documents quickly but may not rank them optimally.
Reranking improves answer quality by selecting the most relevant context.

---

## 32. What is Hybrid Search?

Hybrid Search combines semantic vector search with keyword-based search.
It improves retrieval accuracy by leveraging both approaches.

---

## 33. Why use Hybrid Search?

Some queries depend on exact keywords while others rely on semantic meaning.
Hybrid Search performs well for both query types.

---

## 34. What is metadata filtering?

Metadata filtering restricts retrieval based on document attributes such as source, department, or date.
It reduces irrelevant search results.

---

## 35. Why store metadata?

Metadata enables filtering, document tracking, and incremental updates.
It also improves retrieval precision.

---

## 36. What happens if no relevant chunks are retrieved?

The application should avoid hallucinating and return an appropriate fallback response.
For example: "I couldn't find enough information to answer this question."

---

## 37. What are common RAG challenges?

Common challenges include poor chunking, weak embeddings, irrelevant retrieval, hallucinations, and stale data.
Continuous evaluation and monitoring help address these issues.

---

## 38. How do you improve retrieval quality?

Improve chunking, use better embedding models, apply reranking, tune Top-K, and use metadata filters.
Evaluate retrieval metrics regularly.

---

## 39. How do you evaluate retrieval quality?

Measure Precision@K, Recall@K, MRR, and NDCG using a benchmark dataset.
Analyze failed retrieval cases and adjust the pipeline accordingly.

---

## 40. How do you evaluate the final answer?

Evaluate correctness, faithfulness, relevance, completeness, and safety.
Use automated metrics, LLM judges, and human reviewers.

---

## 41. What is grounding?

Grounding ensures the LLM generates responses only from retrieved evidence.
It minimizes fabricated or unsupported information.

---

## 42. What is context window?

The context window is the maximum amount of input tokens an LLM can process.
It includes the prompt, retrieved documents, and conversation history.

---

## 43. What happens if the context is too large?

The model may exceed its token limit or become less focused.
Use chunk selection, reranking, or summarization to reduce context size.

---

## 44. What is prompt engineering in RAG?

Prompt engineering structures the retrieved context and user query effectively.
Clear instructions improve grounded and accurate responses.

---

## 45. How do you prevent prompt injection in RAG?

Validate user input, isolate system prompts, restrict tool access, and use output guardrails.
Never allow user input to override system instructions.

---

## 46. What is query rewriting?

Query rewriting reformulates the user's question to improve retrieval performance.
It helps retrieve better documents for ambiguous queries.

---

## 47. What is document ingestion?

Document ingestion involves loading, cleaning, chunking, embedding, and storing documents in a vector database.
It prepares knowledge for retrieval.

---

## 48. How do you handle document updates?

Re-embed only modified chunks, replace existing vectors using document IDs, and remove outdated embeddings.
This avoids reprocessing the entire document.

---

## 49. What is incremental indexing?

Incremental indexing updates only new or modified documents instead of rebuilding the entire index.
It reduces processing time and cost.

---

## 50. How do you monitor a production RAG system?

Monitor retrieval latency, LLM latency, Precision@K, error rates, token usage, hallucinations, and user feedback.
Use tools like **LangSmith**, **Prometheus**, **Grafana**, and **CloudWatch** for end-to-end observability.

## 1. What is a vector database?

A vector database stores embeddings and performs fast similarity search on high-dimensional vectors.
It is the core storage layer for semantic search in RAG applications.

## 2. Why do we need a vector database?

Traditional databases search using exact values, while vector databases search by semantic meaning.
They enable fast retrieval of relevant documents for LLMs.

## 3. Which vector databases have you worked with?

I have worked with **Qdrant** and have knowledge of **Pinecone, Milvus, and pgvector**.
Qdrant is my primary choice for RAG applications.

---

## 4. What is Qdrant?

Qdrant is an open-source vector database optimized for semantic search and filtering.
It supports metadata filtering, HNSW indexing, and REST/gRPC APIs.

---

## 5. Why did you choose Qdrant?

Qdrant provides high-performance similarity search, metadata filtering, and easy integration with LangChain.
It is lightweight, open-source, and well suited for production RAG systems.

---

## 6. What is Pinecone?

Pinecone is a fully managed cloud vector database.
It handles indexing, scaling, and infrastructure automatically.

---

## 7. What is Milvus?

Milvus is an open-source vector database designed for large-scale AI workloads.
It supports billions of vectors with distributed deployment.

---

## 8. What is pgvector?

pgvector is a PostgreSQL extension that enables vector storage and similarity search.
It is ideal when relational and vector data need to coexist.

---

## 9. When would you choose Qdrant over Pinecone?

Choose Qdrant for self-hosted deployments and full infrastructure control.
Choose Pinecone when you want a managed cloud service with minimal operational overhead.

---

## 10. What data is stored in a vector database?

A vector database stores embeddings along with metadata and unique IDs.
The original document may also be stored depending on the design.

---

## 11. What metadata do you store?

Typical metadata includes document ID, page number, source, filename, chunk ID, and timestamps.
Metadata improves filtering and document management.

---

## 12. How would you insert an embedding into Qdrant?

Generate an embedding, assign a unique point ID, attach metadata, and upsert it into a Qdrant collection.
The vector and payload are stored together for retrieval.

---

## 13. What is a collection in Qdrant?

A collection is similar to a table in a relational database.
It stores vectors with the same embedding dimensions and configuration.

---

## 14. What is a point in Qdrant?

A point is a single record containing an ID, vector embedding, and metadata payload.
Each chunk typically corresponds to one point.

---

## 15. What is payload in Qdrant?

Payload refers to metadata associated with a vector.
It enables filtering without affecting similarity search.

---

## 16. How do you search in Qdrant?

Convert the query into an embedding and perform a similarity search.
Qdrant returns the most similar vectors along with their metadata.

---

## 17. What is metadata filtering?

Metadata filtering limits retrieval based on attributes such as department, document type, or source.
It improves retrieval accuracy by excluding irrelevant documents.

---

## 18. How do you update vectors?

Generate a new embedding and upsert it using the same point ID.
The existing vector is replaced automatically.

---

## 19. How do you delete vectors?

Delete vectors using their point ID or metadata filters.
This removes outdated or unnecessary embeddings.

---

## 20. How would you re-ingest only one page of a PDF?

Identify the affected page, regenerate its chunks and embeddings, then update only those vectors.
This avoids rebuilding the entire vector database.

---

## 21. How would you replace an existing vector?

Retrieve the point ID using metadata and perform an upsert with the new embedding.
The previous vector is overwritten.

---

## 22. Why assign unique IDs to vectors?

Unique IDs allow updates, deletions, and incremental indexing.
They prevent duplicate embeddings.

---

## 23. What is HNSW?

HNSW (Hierarchical Navigable Small World) is an approximate nearest neighbor indexing algorithm.
It enables very fast vector similarity searches.

---

## 24. What is ANN search?

Approximate Nearest Neighbor (ANN) search finds highly similar vectors quickly.
It trades a small amount of accuracy for much better performance.

---

## 25. What is chunking?

Chunking is the process of splitting large documents into smaller pieces before embedding.
It improves retrieval accuracy and fits within the LLM's context window.

---

## 26. Why is chunking required?

LLMs have token limits and embeddings work better with focused text.
Chunking improves semantic search quality.

---

## 27. What is chunk overlap?

Chunk overlap repeats a small portion of text between adjacent chunks.
It preserves context across chunk boundaries.

---

## 28. Why do we use chunk overlap?

Overlap prevents important information from being split across chunks.
It improves retrieval accuracy for boundary-spanning content.

---

## 29. How do you decide chunk overlap?

The overlap depends on document structure and chunk size, typically 10–20%.
More overlap improves context but increases storage and cost.

---

## 30. What happens if there is no overlap?

Important context may be split between chunks, leading to incomplete retrieval.
This can reduce answer quality.

---

## 31. What chunking strategies have you used?

I have used recursive chunking, custom JSON chunking, page-based chunking, paragraph-based chunking, and semantic chunking.
The strategy depends on document type and use case.

---

## 32. What is recursive chunking?

Recursive chunking splits text using separators such as paragraphs, sentences, and words.
It preserves natural document structure.

---

## 33. What is custom chunking?

Custom chunking uses application-specific rules instead of generic text splitting.
For example, splitting a JSON array into individual objects.

---

## 34. Why did you use custom chunking for JSON?

Each JSON object represented a complete business entity such as a location or qualification.
Keeping one object per chunk preserved semantic integrity.

---

## 35. What is page-based chunking?

Each PDF page becomes one or more chunks while preserving page information.
It is useful for page-specific retrieval and citations.

---

## 36. What is paragraph-based chunking?

Documents are split at paragraph boundaries.
This keeps related information together.

---

## 37. What is semantic chunking?

Semantic chunking groups text based on meaning rather than fixed size.
It avoids splitting logically connected content.

---

## 38. How does semantic chunking work?

It computes embeddings for sentences and groups semantically similar sentences together.
Similarity thresholds determine chunk boundaries.

---

## 39. Which libraries can be used for semantic chunking?

SentenceTransformers, LangChain SemanticChunker, Hugging Face, and spaCy are commonly used.
Embeddings determine semantic similarity.

---

## 40. How do you measure semantic similarity?

Generate embeddings and compute cosine similarity.
Higher similarity scores indicate closer semantic meaning.

---

## 41. Which chunking strategy would you use for JSON?

Custom chunking is preferred because each JSON object represents a complete entity.
Splitting within an object can break its context.

---

## 42. Which chunking strategy would you use for PDFs with text and tables?

Use layout-aware or document-aware chunking that preserves tables and surrounding text.
This maintains the relationship between tabular and textual information.

---

## 43. Which chunking strategy would you use for FAQs?

Use one question-answer pair as a chunk.
Each chunk remains self-contained and easy to retrieve.

---

## 44. What chunk size do you typically use?

A common range is **300–800 tokens**, depending on the embedding model and document type.
The goal is to balance context and retrieval precision.

---

## 45. Can chunk size affect retrieval?

Yes. Very small chunks lose context, while very large chunks introduce irrelevant information.
An optimal chunk size improves retrieval quality.

---

## 46. How do you evaluate chunk quality?

Measure retrieval metrics and manually inspect retrieved chunks.
Poor retrieval often indicates inappropriate chunk boundaries or size.

---

## 47. What happens if duplicate chunks are indexed?

Duplicate chunks increase storage, retrieval noise, and token cost.
Deduplication should be performed during ingestion.

---

## 48. How do you handle document updates?

Re-embed only modified chunks and replace their vectors using document metadata.
This supports efficient incremental indexing.

---

## 49. How do you optimize vector search performance?

Use ANN indexes such as HNSW, metadata filtering, optimized embedding models, and appropriate Top-K values.
These improve latency while maintaining retrieval quality.

---

## 50. How do you monitor a vector database in production?

Monitor search latency, indexing time, collection size, memory usage, query throughput, and retrieval accuracy.
Use Prometheus, Grafana, CloudWatch, and application logs to detect performance or availability issues.

## 1. What is prompt injection?

Prompt injection is an attack where a user tries to manipulate an LLM into ignoring its intended instructions.
It can lead to unauthorized actions or sensitive information disclosure.

---

## 2. How can prompt injection change LLM behavior?

It attempts to override the system prompt and force the model to follow malicious instructions.
This may cause unsafe outputs or unauthorized tool execution.

---

## 3. How can prompt injection expose confidential information?

Attackers may trick the LLM into revealing system prompts, retrieved documents, or sensitive data.
Without guardrails, the model may disclose information it should not.

---

## 4. How do you protect against prompt injection?

Use strong system prompts, input validation, RBAC, output filtering, tool validation, and secure RAG retrieval.
Never allow user prompts to override system instructions.

---

## 5. What are guardrails?

Guardrails are security controls that ensure LLMs generate safe, compliant, and authorized responses.
They validate inputs, outputs, and tool usage before executing actions.

Use authentication, RBAC, SQL validation, parameterized queries, read-only users, and output filtering.
Allow only authorized and validated database operations.

## 7. How do you prevent unauthorized SQL execution?

Allow only approved SQL operations, validate generated queries, and use read-only database accounts.
Sensitive write operations should require human approval.

Grant access only to required tables and operations.
