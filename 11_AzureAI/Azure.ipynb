# Azure AI vs AWS Bedrock Mapping

| Azure | AWS Equivalent | Purpose |
|--------|----------------|----------|

# 1. Azure OpenAI ⭐⭐⭐⭐⭐

### What is it?

Azure OpenAI is Microsoft's managed service for accessing OpenAI models.

Instead of calling OpenAI directly:

```text
Application
      ↓
Azure OpenAI
      ↓
GPT-4 / GPT-4o
      ↓
Response
```

Think of it as:

> **Azure OpenAI = AWS Bedrock for OpenAI models**

---

## Interview Answer

> Azure OpenAI is an Azure-managed service that provides enterprise access to OpenAI models like GPT-4 and embedding models. It integrates with Azure security, networking, identity, and monitoring services.

---

## Azure OpenAI vs Bedrock

| Azure OpenAI | Bedrock |
|--------------|----------|
| GPT-4 | Claude |
| GPT-4o | Llama |
| Embedding Models | Titan Embeddings |
| Azure Managed | AWS Managed |

The architecture is nearly identical.

---

# 2. Azure AI Search ⭐⭐⭐⭐⭐

This is probably the most important Azure service after Azure OpenAI.

## What is it?

Azure AI Search is Microsoft's managed search service.

It supports:

- Keyword Search
- Vector Search
- Hybrid Search
- Semantic Search
- Metadata Filtering

---

### RAG Architecture

```text
PDF

↓

Chunking

↓

Embeddings

↓

Azure AI Search

↓

Top-K Documents

↓

Azure OpenAI

↓

Answer
```

Exactly like:

```text
PDF

↓

Chunking

↓

Embeddings

↓

Qdrant

↓

Top-K

↓

Claude

↓

Answer
```

---

## Interview Answer

> Azure AI Search stores embeddings and supports vector, keyword, and hybrid search. In a RAG system, it retrieves the most relevant documents before sending them to the LLM.

---

## Azure AI Search vs Qdrant

| Azure AI Search | Qdrant |
|-----------------|---------|
| Managed Azure service | Open-source vector database |
| Hybrid Search | Vector-first (keyword search can be combined externally) |
| Semantic Ranking | No built-in semantic ranking |
| Azure Integration | Excellent |
| Enterprise Ready | Yes |

# 4. Managed Identity ⭐⭐⭐⭐⭐

This is a very common interview topic.

## Problem

Suppose your application calls Azure OpenAI.

Without Managed Identity:

```python
API_KEY = "xxxxxxxxxxxxxxxx"
```

Problems:

- Secret stored in code
- Can leak
- Hard to rotate

---

## Managed Identity

Azure automatically gives your application an identity.

Example:

```text
FastAPI

↓

Managed Identity

↓

Azure OpenAI
```

No API key needed.

---

### AWS Equivalent

Exactly like:

**IAM Role**

```text
EC2

↓

IAM Role

↓

Bedrock
```

No Access Key.

No Secret Key.

---

## Interview Answer

> Managed Identity allows Azure resources to authenticate with other Azure services securely without storing credentials. It is similar to IAM Roles in AWS.
