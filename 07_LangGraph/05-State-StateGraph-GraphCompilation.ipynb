# Part 1: State ⭐⭐⭐⭐⭐

# 1. What is State?

**State** is the shared data object that flows through every node in a LangGraph.

Every node:

- Reads the state
- Updates the state
- Returns the updated state

---

# Interview Definition ⭐⭐⭐⭐⭐

> **State is the shared application data that is passed between nodes during graph execution. It represents the current context of the workflow.**

---

# 2. Why Do We Need State?

Imagine this workflow

```text
Retrieve Documents

↓

LLM

↓

Generate Answer
```

Question:

How does the LLM know which documents were retrieved?

Answer:

Through **State**.

# 24. Interview Questions

## Q1. What is State?

**Answer:**

State is the shared application data passed between all nodes during graph execution. Every node reads, updates, and returns the state.

---

## Q2. Why is State important?

**Answer:**

State enables nodes to share information without relying on global variables, making workflows modular and maintainable.

---

## Q3. What is StateGraph?

**Answer:**

StateGraph is the graph builder that defines the workflow by combining a shared state, nodes, and edges into a graph structure.

---

## Q4. What does compile() do?

**Answer:**

compile() validates the graph definition and converts it into an executable graph that can be invoked.

---

## Q5. Can you invoke a StateGraph directly?

**Answer:**

No. A StateGraph is only a builder. You must first call `compile()` to create an executable graph object.

---

## Q6. What is the difference between StateGraph and Graph?

| StateGraph | Graph |
|------------|-------|
| Builder | Executable |
| Define workflow | Run workflow |
| Add nodes & edges | invoke(), stream() |
| Needs compile() | Returned by compile() |

# 26. Real Interview Scenario

**Question:**

> Walk me through the lifecycle of a LangGraph application.

**Answer:**

First, I define the shared state using a TypedDict. Then I create a StateGraph with that state type, register nodes, and connect them using edges. Once the workflow is complete, I call `compile()` to validate and build the executable graph. Finally, I execute the graph using `invoke()` or `stream()`, allowing each node to read and update the shared state until the workflow reaches the END node.

---

# 30-Second Interview Answer

> **State is the shared data that flows through every node in a LangGraph. StateGraph is the builder used to define nodes, edges, and the shared state. Once the graph is defined, `compile()` validates the workflow and creates an executable graph, which is then executed using methods like `invoke()` or `stream()`. This separation between definition and execution makes LangGraph modular, type-safe, and production-ready.**
