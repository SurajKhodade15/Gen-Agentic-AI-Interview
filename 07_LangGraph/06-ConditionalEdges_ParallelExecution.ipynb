# Part 1: Conditional Edges ⭐⭐⭐⭐⭐

# 1. What are Conditional Edges?

A **Conditional Edge** allows LangGraph to dynamically decide **which node should execute next** based on the current state.

Instead of always following a fixed path, the graph evaluates a condition and routes execution accordingly.

---

# Interview Definition ⭐⭐⭐⭐⭐

> **A Conditional Edge is a routing mechanism that determines the next node at runtime by evaluating the current state.**

---

# 2. Why Do We Need Conditional Edges?

Without Conditional Edges

```text
Question

↓

LLM

↓

END
```

Same path every time.

---

With Conditional Edges

```text
Question

↓

Need Tool?

│

├── Yes

│      ↓

│   Tool

│      ↓

│     LLM

│

└── No

       ↓

      LLM
```

Execution depends on the state.

# Part 2: Parallel Execution ⭐⭐⭐⭐

# 11. What is Parallel Execution?

Parallel Execution allows **multiple nodes to execute simultaneously** instead of one after another.

---

# Interview Definition ⭐⭐⭐⭐

> **Parallel Execution is the ability to run multiple independent nodes concurrently and combine their results later in the workflow.**

---

# 12. Sequential vs Parallel

Sequential

```text
Retrieve Docs

↓

Search Web

↓

LLM
```

Total Time

```
2s + 3s = 5s
```

---

Parallel

```text
        START

           │

     ┌─────┴─────┐

     ▼           ▼

Retrieve     Search Web

     │           │

     └─────┬─────┘

           ▼

         Combine

           ▼

          LLM
```

Total Time

```
≈ 3 seconds
```

The two independent tasks execute together.

---

# 13. When Can We Use Parallel Execution?

When tasks are independent.

Examples

- Search Vector DB
- Search Web
- Query SQL
- Call Weather API

These don't depend on each other.

---

# 14. When NOT to Use Parallel Execution?

Bad Example

```text
Retrieve Docs

↓

Summarize Docs
```

Summarization depends on retrieval.

Cannot run simultaneously.

# 17. Conditional Edge vs Parallel Execution

| Conditional Edge | Parallel Execution |
|------------------|-------------------|
| One path is selected | Multiple paths execute |
| Decision based | Concurrent execution |
| Branching | Fan-out |
| Router | Independent tasks |

# 18. Interview Questions

## Q1. What is a Conditional Edge?

**Answer:**

A Conditional Edge evaluates the current state and dynamically determines which node should execute next.

---

## Q2. What is Parallel Execution?

**Answer:**

Parallel Execution allows multiple independent nodes to run simultaneously, reducing overall execution time.

---

## Q3. When should you use Conditional Edges?

**Answer:**

When the next execution path depends on runtime information, such as user intent, tool availability, or intermediate results.

---

## Q4. When should you use Parallel Execution?

**Answer:**

When multiple tasks are independent and can execute concurrently without waiting for each other.

---

## Q5. Can Conditional Edges and Parallel Execution be used together?

**Answer:**

Yes. A Conditional Edge can first decide the execution path, and one of those paths may launch multiple nodes in parallel before merging the results.

# 20. Real Interview Scenario

**Question:**

> You're building a financial research agent. It should retrieve internal reports, fetch live stock prices, and query market news before generating an investment summary. How would you model this in LangGraph?

**Answer:**

I would use **Parallel Execution** to retrieve internal reports, fetch stock prices, and query market news simultaneously because these tasks are independent. Once all results are available, a merge node would combine them into the shared state, and the final LLM node would generate the investment summary. If routing depends on the user's request—for example, choosing between stock analysis and mutual fund analysis—I would use a **Conditional Edge** before launching the parallel tasks.

# 30-Second Interview Answer

> **Conditional Edges enable dynamic routing by selecting the next node based on the current state, while Parallel Execution allows multiple independent nodes to run concurrently. Conditional Edges are used for decision-making, whereas Parallel Execution improves performance by reducing latency. Together, they make LangGraph highly suitable for production-grade AI workflows.**

---

# Key Takeaway

> **Conditional Edges answer the question *"Which path should I take?"* while Parallel Execution answers *"Which tasks can I perform at the same time?"*. These two capabilities make LangGraph significantly more powerful than traditional sequential workflows.**
