# 1. What are Loops & Cycles?

A **Loop (Cycle)** is a workflow where execution returns to a previously executed node until a condition is satisfied.

Unlike traditional workflows that execute only once from START to END, LangGraph allows nodes to execute multiple times.

---

# Interview Definition ⭐⭐⭐⭐⭐

> **A Loop (Cycle) is a graph execution pattern where nodes are revisited until a termination condition is met. It enables iterative reasoning and multi-step decision making.**

---

# 2. Why Do We Need Loops?

Many AI tasks require multiple iterations.

Examples:

- Agent reasoning
- Tool calling
- Self-correction
- Planning
- Reflection
- Multi-step problem solving

Without loops, the agent gets only **one chance** to solve the problem.

---

# 3. Workflow Without Loop

```text
START

↓

LLM

↓

END
```

One execution.

---

# 4. Workflow With Loop

```text
START

↓

Reason

↓

Need Tool?

│

├── Yes

│

▼

Tool

↓

Reason

│

└─────────────┐
              │
              ▼

No More Tools

↓

END
```

The agent keeps reasoning until it has enough information.

---

# 5. Why LangChain Chains Cannot Easily Do This?

Traditional Chain

```text
Prompt

↓

LLM

↓

Output
```

No built-in support for revisiting previous steps.

---

LangGraph

```text
Reason

↓

Tool

↓

Reason

↓

Tool

↓

END
```

Supports cycles naturally.

# 7. Small Coding Example

A routing function decides whether to continue looping.

```python
def route(state):

    if state["tool_required"]:
        return "tool"

    return END
```

Loop edge

```python
builder.add_conditional_edges(
    "reason",
    route
)

builder.add_edge(
    "tool",
    "reason"
)
```

Here,

```
Tool

↓

Reason

↓

Tool

↓

Reason
```

creates the cycle.

# 11. Loop Termination

Every loop **must have an exit condition**.

Example

```python
if state["finished"]:
    return END
```

Without this,

```
Infinite Loop
```

# 16. Interview Questions

## Q1. What is a Loop in LangGraph?

**Answer:**

A loop is a cycle where execution returns to a previously executed node until a stopping condition is met.

---

## Q2. Why are loops important for AI agents?

**Answer:**

Agents often need multiple reasoning steps, tool calls, or self-corrections before producing a final answer. Loops make this iterative process possible.

---

## Q3. How is a loop created in LangGraph?

**Answer:**

By connecting a node back to a previously executed node using edges, typically combined with a conditional edge to determine when to stop.

---

## Q4. How do you prevent infinite loops?

**Answer:**

Define a clear termination condition, such as task completion, maximum iterations, or a successful validation.

---

## Q5. Where are loops commonly used?

**Answer:**

Tool calling, ReAct agents, planning, reflection, multi-agent workflows, and human-in-the-loop applications.

# 18. Real Interview Scenario

**Question:**

> You are building a research agent that searches documents until it finds sufficient information. How would you model this in LangGraph?

**Answer:**

I would create a reasoning node that evaluates whether enough information has been collected. If additional information is required, a conditional edge routes execution to a retrieval node. After retrieval, the graph loops back to the reasoning node. Once the state indicates sufficient context, the graph exits the loop and generates the final response.
