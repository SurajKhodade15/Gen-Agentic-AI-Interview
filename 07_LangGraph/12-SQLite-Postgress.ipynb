# 1. What are SQLite/Postgres Checkpointers?

SQLite and PostgreSQL Checkpointers are **persistent implementations** of LangGraph Checkpointers.

Unlike `MemorySaver`, they store checkpoints in a database, allowing workflows to survive application restarts.

---

# Interview Definition ⭐⭐⭐⭐⭐

> **SQLite and PostgreSQL Checkpointers persist LangGraph execution state in a database, enabling durable workflow recovery, long-running agents, and production deployments.**

---

# 2. Why Do We Need Persistent Checkpointers?

Imagine a workflow that runs for **2 hours**.

```text
START

↓

Retrieve Data

↓

Analyze

↓

Generate Report

↓

Email Report
```

If the server crashes after 90 minutes:

With MemorySaver

```text
Crash

↓

Restart

↓

All Progress Lost ❌
```

With SQLite/Postgres

```text
Crash

↓

Read Database

↓

Resume from Last Checkpoint ✅
```

# 4. SQLite Checkpointer

SQLite stores checkpoints in a local database file.

Suitable for

- Local Development
- Single Server
- Small Applications

Architecture

```text
Graph

↓

SQLite Database

↓

checkpoint.db
```

---

# 5. PostgreSQL Checkpointer

PostgreSQL stores checkpoints in a centralized database.

Suitable for

- Enterprise Systems
- Cloud Applications
- Multi-Server Deployments
- Production Workflows

Architecture

```text
Multiple Servers

      │

      ▼

LangGraph

      │

      ▼

 PostgreSQL

      │

      ▼

 Shared Checkpoints
```

# 6. Small Coding Example (SQLite)

```python
from langgraph.checkpoint.sqlite import SqliteSaver

checkpointer = SqliteSaver.from_conn_string(
    "checkpoints.db"
)

graph = builder.compile(
    checkpointer=checkpointer
)
```

> **Note:** The exact import path may vary slightly depending on the LangGraph version.

# 7. Small Coding Example (PostgreSQL)

```python
from langgraph.checkpoint.postgres import PostgresSaver

checkpointer = PostgresSaver.from_conn_string(
    DATABASE_URL
)

graph = builder.compile(
    checkpointer=checkpointer
)
```

# 14. Interview Questions

## Q1. Why use SQLite/Postgres Checkpointers?

**Answer:**

They persist workflow execution state in a database, allowing workflows to resume after crashes or application restarts.

---

## Q2. Why is PostgreSQL preferred for production?

**Answer:**

PostgreSQL supports concurrent access, high availability, scalability, and multiple application instances, making it suitable for enterprise deployments.

---

## Q3. When would you choose SQLite?

**Answer:**

For local development, desktop applications, or small deployments where a lightweight database is sufficient.

---

## Q4. Difference between MemorySaver and PostgreSQL Checkpointer?

**Answer:**

MemorySaver stores checkpoints in RAM and loses them after restart, whereas PostgreSQL stores checkpoints persistently in a database and supports production-grade recovery.

---

## Q5. What is the main advantage of persistent checkpointing?

**Answer:**

It enables long-running workflows to resume from the latest checkpoint instead of restarting from the beginning.
