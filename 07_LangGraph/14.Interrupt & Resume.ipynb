# 1. What is Interrupt?

An **Interrupt** temporarily pauses the execution of a LangGraph workflow and waits for an external input before continuing.

Unlike an application crash, this is an **intentional pause**.

---

# Interview Definition ⭐⭐⭐⭐⭐

> **Interrupt is a LangGraph feature that intentionally pauses graph execution, saves the current state, waits for external input, and later resumes execution from the same point.**

---

# 2. Why Do We Need Interrupt?

Many enterprise workflows cannot finish immediately.

Examples:

- Manager approval
- Human review
- User confirmation
- Payment verification
- OTP verification

The workflow must pause until the required action is completed.

# 6. Small Coding Example

```python
from langgraph.types import interrupt

def approval_node(state):

    approval = interrupt(
        "Manager approval required"
    )

    return {
        "approved": approval
    }
```

When execution reaches `interrupt()`, the graph pauses.

After receiving the approval value, execution resumes from the same node.

---

# 7. Resume Example

Later, the workflow resumes with:

```python
graph.invoke(
    True,
    config=config
)
```

Here:

- `True` → Approval received
- Graph continues from the interrupted node

> **Note:** The exact resume syntax can vary slightly across LangGraph versions.

# 9. Interrupt vs Checkpointer

Common interview question.

| Interrupt | Checkpointer |
|-----------|--------------|
| Pauses execution | Saves execution |
| Intentional | Automatic |
| Waits for input | Doesn't wait |
| Uses Checkpointer | Provides persistence |

Think of it like this:

- **Interrupt** = Pause button
- **Checkpointer** = Save button

Interrupt depends on checkpointing to restore execution later.

# 15. Interview Questions

## Q1. What is Interrupt in LangGraph?

**Answer:**

Interrupt pauses graph execution intentionally, saves the current state using a Checkpointer, waits for external input, and later resumes execution from the same point.

---

## Q2. Why do we need Interrupt?

**Answer:**

Many workflows require human approval or external events before continuing. Interrupt allows the workflow to pause safely without losing progress.

---

## Q3. Does Interrupt work without a Checkpointer?

**Answer:**

No. Interrupt relies on a Checkpointer to persist the workflow state so execution can resume later.

---

## Q4. Difference between Interrupt and Checkpointer?

**Answer:**

A Checkpointer automatically saves execution state, while Interrupt intentionally pauses execution. Interrupt uses the Checkpointer to restore the workflow during resume.

---

## Q5. Give a real-world example.

**Answer:**

A loan approval workflow pauses after risk analysis until a manager approves the request. Once approved, the workflow resumes from the same point and continues processing.

# 17. Real Interview Scenario

**Question:**

> You're building an AI loan processing system. Loans above ₹50 lakh require manager approval before disbursement. How would you implement this in LangGraph?

**Answer:**

After validating the application, the approval node calls `interrupt()` to pause the workflow. The current state is saved using a Checkpointer. Once the manager approves the loan, the workflow is resumed using the same thread/config, and execution continues directly to the disbursement node instead of restarting the process.

# 30-Second Interview Answer

> **Interrupt is a LangGraph feature that intentionally pauses workflow execution for external input such as human approval. It works together with a Checkpointer to persist the current state, allowing the workflow to resume from the exact point where it was interrupted. This capability is essential for long-running, human-in-the-loop, production-grade AI systems.**

# Key Takeaway

> **Checkpointer saves the workflow. Interrupt pauses the workflow. Resume continues the workflow. Together, they enable reliable long-running AI applications with human-in-the-loop capabilities.**
