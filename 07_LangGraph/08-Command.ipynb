# 1. What is Command?

Normally, a node only returns an updated **State**.

With **Command**, a node can return:

- Updated State
- Next Node to execute

In other words,

**Command allows a node to update the state and control graph execution simultaneously.**

---

# Interview Definition ⭐⭐⭐⭐⭐

> **Command is a LangGraph object that allows a node to both update the shared state and explicitly control the next execution step, such as routing to another node.**

---

# 2. Why Do We Need Command?

Without Command

```text
Node

↓

Return State

↓

Conditional Edge

↓

Next Node
```

Routing is handled separately.

---

With Command

```text
Node

↓

Update State

↓

Decide Next Node

↓

Continue Execution
```

Everything happens inside one node.

# 6. Basic Syntax

```python
return Command(

    update={...},

    goto="next_node"

)
```

Two important parts

- **update** → Update State
- **goto** → Next Node

---

# 7. Small Coding Example

```python
from langgraph.types import Command

def router(state):

    if state["approved"]:

        return Command(
            goto="payment"
        )

    return Command(
        goto="manager_review"
    )
```

Flow

```text
Approved?

│

├── Yes

│

▼

Payment

│

└── No

↓

Manager Review
```

# 8. Updating State

Command can update state while routing.

# 10. Command vs Conditional Edge

Very common interview question.

| Conditional Edge | Command |
|------------------|---------|
| Routing outside node | Routing inside node |
| Separate routing function | Node controls routing |
| Returns next node | Returns Command |
| Better for simple routing | Better for dynamic workflows |

# 13. When Should You Use Command?

Use Command when:

✅ Node decides routing

✅ Update state and route together

✅ Dynamic workflows

✅ Agent decision making

---

# 14. When Use Conditional Edge Instead?

Use Conditional Edge when:

- Routing logic should remain separate
- Multiple nodes share the same routing logic
- Simpler graph design is preferred

---

# 15. Common Mistakes

❌ Returning invalid node names.

❌ Forgetting `goto`.

❌ Updating state but not routing correctly.

❌ Using Command when simple edges are sufficient.

---

# 16. Best Practices

✅ Keep Command logic simple.

✅ Return meaningful state updates.

✅ Use descriptive node names.

✅ Prefer Conditional Edges for reusable routing.

# 17. Interview Questions

## Q1. What is Command?

**Answer:**

Command is a LangGraph object that allows a node to update the shared state and explicitly control which node executes next.

---

## Q2. Why use Command?

**Answer:**

It combines state updates and routing into a single return value, making dynamic workflows simpler and more flexible.

---

## Q3. What are the two important fields in Command?

**Answer:**

- `update` → Modifies the shared state.
- `goto` → Specifies the next node to execute.

---

## Q4. Difference between Command and Conditional Edge?

**Answer:**

Conditional Edges keep routing outside the node using a separate routing function, whereas Command allows the node itself to decide the next execution path while optionally updating the state.

---

## Q5. Can Command create loops?

**Answer:**

Yes. A node can return `Command(goto="previous_node")`, causing the graph to revisit an earlier node until a termination condition is met.

# 19. Real Interview Scenario

**Question:**

> Your AI loan approval system validates a request. If the amount is below ₹5 lakh, it should go directly to payment; otherwise, it should go to a manager for approval. How would you implement this?

**Answer:**

I would implement the routing inside the validation node using `Command`. The node would evaluate the loan amount, update the application status if required, and return `Command(goto="payment")` for smaller amounts or `Command(goto="manager_review")` for larger amounts. This keeps the decision and routing together.

# 30-Second Interview Answer

> **Command is a LangGraph mechanism that allows a node to both update the shared state and determine the next node to execute. Unlike Conditional Edges, which keep routing logic separate, Command embeds routing directly inside the node, making it useful for dynamic agent workflows and complex decision-making.**

---

# Key Takeaway

> **Conditional Edge asks, "Which node should run next?" from outside the node, while Command lets the node itself answer that question and update the state at the same time. This makes Command a powerful feature for implementing intelligent, dynamic AI workflows.**
