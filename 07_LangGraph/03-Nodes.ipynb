# 1. What is a Node?

A **Node** is the smallest executable unit in a LangGraph.

A node performs a specific task, such as:

- Calling an LLM
- Calling a Tool
- Querying a Database
- Searching Documents
- Processing Data
- Updating State

Think of a node as a **Python function** that receives the current state, performs some work, and returns an updated state.

---

# Interview Definition ⭐⭐⭐⭐⭐

> **A Node is an executable function in LangGraph that receives the current state, performs a task, optionally updates the state, and returns the updated state to the graph.**

---

# 2. Why Do We Need Nodes?

Instead of writing one large function:

```python
process_everything()
```

Break the workflow into smaller reusable functions.

Example

```text
Validate User

↓

Retrieve Documents

↓

Call LLM

↓

Generate Answer
```

Each step becomes a node.

---

# 3. Node Architecture

```text
          Current State

                │

                ▼

             Node

      (Business Logic)

                │

                ▼

          Updated State
```

A node never decides **where** to go next.

It only performs work.

---

# 4. Node Lifecycle

Every node follows the same lifecycle.

```text
Receive State

↓

Read Values

↓

Execute Logic

↓

Update State

↓

Return State
```

The graph runtime handles the next step.

# 14. Node Execution Order

Nodes do **not** decide execution order.

The graph controls execution.

```text
Node

↓

Edge

↓

Next Node
```

# 15. Node vs Function

Common interview question.

| Python Function | LangGraph Node |
|-----------------|----------------|
| Any function | Function registered in graph |
| No shared workflow | Works inside graph |
| Independent | Reads & updates state |
| Manual execution | Graph runtime execution |

Every node is usually a Python function, but not every Python function is part of a LangGraph.

# 16. Best Practices

✅ One responsibility per node.

✅ Keep nodes small.

✅ Read from state.

✅ Return updated state.

✅ Avoid side effects unless necessary.

---

# 17. Common Mistakes

❌ Storing data inside global variables.

❌ Making one node perform every task.

❌ Modifying state without returning it.

❌ Putting routing logic inside nodes.

# 18. Interview Questions

## Q1. What is a Node in LangGraph?

**Answer:**

A Node is an executable function that receives the current state, performs a task, updates the state if necessary, and returns the updated state.

---

## Q2. Are Nodes stateful?

**Answer:**

No. Nodes are stateless functions. The shared State stores application data, while nodes simply read and update it.

---

## Q3. Can a Node call an LLM?

**Answer:**

Yes. A node can invoke an LLM, call tools, query databases, perform retrieval, or execute any Python logic.

---

## Q4. Who decides which node executes next?

**Answer:**

The graph runtime determines the next node based on the configured edges. Nodes only perform work and return state.

---

## Q5. Why should nodes have a single responsibility?

**Answer:**

Small, focused nodes are easier to test, debug, reuse, and maintain. They also make workflows more modular and easier to extend.

# 20. Real Interview Scenario

**Question:**

> You are building an AI customer support workflow. The process includes user validation, document retrieval, answer generation, and response formatting. How would you design the nodes?

**Answer:**

I would create separate nodes for each responsibility:
1. **Validation Node** to verify the user.
2. **Retriever Node** to fetch relevant support documents.
3. **LLM Node** to generate the response using the retrieved context.
4. **Formatter Node** to produce the final response format.

Each node would receive the shared state, perform a single task, update the state, and return it to the graph.

# 30-Second Interview Answer

> **A Node is the fundamental execution unit in LangGraph. It is typically a Python function that receives the current shared state, performs a specific task such as calling an LLM or a tool, updates the state, and returns it. Nodes are stateless, reusable, and follow the single-responsibility principle, while the graph controls the execution flow through edges.**

# Key Takeaway

> **A Node defines *what* work is performed. It does not decide *where* execution goes next—that responsibility belongs to the graph through edges.**
