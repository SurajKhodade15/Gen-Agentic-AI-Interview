# 1. What is Send API?

**Send** is a LangGraph feature used to execute **multiple instances of the same node** with different inputs.

Think of it as a **fan-out** mechanism.

Instead of executing one node once,

LangGraph executes the same node multiple times.

---

# Interview Definition ⭐⭐⭐⭐⭐

> **Send is a LangGraph mechanism used to dynamically create parallel executions of the same node by sending different state values to each execution.**

---

# 2. Why Do We Need Send?

Suppose a user uploads

```
100 PDF files
```

Should we process them

```text
PDF1

↓

PDF2

↓

PDF3

↓

...
```

No.

Instead

```text
           START

              │

     ┌────────┼─────────┐

     ▼        ▼         ▼

   PDF1     PDF2      PDF3

     │        │         │

     └────────┼─────────┘

              ▼

        Merge Results
```

This is where Send is useful.

# 3. Parallel Execution vs Send

Parallel Execution

```text
START

├── Search PDF

├── Search SQL

└── Search Web
```

Different nodes.

---

Send

```text
START

↓

Process PDF

↓

Process PDF

↓

Process PDF
```

Same node.

Different inputs.

---

# 4. How Send Works

```text
State

↓

Create Multiple Send Objects

↓

Same Node

↓

Runs Multiple Times

↓

Merge Results
```

# 6. Small Coding Example

Input

```python
state = {
    "files": [
        "a.pdf",
        "b.pdf",
        "c.pdf"
    ]
}
```

Create Send objects

```python
from langgraph.types import Send

def router(state):

    return [

        Send(
            "process_pdf",
            {"file": file}
        )

        for file in state["files"]

    ]
```

Each file is processed independently.

# 11. Send vs Command

Very common interview question.

| Command | Send |
|----------|------|
| Route to one node | Route to many executions |
| Single execution | Multiple executions |
| Dynamic routing | Fan-out processing |
| Decision making | Parallel workload distribution |

---

# 12. Send vs Conditional Edge

| Conditional Edge | Send |
|------------------|------|
| Select one path | Create multiple executions |
| Routing | Parallel distribution |
| One next node | Many instances of one node |

# 15. Best Practices

✅ Use Send for independent tasks.

✅ Keep processing nodes stateless.

✅ Merge results after execution.

# 16. Common Mistakes

❌ Using Send for dependent tasks.

❌ Sharing mutable state between executions.

❌ Forgetting to aggregate results.

# 17. Interview Questions

## Q1. What is Send API?

**Answer:**

Send is a LangGraph feature that creates multiple parallel executions of the same node using different state values.

---

## Q2. Why use Send?

**Answer:**

It enables scalable processing of collections such as files, documents, invoices, or resumes by distributing work across multiple executions.

---

## Q3. Difference between Send and Command?

**Answer:**

Command routes execution to a single next node, while Send creates multiple parallel executions of the same node.

---

## Q4. Can Send improve performance?

**Answer:**

Yes. Independent tasks can execute concurrently, reducing the total processing time compared to sequential execution.

---

## Q5. Give a real-world use case for Send.

**Answer:**

Processing hundreds of uploaded PDF files, where each document can be analyzed independently before combining the results.

# 19. Real Interview Scenario

**Question:**

> You receive 1,000 customer invoices and need to extract invoice details as quickly as possible. How would you implement this in LangGraph?

**Answer:**

I would use a router node that creates a `Send` object for each invoice. LangGraph would execute the same invoice-processing node in parallel for each document. Once all executions complete, the results would be merged into a single state for further processing or database storage. This approach is faster and scales much better than sequential processing.

> **Send is a LangGraph feature used for fan-out processing. It creates multiple parallel executions of the same node using different state values, making it ideal for batch processing tasks such as document analysis, invoice extraction, and resume screening. # Key Takeaway


> **Command routes one workflow, while Send distributes one workflow into many parallel executions. Use Command for decision-making and Send for scalable batch processing.**
