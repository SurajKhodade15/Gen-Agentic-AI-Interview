# Part 1: Human Approval

# 1. What is Human-in-the-Loop?

**Human-in-the-Loop (HITL)** is a workflow where an AI agent pauses execution and waits for a human to review, approve, reject, or modify the result before continuing.

---

# Interview Definition ⭐⭐⭐⭐⭐

> **Human-in-the-Loop is a workflow pattern where an AI agent collaborates with a human by pausing execution for review or approval before continuing.**

---

# 2. Why Do We Need Human Approval?

AI should **not** make every decision autonomously.

Examples requiring human approval:

- Loan Approval
- Medical Diagnosis
- Insurance Claims
- Legal Documents
- Large Financial Transactions
- HR Hiring Decisions

# Part 2: interrupt()

# 5. What is interrupt()?

`interrupt()` pauses graph execution and returns control to the application.

The graph waits until an external response is provided.

---

# Interview Definition

> **interrupt() is a LangGraph API that pauses execution, saves the current state, and waits for external input before resuming.**

---

# 6. Small Coding Example

```python
from langgraph.types import interrupt

def approval_node(state):

    approval = interrupt(
        "Approve this request?"
    )

    return {
        "approved": approval
    }
```

Execution pauses here.

# Part 3: Resume Execution

# 8. What is Resume?

Resume continues execution from the exact point where `interrupt()` paused the workflow.

It does **not** restart the graph.

---

# Interview Definition

> **Resume Execution restores the graph from the latest checkpoint and continues execution after an interrupt.**

---

# 9. Small Coding Example

Resume with approval

```python
graph.invoke(
    True,
    config=config
)
```

Resume with rejection

```python
graph.invoke(
    False,
    config=config
)
```

The value passed becomes the return value of `interrupt()`.

# 17. Interview Questions

## Q1. What is Human-in-the-Loop?

**Answer:**

Human-in-the-Loop is a workflow where AI pauses execution for human review or approval before continuing.

---

## Q2. Why is interrupt() needed?

**Answer:**

interrupt() intentionally pauses workflow execution so external input, such as manager approval, can be collected before continuing.

---

## Q3. Does Resume restart the workflow?

**Answer:**

No. Resume continues execution from the exact point where the workflow was interrupted.

---

## Q4. Can interrupt() work without a Checkpointer?

**Answer:**

No. A Checkpointer is required to save the execution state so it can be restored during resume.

---

## Q5. Give a real-world use case.

**Answer:**

A bank pauses a loan approval workflow until a manager reviews high-value applications. After approval, execution resumes from the approval step.

# 19. Real Interview Scenario

**Question:**

> You're building an AI medical diagnosis system. Before showing the diagnosis to a patient, a doctor must verify it. How would you implement this in LangGraph?

**Answer:**

The diagnosis node generates the initial diagnosis and then calls `interrupt()` to pause execution. A Checkpointer saves the current workflow state. The doctor reviews the diagnosis and either approves or modifies it. Once the review is complete, the workflow resumes from the interruption point and proceeds to deliver the verified diagnosis to the patient.
