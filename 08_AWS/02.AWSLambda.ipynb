# 1. What is AWS Lambda?

AWS Lambda is a **serverless compute service** that lets you run code **without provisioning or managing servers**.

You upload your code, and Lambda automatically executes it when triggered by an event.

---

# Interview Definition ⭐⭐⭐⭐⭐

> **AWS Lambda is a serverless compute service that executes code in response to events and automatically manages the underlying infrastructure, scaling, and availability.**

---

# 2. Why Use Lambda in GenAI?

Instead of running a backend server 24/7,

you can execute AI logic **only when a request arrives**.

Benefits:

- No server management
- Auto scaling
- Pay only for execution time
- Easy integration with AWS services

# 4. How Lambda Works

```text
Event

↓

Lambda Triggered

↓

Execute Code

↓

Return Response

↓

Lambda Stops
```

Unlike traditional servers, Lambda runs **only when needed**.

---

# 5. Lambda Lifecycle

```text
Request

↓

Initialize Runtime (Cold Start)

↓

Execute Function

↓

Return Response

↓

Environment Reused (Warm Start)
```

# 9. Calling Amazon Bedrock from Lambda

```python
import boto3

bedrock = boto3.client("bedrock-runtime")

response = bedrock.invoke_model(
    modelId="anthropic.claude-3-sonnet",
    body='{"prompt":"Explain AI"}'
)
```

Typical Flow

```text
Lambda

↓

Bedrock Runtime

↓

Claude

↓

Response
```

# 12. Lambda vs EC2

| Lambda | EC2 |
|----------|-----|
| Serverless | Virtual Machine |
| Auto Scaling | Manual Scaling |

# 15. Best Practices

✅ Keep Lambda functions small.

✅ Store secrets in AWS Secrets Manager.

✅ Reuse SDK clients outside the handler.

✅ Set appropriate memory and timeout values.

✅ Use IAM roles with least privilege.

---

# 16. Common Mistakes

❌ Creating AWS clients inside every request.

❌ Hardcoding credentials.

❌ Using Lambda for long-running batch jobs.

❌ Ignoring cold starts.

# 17. Interview Questions

### Q1. What is AWS Lambda?

**Answer**

AWS Lambda is a serverless compute service that executes code in response to events without requiring server management.

---

### Q2. Why is Lambda used in GenAI applications?

**Answer**

It provides a scalable, cost-effective backend for calling foundation models such as Amazon Bedrock without maintaining servers.

---

### Q3. How does Lambda communicate with Bedrock?

**Answer**

Using the AWS SDK (`boto3`) and the Bedrock Runtime APIs such as `InvokeModel` or `Converse`.

---

### Q4. What is a Cold Start?

**Answer**

A Cold Start occurs when AWS creates a new execution environment before running the Lambda function, adding extra startup latency.

---

### Q5. What is the difference between a Cold Start and a Warm Start?

| Cold Start | Warm Start |
|------------|------------|
| New execution environment | Reuses existing environment |
| Higher latency | Lower latency |

---

### Q6. Why is Lambda preferred over EC2 for GenAI APIs?

**Answer**

Lambda automatically scales, requires no infrastructure management, and charges only for execution time, making it ideal for event-driven AI APIs.

# 19. Real Interview Scenario

**Question:**

> You built a chatbot using Amazon Bedrock. How would you expose it as a REST API?

**Answer:**

I would create a REST endpoint using **API Gateway**, which triggers an **AWS Lambda** function. The Lambda function validates the request, invokes the appropriate Bedrock model using the AWS SDK, processes the response, and returns it to the client. This architecture is serverless, scalable, and cost-efficient.
