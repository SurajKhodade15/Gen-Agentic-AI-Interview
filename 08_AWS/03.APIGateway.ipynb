# 1. What is Amazon API Gateway?

Amazon API Gateway is a **fully managed service** used to **create, publish, secure, monitor, and manage APIs**.

It acts as the **entry point** for your backend services.

---

# Interview Definition ⭐⭐⭐⭐⭐

> **Amazon API Gateway is a managed service that enables developers to create, secure, publish, and monitor REST, HTTP, and WebSocket APIs while routing client requests to backend services such as AWS Lambda.**

---

# 20. Interview Questions

### Q1. What is API Gateway?

**Answer**

API Gateway is a managed AWS service that exposes, secures, and manages APIs while routing requests to backend services such as Lambda.

---

### Q2. Why is API Gateway used with Lambda?

**Answer**

API Gateway provides a secure HTTP endpoint, authentication, request validation, throttling, and monitoring before invoking Lambda.

---

### Q3. Can API Gateway call Bedrock directly?

**Answer**

Typically, **No**.

The common production architecture is:

```
API Gateway

↓

Lambda

↓

Bedrock
```

Lambda performs validation, business logic, logging, and Bedrock API calls.

---

### Q4. What authentication methods does API Gateway support?

**Answer**

- IAM
- JWT
- Cognito
- API Keys
- Lambda Authorizers

---

### Q5. Why is Rate Limiting important?

**Answer**

It prevents abuse, protects backend services from excessive traffic, and helps control operational costs.

---

### Q6. What is the difference between REST API and HTTP API?

| REST API | HTTP API |
|-----------|----------|
| More features | Simpler & faster |
| Higher cost | Lower cost |
| Enterprise integrations | Most modern serverless APIs |

# 22. Real Interview Scenario

**Question:**

> How would you expose a Bedrock-powered chatbot to mobile and web clients?

**Answer:**

I would create an HTTP API using **Amazon API Gateway**. Incoming requests would be authenticated and validated by API Gateway, which would invoke an **AWS Lambda** function. The Lambda function would call the **Amazon Bedrock Converse API**, process the model response, and return it to the client. This architecture is secure, serverless, scalable, and easy to monitor with CloudWatch.

---

# 23. 30-Second Interview Answer

> **Amazon API Gateway is a fully managed service that exposes secure HTTP endpoints for applications. In a GenAI architecture, it acts as the entry point for client requests, providing authentication, request validation, throttling, and monitoring before invoking AWS Lambda. The Lambda function then communicates with Amazon Bedrock to generate AI responses, making API Gateway an essential component of scalable serverless GenAI applications.**
