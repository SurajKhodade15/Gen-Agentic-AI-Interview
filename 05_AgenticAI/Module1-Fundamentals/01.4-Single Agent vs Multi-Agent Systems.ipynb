# 1. What is a Single Agent?

A **Single Agent** is one AI Agent responsible for completing the entire task.

It performs:

- Reasoning
- Planning
- Tool Calling
- Decision Making
- Response Generation

using a single agent.

---

## Architecture

```text
User → Single Agent → Tools → Final Response
```

---

## Example

User:

```
Summarize this PDF.
```

Flow:

```text
User → Agent → Read PDF → Summarize → Return Response
```

One agent performs everything.

---

# 2. What is a Multi-Agent System?

A **Multi-Agent System (MAS)** consists of multiple AI Agents, each with a specialized role.

Each agent focuses on a specific task and collaborates with other agents.

---

## Architecture

```text
User → Supervisor Agent → Research Agent → Analysis Agent → Report Agent → User
```

Instead of one large agent, work is divided among specialized agents.

---

# 3. Why Do We Need Multiple Agents?

Imagine building a company.

Would one employee handle:

- Sales
- HR
- Finance
- Marketing
- Development

No.

Instead, specialists handle different responsibilities.

AI Agents follow the same principle.

# 20. Interview Questions

## Q1. What is the difference between a Single Agent and a Multi-Agent System?

**Answer:**

A Single Agent performs all reasoning, planning, and execution itself. A Multi-Agent System distributes responsibilities across multiple specialized agents that collaborate to complete the task.

---

## Q2. Why use Multi-Agent Systems?

**Answer:**

They improve scalability, specialization, maintainability, and allow parallel execution of complex workflows.

---

## Q3. Does every application need multiple agents?

**Answer:**

No. Many applications such as chatbots, document summarization, and simple RAG systems work effectively with a Single Agent. Multi-Agent Systems are more suitable for complex enterprise workflows.

---

## Q4. Which frameworks support Multi-Agent Systems?

**Answer:**

CrewAI, AutoGen, LangGraph, OpenAI Agents SDK, and LlamaIndex all support building Multi-Agent applications.

---

## Q5. What is the role of a Supervisor Agent?

**Answer:**

The Supervisor Agent coordinates the workflow by assigning tasks to specialized agents, collecting their outputs, and producing the final response.

---

# 21. Common Interview Mistakes

❌ **"Multi-Agent is always better."**

Wrong.

For simple applications, a Single Agent is usually faster, cheaper, and easier to maintain.

---

❌ **"Each agent should have the same responsibility."**

Wrong.

Each agent should have a **specialized role**.

---

❌ **"Agents communicate directly with users."**

Usually the Supervisor (or Orchestrator) communicates with the user, while worker agents communicate internally.

---

# Interview Summary (30-Second Answer)

> **A Single Agent handles the complete workflow itself, making it suitable for simple tasks such as chatbots or document summarization. A Multi-Agent System consists of specialized agents coordinated by a supervisor, making it ideal for complex enterprise workflows that benefit from parallel execution, modularity, and specialization.**

---

# Key Takeaway

> **Single Agent = One expert doing everything.**
>
> **Multi-Agent = Multiple specialists collaborating under a coordinator.**
