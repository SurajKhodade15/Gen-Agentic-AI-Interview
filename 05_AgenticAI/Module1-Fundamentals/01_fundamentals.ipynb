# 1. Definition

**Agentic AI** is an AI system capable of **autonomously** achieving a goal by:

- Understanding the objective
- Planning the required steps
- Reasoning about what to do next
- Using external tools
- Observing results
- Taking the next action
- Repeating the process until the goal is completed

Unlike a traditional LLM that only generates text, an AI Agent can **think, decide, and act**.

---

## Interview Definition

> Agentic AI is an AI system that autonomously plans, reasons, uses tools, makes decisions, and executes multiple steps to achieve a user-defined goal.

---

# 2. Why Agentic AI?

Traditional LLMs are excellent at generating text but have several limitations.

## Example

User asks:

> "Book the cheapest flight from Pune to Bangalore for tomorrow and email me the itinerary."

A normal LLM cannot:

- Search flight websites
- Compare ticket prices
- Book the ticket
- Send an email

It only generates text.

An AI Agent can perform all these tasks by interacting with external tools.

---

## Limitations of Traditional LLMs

- Cannot access live data
- Cannot call APIs by itself
- Cannot execute code independently
- Cannot make long-term plans
- Cannot retry failed tasks
- Cannot perform multi-step automation

---

## Agentic AI Solves These Problems

It can:

- Plan
- Reason
- Use tools
- Access APIs
- Search the web
- Read databases
- Execute code
- Remember previous interactions
- Continue until the task is complete

# 4. Characteristics of an AI Agent

## 1. Goal-Oriented

Every agent starts with a goal.

Example:

```
Book Hotel
```

Everything the agent does is directed toward completing that goal.

---

## 2. Autonomous

Minimal human intervention.

Instead of asking the user after every step:

```
Should I search?

Should I compare?

Should I book?
```

The agent decides on its own.

---

## 3. Planning

Before acting, the agent creates a plan.

Example

Goal:

```
Generate Monthly Sales Report
```

Plan:

```
Read Database

↓

Calculate Metrics

↓

Generate Charts

↓

Create PDF

↓

Email PDF
```

---

## 4. Reasoning

The agent decides the next best action.

Example:

```
Search Result Not Found

↓

Try another website

↓

Still not found

↓

Ask User
```

Reasoning helps the agent adapt.

---

## 5. Tool Usage

Agents can use external tools.

Examples:

- Calculator
- SQL Database
- Python
- Search Engine
- Email API
- Weather API
- File System

Without tools:

LLM only knows text.

With tools:

Agent performs actions.

---

## 6. Memory

Agents remember information.

Example

User:

```
Always book aisle seats.
```

Later:

```
Book my flight.
```

The agent remembers:

```
Preferred seat = Aisle
```

---

## 7. Iterative Execution

The agent keeps working until the goal is achieved.

```
Think

↓

Act

↓

Observe

↓

Think Again

↓

Act Again

↓

Finished
```

This loop is one of the defining features of Agentic AI.

# 6. Architecture Overview

```
                 User Goal
                     │
                     ▼
              +---------------+
              |     LLM       |
              | (Reasoning)   |
              +---------------+
                     │
         Decide Next Action
                     │
     ┌───────────────┼───────────────┐
     ▼               ▼               ▼
 Planner          Memory         Tool Selector
     │               │               │
     └───────────────┼───────────────┘
                     ▼
              Execute Tool
                     │
             Observe Result
                     │
         Goal Completed?
             │
      Yes ───┴──► Return Result
      No
             │
             ▼
        Continue Reasoning
```

---

## Components

| Component | Responsibility |
|------------|----------------|
| LLM | Thinks and reasons |
| Planner | Creates execution steps |
| Memory | Stores context and preferences |
| Tools | Perform real-world actions |
| Executor | Calls tools/APIs |
| Observer | Reads tool outputs |
| Controller | Decides whether to continue or stop |

# 8. Interview Questions

## Q1. What is Agentic AI?

**Answer:**

Agentic AI is an AI system that autonomously plans, reasons, uses tools, makes decisions, and executes multiple steps to achieve a goal with minimal human intervention.

---

## Q2. Why do we need Agentic AI?

**Answer:**

Traditional LLMs only generate text. Agentic AI extends LLMs with planning, memory, and tool usage, enabling them to complete complex real-world tasks such as booking flights, querying databases, or automating workflows.

---

## Q3. What is the biggest difference between an LLM and an AI Agent?

**Answer:**

An LLM predicts text based on a prompt, whereas an AI Agent uses an LLM as its reasoning engine and combines it with planning, memory, and tools to perform autonomous, multi-step task execution.

---

## Q4. Can an AI Agent work without an LLM?

**Answer:**

Yes. An agent is a decision-making entity. While modern agents often use LLMs for reasoning, rule-based systems, classical planners, or reinforcement learning policies can also serve as the decision engine.

---

## Q5. Give one real-world example of Agentic AI.

**Answer:**

A travel booking assistant that searches flights and hotels, compares prices, books reservations, generates an itinerary, and emails the user without requiring manual intervention at each step.

---

# 9. Common Interview Mistakes

❌ "Agentic AI is just ChatGPT."

> Wrong. ChatGPT is primarily an LLM-based assistant. An AI Agent adds planning, memory, tools, and autonomous execution.

---

❌ "LLM and Agent are the same."

> Wrong. An LLM is the reasoning model. An Agent is the system that orchestrates reasoning, tool use, memory, and execution.

---

❌ "Agentic AI always works without humans."

> Wrong. Many production agents include **Human-in-the-Loop (HITL)** for approvals, especially in high-risk domains.

---

# 10. Quick Revision

- Agents combine **reasoning + planning + memory + tools + execution**.
- Agents follow a loop:

```
Think
 ↓
Plan
 ↓
Act
 ↓
Observe
 ↓
Repeat
```

- Real-world applications include travel assistants, HR automation, coding assistants, finance, and customer support.

# 1. Why This Comparison is Important

One of the most common interview questions is:

Many candidates incorrectly assume they are the same.

Each layer adds more intelligence and autonomy.

# 4. What is a Workflow?

A workflow is a **predefined sequence of steps**.

The path is fixed.

```
Step 1

↓

Step 2

↓

Step 3

↓

Done
```

The workflow never decides to change its path.

Every invoice follows exactly the same process.

An AI Agent is a system that can:

# 11. Interview Questions

## Q1. What is the difference between an LLM and an AI Agent?

**Answer:**

An LLM generates text based on a prompt. An AI Agent uses an LLM as its reasoning engine and adds planning, memory, tool usage, and iterative execution to achieve goals autonomously.

---

## Q2. Is every chatbot an AI Agent?

**Answer:**

No.

Most chatbots simply respond to user queries. An AI Agent can independently plan, make decisions, use tools, and execute multi-step tasks.

---

## Q3. What is the difference between a Workflow and an AI Agent?

**Answer:**

A workflow follows a fixed sequence of predefined steps. An AI Agent dynamically decides which steps to execute based on the current situation and observed results.

---

## Q4. Can an AI Agent contain workflows?

**Answer:**

Yes.

Agents often invoke workflows for repetitive, deterministic tasks such as document processing or report generation.

---

## Q5. Which is more reliable: Workflow or AI Agent?

**Answer:**

For repetitive business processes, workflows are generally more predictable. AI Agents are more flexible and capable of handling dynamic, unstructured tasks but require additional safeguards and evaluation.
