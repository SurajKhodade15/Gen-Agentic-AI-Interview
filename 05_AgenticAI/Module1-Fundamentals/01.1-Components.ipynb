# 3. Components of an AI Agent

Most modern agent frameworks (LangGraph, CrewAI, AutoGen, OpenAI Agents SDK) are built around these core components:

1. Goal
2. LLM (Brain)
3. Planner
4. Memory
5. Tools
6. Executor
7. Observer
8. Controller

Let's understand each one.

---

# 4. Goal

Every AI Agent starts with a **goal**.

Without a goal, the agent has nothing to achieve.

Example:

```
Find the cheapest hotel in Goa.
```

Everything the agent does is driven by this goal.

---

# 5. LLM (The Brain)

The LLM is responsible for:

- Understanding language
- Reasoning
- Generating responses
- Choosing the next action

Examples:

- GPT-4.1
- GPT-5
- Claude
- Gemini
- Llama

The LLM **does not directly execute actions**.

It decides **what should happen next**.

---

### Example

User:

```
Book my flight.
```

LLM thinks:

```
First search flights.

Then compare prices.

Finally book.
```

---

# 6. Planner

The Planner breaks a complex goal into smaller tasks.

Example:

Goal:

```
Generate Sales Report
```

Planner creates:

```
Read Database

↓

Calculate Revenue

↓

Create Charts

↓

Generate PDF

↓

Email Report
```

Without a planner, the agent would not know where to start.

---

### Analogy

Project Manager

Receives:

```
Build a website.
```

Creates tasks:

- Design UI
- Backend
- Testing
- Deployment

Exactly what the planner does.

---

# 7. Memory

Memory stores useful information.

There are two main types:

## Short-Term Memory

Stores current conversation.

Example

```
User:

Book a hotel in Goa.
```

Later:

```
Make it near the beach.
```

The agent remembers:

```
Location = Goa
```

---

## Long-Term Memory

Stores information across sessions.

Example

```
Preferred Airline = Indigo

Seat = Window

Language = English
```

Next month...

User:

```
Book a flight.
```

The agent already knows preferences.

---

# 8. Tools

The LLM cannot perform real-world actions.

It needs tools.

Examples:

- Calculator
- SQL Database
- Python
- Web Search
- Email API
- Weather API
- File Reader
- RAG Retriever

---

### Example

User

```
What's today's weather?
```

LLM cannot know live weather.

Instead:

```
Weather Tool

↓

Current Weather

↓

LLM explains result
```

---

# 9. Executor

The Executor actually runs the selected tool.

Example

Planner decides:

```
Search Google
```

Executor performs:

```
GET https://search-api.com?q=...
```

Result:

```
Returned to LLM
```

Think of the Executor as the employee who carries out the task.

---

# 10. Observer

After a tool executes, someone must inspect the result.

That is the Observer's job.

Example

```
Flight Search

↓

0 Flights Found
```

Observer reports:

```
Search failed.
```

The LLM now reasons:

```
Try another airline.
```

---

# 11. Controller

The Controller coordinates everything.

Responsibilities:

- Starts execution
- Maintains state
- Calls planner
- Invokes tools
- Tracks progress
- Stops when the goal is achieved

It acts as the **orchestrator**.

---

# 12. How All Components Work Together

# 16. Interview Questions

## Q1. What is the brain of an AI Agent?

**Answer:**

The LLM acts as the reasoning engine. It understands the user's request, generates plans, selects tools, and determines the next action.

---

## Q2. Why does an AI Agent need tools?

**Answer:**

An LLM cannot directly interact with external systems. Tools allow the agent to access databases, APIs, search engines, file systems, and other real-world resources.

---

## Q3. What is the difference between Planner and Controller?

**Answer:**

The Planner decides **what tasks need to be performed**. The Controller manages **how and when those tasks are executed**, tracks state, and determines when the workflow should stop.

---

## Q4. Why is memory important?

**Answer:**

Memory enables the agent to retain context, user preferences, previous actions, and past observations, leading to more coherent and personalized behavior.

---

## Q5. Which component actually executes the tool?

**Answer:**

The Executor runs the selected tool or API. The LLM only decides which tool should be used; it does not execute the action itself.

---

# 17. Common Interview Mistakes

❌ **"The LLM executes APIs."**

Wrong.

The LLM selects the tool. The Executor performs the API call.

---

❌ **"Planner and Controller are the same."**

Wrong.

Planner creates the execution plan.

Controller orchestrates the execution.

---

❌ **"Memory is just chat history."**

Wrong.

Memory can include conversation history, user preferences, retrieved documents, execution results, and long-term knowledge.

# Interview Summary (30-Second Answer)

> **An AI Agent consists of a Goal, an LLM for reasoning, a Planner for task decomposition, Memory for storing context, Tools for interacting with external systems, an Executor to invoke those tools, an Observer to analyze results, and a Controller that orchestrates the entire execution loop until the goal is achieved.**

---

# Key Takeaway

> **The LLM is the brain, but an AI Agent is the complete system that combines reasoning, planning, memory, tools, execution, and orchestration to solve real-world tasks.**
