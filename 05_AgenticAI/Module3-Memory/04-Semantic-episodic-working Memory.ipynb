> **Interview Difficulty:** ⭐⭐⭐⭐⭐ (Very Frequently Asked)

> Modern AI Agents don't use just one type of memory. Like humans, they use different memories for **current tasks, factual knowledge, and past experiences**.

# 3. Working Memory

Working Memory stores information required to complete the **current task**.

---

## Interview Definition

> Working Memory is temporary memory that stores the information needed to complete the current conversation or workflow.

---

Example

```text
User

Book a flight

↓

Destination = Delhi

↓

Date = 15 Aug

↓

Passengers = 2
```

Once booking is complete,

the memory is cleared.

---

### Working Memory Flow

```text
User
→ Current Task
→ Temporary State
→ Response
→ Clear
```

---

# 4. Episodic Memory

Episodic Memory stores **past experiences**.

Instead of remembering facts,

it remembers **events**.

---

## Interview Definition

> Episodic Memory stores previous experiences, interactions, and completed tasks that may help solve future problems.

---

Example

```text
Yesterday

↓

User asked

↓

Build FastAPI API

↓

Successfully Completed
```

Next time

```text
Agent

↓

Uses previous experience
```

---

Another Example

```text
Last Month

↓

SQL Query Failed

↓

Cause

Missing Index

↓

Remember Solution
```

Future queries can reuse that experience.

---

### Episodic Memory Flow

```text
Experience
→ Store
→ Retrieve Similar Experience
→ Apply Learning
```

---

# 5. Semantic Memory

Semantic Memory stores **facts and knowledge**.

It answers

```
What do I know?
```

instead of

```
What happened?
```

---

## Interview Definition

> Semantic Memory stores structured knowledge, facts, concepts, and relationships that are independent of specific experiences.

---

Examples

```text
Python is a programming language.

↓

LangGraph is an orchestration framework.

↓

Transformer introduced Attention.
```

These are facts,

not experiences.

---

### Semantic Memory Flow

```text
Knowledge
→ Store
→ Retrieve Fact
→ Answer
```

---

# 13. Interview Questions

## Q1. What is Working Memory?

**Answer:**

Working Memory stores temporary information required during the current task or conversation.

---

## Q2. What is Episodic Memory?

**Answer:**

Episodic Memory stores previous experiences, completed tasks, and interactions that help the AI Agent learn from past events.

---

## Q3. What is Semantic Memory?

**Answer:**

Semantic Memory stores factual knowledge, concepts, and relationships independent of specific experiences.

---

## Q4. Which memory stores previous conversations?

**Answer:**

If the goal is to remember the **experience or interaction**, it is Episodic Memory.

If the goal is to remember **facts extracted from conversations** (for example, "User prefers Python"), it can be stored as Semantic or profile memory, depending on the implementation.

---

## Q5. Which memory is used in LangGraph State?

**Answer:**

Working Memory.

The graph state represents the temporary workflow state.

---

# 14. Common Interview Mistakes

❌ **"Working Memory and Long-Term Memory are the same."**

Wrong.

Working Memory exists only during the current task.

---

❌ **"Episodic Memory stores facts."**

Wrong.

It stores experiences.

---

❌ **"Semantic Memory stores conversations."**

Wrong.

It stores knowledge extracted from conversations, not the conversation itself.

# Interview Summary (30-Second Answer)

> **AI Agents typically use three memory types. Working Memory stores temporary information required for the current task. Episodic Memory stores past experiences and completed interactions that help the agent learn from history. Semantic Memory stores factual knowledge and concepts, such as programming languages or business rules. Together, these memory types enable context-aware, personalized, and intelligent behavior.**
