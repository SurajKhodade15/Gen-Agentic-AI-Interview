# 2. What is Memory?

Memory is the ability of an AI Agent to store, retrieve, and use information from previous interactions or experiences to improve future responses and decision-making.

---

## Interview Definition

> Memory is the mechanism that allows an AI Agent to retain and retrieve relevant information across one or multiple interactions, enabling personalized and context-aware behavior.

---

# 9. Memory vs Context Window

This is a **very common interview question**.

| Context Window | Memory |
|----------------|--------|
| Temporary | Persistent |
| Exists during one conversation | Can exist across conversations |
| Limited by model tokens | Stored externally |
| Automatically removed | Retrieved when needed |

---

Example

Context Window

```text
Current Chat Only
```

Memory

```text
Remember

↓

User prefers Python

↓

Available next week

↓

Available next month
```

# 10. Where is Memory Stored?

Memory is usually stored outside the LLM.

Examples

- SQL Database
- Redis
- Vector Database
- Graph Database
- Files
- Cloud Storage

The LLM does **not** permanently store user information itself.

---

# 11. Memory Lifecycle

```text
User Input
→ Store Memory
→ Retrieve Memory
→ Update Memory
→ Use Memory
→ Generate Response
```

# 17. Interview Questions

## Q1. What is Memory in AI Agents?

**Answer:**

Memory is the ability of an AI Agent to store and retrieve relevant information from previous interactions or experiences, enabling personalized and context-aware responses.

---

## Q2. Why is Memory important?

**Answer:**

Memory enables AI Agents to remember user preferences, previous conversations, completed tasks, and learned experiences, improving continuity and personalization.

---

## Q3. Is Memory the same as the Context Window?

**Answer:**

No.

The context window is temporary and limited to the current interaction, whereas memory is persistent and stored externally for future retrieval.

---

## Q4. Where is memory stored?

**Answer:**

Typically in external systems such as SQL databases, Redis, vector databases, graph databases, or cloud storage.

---

## Q5. Does the LLM permanently remember users?

**Answer:**

No.

The LLM relies on external memory systems managed by the application.

---

# 18. Common Interview Mistakes

❌ **"LLMs permanently remember everything."**

Wrong.

Persistent memory is implemented by the application, not the model itself.

---

❌ **"Memory and RAG are the same."**

Wrong.

RAG retrieves external knowledge, while memory stores user- or agent-specific information.

---

❌ **"Context window is memory."**

Wrong.

The context window is temporary; memory is persistent.

# Interview Summary (30-Second Answer)

> **Memory enables AI Agents to store and retrieve relevant information across interactions, allowing them to personalize responses, maintain continuity, and improve decision-making. Unlike the context window, which is temporary and limited by tokens, memory is persistent and typically stored in external systems such as databases or vector stores.**

# 20. Memory vs RAG vs Context Window

| Feature | Memory | RAG | Context Window |
|---------|--------|-----|----------------|
| Purpose | Remember user/agent information | Retrieve external knowledge | Hold current conversation |
| Persistent | ✅ | ✅ | ❌ |
| Uses Vector DB | Sometimes | Usually | ❌ |
| Personalization | ✅ | ❌ | Limited |
| Example | User likes Python | Company HR Policy | Current chat messages |

# Key Takeaway

> **Memory answers: "What do I remember about this user?"**
>
> **RAG answers: "What information should I retrieve?"**
>
> **Context Window answers: "What is being discussed right now?"**
