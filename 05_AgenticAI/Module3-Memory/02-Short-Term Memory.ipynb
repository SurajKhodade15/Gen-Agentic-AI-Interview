# 2. What is Short-Term Memory?

Short-Term Memory stores temporary information needed to complete the **current conversation or task**.

Once the task finishes, this memory is usually cleared.

---

## Interview Definition

> Short-Term Memory is temporary memory that stores information required during the current interaction or workflow, enabling the AI Agent to maintain context until the task is completed.

---

# 3. Human Analogy

Imagine solving a math problem.

While solving it, you remember:

- Numbers
- Intermediate calculations
- Current formula

After finishing,

you usually forget these temporary details.

This is exactly how Short-Term Memory works.

# 10. Short-Term vs Long-Term Memory

| Short-Term Memory | Long-Term Memory |
|-------------------|------------------|
| Temporary | Persistent |
| Current task only | Across sessions |
| Cleared after task | Stored permanently |
| Conversation state | User profile, preferences, history |

# 11. LangGraph Perspective

LangGraph uses **State** as Short-Term Memory.

```text
User
→ State Object
→ Node A
→ Update State
→ Node B
→ Update State
→ Final Response
```

Each node reads and updates the shared state.

---

Example State

```python
state = {
    "goal": "Book Flight",
    "destination": "Delhi",
    "date": "15 Aug",
    "source": "Pune"
}
```

---

# 12. Python Example

```python
state = {}

state["destination"] = "Delhi"
state["date"] = "15 Aug"
state["source"] = "Pune"

print(state)
```

Output

```python
{
    "destination": "Delhi",
    "date": "15 Aug",
    "source": "Pune"
}
```

# 17. Interview Questions

## Q1. What is Short-Term Memory?

**Answer:**

Short-Term Memory stores temporary information required during the current conversation or workflow. It helps the AI Agent maintain context until the task is completed.

---

## Q2. What kind of information is stored?

**Answer:**

Current conversation state, intermediate reasoning, tool outputs, workflow progress, and temporary user inputs.

---

## Q3. How is Short-Term Memory different from Long-Term Memory?

**Answer:**

Short-Term Memory exists only during the current task, whereas Long-Term Memory persists across multiple conversations and sessions.

---

## Q4. How is Short-Term Memory implemented in LangGraph?

**Answer:**

LangGraph uses a shared **State** object. Each node reads from and writes to this state, allowing the workflow to maintain context across execution steps.

---

## Q5. Is Short-Term Memory stored in a Vector Database?

**Answer:**

Usually, no.

It is commonly stored in memory, session state, or temporary caches. Vector databases are more commonly used for Long-Term Memory or RAG.

---

# 18. Common Interview Mistakes

❌ **"Short-Term Memory is the same as chat history."**

Wrong.

It includes the **entire workflow state**, not just previous messages.

---

❌ **"Short-Term Memory survives after the conversation."**

Wrong.

It is typically cleared once the task finishes.

---

❌ **"Every memory should go into a vector database."**

Wrong.

Short-Term Memory is often stored in RAM or application state because it is temporary.

# Interview Summary (30-Second Answer)

> **Short-Term Memory is temporary memory used by AI Agents to maintain the current conversation or workflow state. It stores information such as user inputs, intermediate reasoning, tool outputs, and execution progress. Once the task is completed, this memory is usually cleared. Frameworks like LangGraph implement this using a shared state object.**

# 20. Memory Hierarchy

```text
User Request
        │
        ▼
Short-Term Memory
(Current Task State)
        │
        ▼
Long-Term Memory
(User Preferences, History)
        │
        ▼
Knowledge Base / RAG
(External Documents)
```

This hierarchy is commonly used in production AI Agent architectures.

# Key Takeaway

> **Short-Term Memory answers: "What is happening right now?"**
>
> **Long-Term Memory answers: "What do I know from the past?"**
