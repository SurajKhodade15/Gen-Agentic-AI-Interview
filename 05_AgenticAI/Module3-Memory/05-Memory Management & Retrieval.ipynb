# 17. Interview Questions

## Q1. Should every conversation be stored?

**Answer:**

No. Only information with long-term value, such as user preferences, goals, or important project details, should be stored.

---

## Q2. Why is memory ranking important?

**Answer:**

Because retrieving every stored memory increases token usage and can reduce response quality. Ranking ensures only the most relevant memories are sent to the LLM.

---

## Q3. What is Top-K retrieval?

**Answer:**

Top-K retrieval selects the highest-ranked **K** memories based on relevance scores and includes only those in the prompt.

---

## Q4. How should conflicting memories be handled?

**Answer:**

The system should detect conflicts, update or replace outdated memories, and maintain a single consistent representation whenever possible.

---

## Q5. What is Hybrid Search?

**Answer:**

Hybrid Search combines keyword-based retrieval with semantic (embedding-based) retrieval to improve both precision and recall.

---

# 18. Common Interview Mistakes

❌ **"Store everything."**

Wrong.

Store only valuable long-term information.

---

❌ **"Retrieve everything."**

Wrong.

Use relevance ranking and Top-K selection.

---

❌ **"Never delete memory."**

Wrong.

Outdated or incorrect memories should be updated or removed.

# Interview Summary (30-Second Answer)

> **Memory management is the process of deciding what information should be stored, how it should be retrieved, updated, ranked, and eventually forgotten. In production AI systems, only valuable long-term information is stored, relevant memories are retrieved using keyword, semantic, or hybrid search, Top-K memories are added to the prompt, and outdated memories are updated or removed to maintain accuracy and efficiency.**
