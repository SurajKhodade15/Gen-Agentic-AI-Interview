# 1. Why Do We Need Tools?

Suppose you ask ChatGPT:

```
What is today's weather in Pune?
```

Can the LLM answer accurately?

**No.**

Because the weather changes every minute.

Instead, the agent should:

```text
User Question
        │
        ▼
Weather API
        │
        ▼
Current Temperature
        │
        ▼
LLM explains the result
```

The Weather API is a **Tool**.

---

# 2. What is a Tool?

A **Tool** is any external capability that an AI Agent can invoke to perform actions or retrieve information beyond the LLM's built-in knowledge.

---

## Interview Definition

> A Tool is an external function, API, database, or software component that an AI Agent can invoke to interact with the outside world or perform specialized tasks.

---

# 3. Why LLMs Need Tools

LLMs are limited to reasoning over the information provided in their prompt and training.

They cannot reliably:

- Access the internet
- Query SQL databases
- Execute Python code
- Read files
- Send emails
- Book flights
- Check bank balances

Tools remove these limitations.

# 13. Tool vs Function

| Tool | Function |
|------|----------|
| High-level capability | Implementation |
| May call multiple functions | Single executable unit |
| Used by the agent | Executed by code |

Example

```text
Tool

↓

Weather Tool

↓

Function

↓

get_weather(city)
```

Every function can be exposed as a tool, but not every tool is just a single function.

---

# 14. Python Example

```python
def get_weather(city):
    return f"Sunny in {city}"

result = get_weather("Pune")

print(result)
```

Output

```text
Sunny in Pune
```

When exposed to an AI Agent, `get_weather()` becomes a tool.

# 19. Interview Questions

## Q1. What is a Tool?

**Answer:**

A Tool is an external capability such as an API, function, database, or software component that an AI Agent invokes to retrieve information or perform actions beyond the LLM's built-in abilities.

---

## Q2. Why are Tools needed?

**Answer:**

LLMs cannot access live data or interact with external systems on their own. Tools allow agents to perform tasks such as web search, database queries, file operations, and API calls.

---

## Q3. Give examples of AI Agent tools.

**Answer:**

- Weather API
- SQL Database
- Python Interpreter
- Web Search
- File Reader
- Email Service
- Calculator

---

## Q4. What is the difference between a Tool and an LLM?

**Answer:**

The LLM performs reasoning and language generation, while tools execute external actions or retrieve live information. The agent combines both to solve tasks.

---

## Q5. Can an AI Agent use multiple tools?

**Answer:**

Yes. A single request may require several tools. For example, an agent can search the web, query a database, execute Python code, and then generate a final response.

---

# 20. Common Interview Mistakes

❌ **"A Tool is the same as an API."**

Wrong.

An API is one type of tool. Tools also include databases, Python interpreters, file systems, and calculators.

---

❌ **"The LLM executes the tool."**

Wrong.

The **agent or orchestration framework** invokes the tool. The LLM decides **which tool** to use.

---

❌ **"Tools replace LLMs."**

Wrong.

Tools provide capabilities; the LLM still performs reasoning and response generation.

# Interview Summary (30-Second Answer)

> **A Tool is an external capability that an AI Agent uses to interact with the outside world. While an LLM is responsible for reasoning and language generation, tools enable actions such as calling APIs, querying databases, executing Python code, reading files, or searching the web. Modern agent frameworks combine LLM reasoning with tool execution to solve real-world tasks.**
