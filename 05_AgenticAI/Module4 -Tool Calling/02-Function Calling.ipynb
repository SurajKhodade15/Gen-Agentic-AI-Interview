```
What's the weather in Pune?
```

```text
User
    │
    ▼
LLM
    │
    ▼
"I think it's sunny."
```

```text
User
    │
    ▼
LLM
    │
    ▼
Generate Function Call
    │
    ▼
get_weather("Pune")
    │
    ▼
Weather API
    │
    ▼
Current Weather
    │
    ▼
LLM
    │
    ▼
Final Answer
```

# 2. What is Function Calling?

Function Calling allows an LLM to request the execution of predefined functions by generating structured arguments instead of directly answering the question.

The application executes the function and returns the result to the LLM.

---

## Interview Definition

> Function Calling is a mechanism where an LLM identifies that a predefined function is needed, generates the required arguments, and lets the application execute the function before producing the final response.

---

# 3. Who Executes the Function?

This is one of the most important interview questions.

Many people think:

```text
LLM
↓

Executes Function
```

❌ Wrong

Correct flow:

```text
LLM
↓

Requests Function

↓

Application Executes Function

↓

Result Sent Back

↓

LLM Generates Final Response
```

**The LLM never executes your Python code directly.**

---

# 4. Function Calling Workflow

```text
User
    │
    ▼
LLM
(Decides a function is needed)
    │
    ▼
Generate Function Name + Arguments
    │
    ▼
Application Executes Function
    │
    ▼
Function Result
    │
    ▼
LLM
    │
    ▼
Final Response
```

# 10. Python Example

```python
def get_weather(city):
    return f"Sunny in {city}"

city = "Pune"

result = get_weather(city)

print(result)
```

Output

```text
Sunny in Pune
```

In an AI Agent, the LLM decides:

- Function name
- Arguments

Your application executes the function.

# 11. Function Calling vs Tool Calling

This is one of the most common interview questions.

| Function Calling | Tool Calling |
|------------------|--------------|
| Calls a predefined function | Uses any external capability |
| Structured arguments | High-level capability |
| Usually one executable unit | May involve multiple functions |
| Often JSON arguments | API, SQL, Python, File, Search, etc. |

Think of it like this:

```text
Tool
    │
    ▼
One or More Functions
```

Every Function can be exposed as a Tool.

Not every Tool is a single Function.

---

# 12. Function Calling vs API Calling

| Function Calling | API Calling |
|------------------|-------------|
| Internal application logic | Network request |
| May or may not call an API | Always communicates with another service |
| Example: calculate_tax() | GET /weather |

A function may internally call an API, but they are not the same thing.

# 15. Production Flow

```text
User Question
        │
        ▼
Reasoning
        │
        ▼
Need Function?
        │
     Yes
        │
        ▼
Generate Function Call
        │
        ▼
Validate Arguments
        │
        ▼
Execute Function
        │
        ▼
Return Result
        │
        ▼
LLM Response
```

# 16. Best Practices

✅ Expose only required functions

✅ Validate generated arguments

✅ Handle exceptions

✅ Log function calls

✅ Retry transient failures

✅ Never blindly trust generated arguments

# 17. Interview Questions

## Q1. What is Function Calling?

**Answer:**

Function Calling allows an LLM to generate structured function requests with arguments. The application executes the function and returns the result, which the LLM uses to produce the final response.

---

## Q2. Does the LLM execute Python code?

**Answer:**

No.

The LLM only decides which function to call and generates its arguments. The application executes the function.

---

## Q3. Why is Function Calling better than asking the LLM to generate code?

**Answer:**

Function Calling provides structured, validated execution using predefined functions, reducing hallucinations and improving reliability.

---

## Q4. What happens if the generated arguments are incorrect?

**Answer:**

The application should validate the arguments, return an error if necessary, or ask the user for clarification before executing the function.

---

## Q5. Can a single user request trigger multiple function calls?

**Answer:**

Yes.

For example, an agent might search the web, query a database, and then send an email before generating the final response.

---

# 18. Common Interview Mistakes

❌ **"The LLM executes functions."**

Wrong.

The application executes them.

---

❌ **"Function Calling only works with APIs."**

Wrong.

Functions can perform calculations, query databases, read files, or execute any business logic.

---

❌ **"Function Calling replaces Tools."**

Wrong.

Function Calling is one implementation mechanism used to invoke tools.

# Interview Summary (30-Second Answer)

> **Function Calling enables an LLM to interact with external systems in a structured way. Instead of executing code itself, the LLM selects a predefined function, generates the required arguments, and asks the application to execute it. The application returns the result, which the LLM uses to produce the final response. This makes AI agents more reliable, secure, and capable of performing real-world tasks.**

# 20. Tool Calling vs Function Calling vs API Calling

| Feature | Tool Calling | Function Calling | API Calling |
|---------|--------------|------------------|-------------|
| Purpose | Use external capability | Execute predefined function | Communicate with external service |
| Scope | Broad | Specific | Network-based |
| Executed By | Application | Application | Application |
| Example | SQL Tool | `run_sql(query)` | `POST /query` |
| May Call APIs | ✅ | ✅ | N/A |

---

# Key Takeaway

> **Tool Calling = "What capability should I use?"**
>
> **Function Calling = "Which function should I execute?"**
>
> **API Calling = "How do I communicate with another service?"**
