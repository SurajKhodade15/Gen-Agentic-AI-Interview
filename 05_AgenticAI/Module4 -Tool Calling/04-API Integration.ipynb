# 1. What is an API?

An **API (Application Programming Interface)** is a contract that allows one application to communicate with another.

Example:

```
Chatbot
        │
        ▼
Weather API
        │
        ▼
Current Weather
```

The chatbot does not know how weather data is generated.

It simply requests the information through an API.

# 4. REST API Basics

The most common APIs used with AI Agents are **REST APIs**.

Common HTTP methods:

| Method | Purpose |
|---------|----------|
| GET | Retrieve data |
| POST | Create data |
| PUT | Update data |
| PATCH | Partially update data |
| DELETE | Remove data |

# 8. Authentication

Most APIs require authentication.

Common methods:

| Method | Description |
|---------|-------------|
| API Key | Simple token |
| Bearer Token (OAuth/JWT) | Most common |
| OAuth 2.0 | User authorization |
| Basic Auth | Username/password (less common today) |

---

Example

```http
Authorization: Bearer <token>
```

Never hardcode secrets in your application.

# 12. Retry Strategy

Some failures are temporary.

```text
Request
        │
        ▼
429 Rate Limited
        │
        ▼
Wait
        │
        ▼
Retry
```

Use retries for transient errors such as:

- Network timeout
- 429 (Rate limit)
- 503 (Service unavailable)

Do **not** blindly retry every error.

---

# 13. Timeout Handling

Never wait forever.

```text
Request

↓

Wait

↓

Timeout

↓

Retry / Fail Gracefully
```

Always configure reasonable timeouts.

# 17. Interview Questions

## Q1. Why do AI Agents use APIs?

**Answer:**

APIs allow AI Agents to interact with external systems, retrieve live information, and perform actions such as creating tickets, sending emails, or querying enterprise applications.

---

## Q2. What is the difference between a Tool and an API?

**Answer:**

A Tool is a capability exposed to the AI Agent. An API is one implementation of that capability. For example, a "Weather Tool" may internally call a weather API.

---

## Q3. What is the most common API style?

**Answer:**

REST APIs are the most commonly used, although GraphQL and gRPC are also used in some systems.

---

## Q4. How should an AI Agent handle API failures?

**Answer:**

Validate responses, retry transient failures, use timeouts, log errors, and return a user-friendly message when the request cannot be completed.

---

## Q5. Where should API keys be stored?

**Answer:**

In secure configuration such as environment variables or a secrets manager—not in source code or prompts.

---

# 18. Common Interview Mistakes

❌ **"The LLM calls the API directly."**

Wrong.

The application performs the API request after the LLM requests the function.

---

❌ **"Retry every failure."**

Wrong.

Retry only transient errors like timeouts or rate limits.

---

❌ **"Store API keys in the code."**

Wrong.

Use environment variables or dedicated secret management systems.

# Interview Summary (30-Second Answer)

> **API Integration enables AI Agents to communicate with external systems such as CRMs, weather services, ticketing platforms, and databases. The LLM determines that an API is needed through function calling, the application executes the request, validates the response, handles authentication and failures, and then provides the result back to the LLM to generate the final response.**

# 20. Relationship Between Concepts

```text
User Request
        │
        ▼
Reasoning (LLM)
        │
        ▼
Tool Selection
        │
        ▼
Function Calling
        │
        ▼
API Integration
        │
        ▼
External Service
        │
        ▼
Result
        │
        ▼
LLM Response
```

This illustrates how the topics in Module 4 connect:

- **Tools** define available capabilities.
- **Function Calling** is how the LLM requests those capabilities.
- **Tool Selection & Routing** chooses the right capability.
- **API Integration** is how many real-world tools communicate with external services.

---

# Key Takeaway

> **Function Calling decides *what* to execute.**
>
> **API Integration determines *how* the application communicates with external systems to perform that execution.**
