# 1. Why Do We Need Reflexion?

Imagine an AI Agent receives this task:

```
Find the cheapest flight to Delhi.
```

Execution

```text
Search Airline A → ₹12,000 → Book Flight
```

Later it discovers:

```
Airline B → ₹9,500
```

The agent made a mistake.

A normal ReAct agent simply completes the task.

A Reflexion agent asks:

```
Why did I fail?

How can I improve next time?
```

This ability is called **Self-Reflection**.

---

# 2. What is Reflexion?

Reflexion is a reasoning framework where an AI Agent evaluates its previous actions, identifies mistakes, stores lessons learned, and uses those lessons to improve future decisions.

Unlike training a new model, Reflexion improves behavior **during execution**.

---

## Interview Definition

> Reflexion is an AI Agent technique in which the agent reviews its previous reasoning and actions, learns from mistakes, stores feedback in memory, and applies those learnings to future decisions.

---

# 3. Normal Agent vs Reflexion Agent

## ReAct

```text
Goal → Think → Act → Observe → Answer
```

Task completed.

---

## Reflexion

```text
Goal → Think → Act → Observe → Reflect → Improve → Repeat → Answer
```

Reflection is an additional learning step.

---

# 4. Reflexion Lifecycle

```text
Goal → Think → Act → Observe → Reflect → Store Lesson → Retry (if needed) → Final Answer
```

Notice the extra **Store Lesson** step.

---

# 5. What Happens During Reflection?

The agent asks questions like:

- Did I choose the correct tool?
- Was my answer accurate?
- Did the API fail?
- Could another strategy work better?
- What should I remember for next time?

# 9. Reflection vs Observation

Many candidates confuse these.

| Observation | Reflection |
|-------------|------------|
| Reads tool output | Evaluates the output |
| "0 flights found" | "Try another airline" |
| Collects information | Improves strategy |
| Passive | Active reasoning |

Example

```text
Observation:
Weather API failed.

Reflection:
Let's retry with another API.
```

# 10. Reflection Memory

One of the biggest advantages of Reflexion is storing lessons.

Example

```text
Lesson Learned

Amazon API often times out.

↓

Use Flipkart first.
```

Next execution

```text
Skip Amazon

↓

Use Flipkart
```

Performance improves over time.

# 11. ReAct vs Reflexion

| ReAct | Reflexion |
|--------|-----------|

| Uses Tools | Uses Tools |
| Observes Results | Observes Results |

# 13. Python Example (Conceptual)

```python
goal = "Solve coding problem"

while True:

    answer = llm.solve(goal)

    feedback = evaluate(answer)

    if feedback == "correct":
        break

    memory.store(feedback)

    goal = goal + "\nImprove using feedback."

print(answer)
```

# 18. Interview Questions

## Q1. What is Reflexion?

**Answer:**

Reflexion is an AI Agent reasoning technique where the agent evaluates its previous actions, learns from mistakes, stores useful feedback, and improves future decisions without retraining the model.

---

## Q2. Why is Reflexion important?

**Answer:**

It enables AI Agents to continuously improve their performance during execution by analyzing failures and adjusting their strategies.

---

## Q3. What is the difference between Observation and Reflection?

**Answer:**

Observation collects the result of an action, while Reflection analyzes that result and determines whether the strategy should change.

---

## Q4. Does Reflexion require model fine-tuning?

**Answer:**

No.

The improvement comes from reasoning and memory, not from updating the model's weights.

---

## Q5. What is the difference between ReAct and Reflexion?

**Answer:**

ReAct alternates between reasoning and acting to complete a task. Reflexion extends ReAct by adding a self-evaluation step that learns from mistakes and improves future decisions.

---

# 19. Common Interview Mistakes

❌ **"Reflection means storing chat history."**

Wrong.

Reflection analyzes actions.

Memory stores information.

---

❌ **"Reflexion retrains the LLM."**

Wrong.

The model parameters remain unchanged.

The agent improves using feedback and memory.

---

❌ **"Reflection happens only after failure."**

Wrong.

Reflection can occur after both successful and unsuccessful actions to identify opportunities for improvement.
