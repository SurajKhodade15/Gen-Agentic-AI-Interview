# 1. Why Do We Need Tree of Thoughts?

Suppose you're asked:

```
Plan a startup that can generate ₹10 Crore in revenue.
```

There isn't a single obvious solution.

Possible approaches:

- SaaS
- E-commerce
- AI Product
- Consulting
- Marketplace

A normal Chain of Thought follows **one path**.

If that path is poor, the final answer may also be poor.

Instead, we want the model to explore multiple possibilities and choose the best one.

That's Tree of Thoughts.

---

# 2. What is Tree of Thoughts?

Tree of Thoughts (ToT) is a reasoning framework where an AI explores multiple reasoning branches, evaluates each branch, discards weak ones, and continues expanding the most promising paths.

---

## Interview Definition

> Tree of Thoughts is a reasoning strategy where an AI model generates multiple possible reasoning paths, evaluates them, and searches for the best solution instead of following a single chain of reasoning.

---

# 3. Chain of Thought vs Tree of Thoughts

## Chain of Thought

```text
Question
→ Thought 1
→ Thought 2
→ Thought 3
→ Answer
```

Only one reasoning path.

---

## Tree of Thoughts

```text
Question
→ Idea A
→ Idea B
→ Idea C

Idea A → Expand
Idea B → Expand
Idea C → Discard

↓

Best Answer
```

Multiple reasoning paths are explored.

---

# 4. ToT Workflow

```text
Goal
→ Generate Multiple Ideas
→ Evaluate Each Idea
→ Keep Best Ideas
→ Expand Them
→ Select Best Solution
```

This is the core interview flow.

---

# 9. Internal Workflow

```text
Problem

↓

Generate Thoughts

↓

Evaluate Thoughts

↓

Discard Weak Thoughts

↓

Expand Best Thoughts

↓

Final Answer
```

Unlike CoT, the model continuously compares different reasoning paths.

# 10. Search Strategies

Tree of Thoughts commonly uses search algorithms.

### Breadth-First Search (BFS)

```text
Root
→ A
→ B
→ C
```

Explore every branch equally.

---

### Depth-First Search (DFS)

```text
Root
→ A
→ A1
→ A2
→ Answer
```

Explore one branch deeply before trying another.

---

### Beam Search

```text
Generate 10 ideas

↓

Keep Top 3

↓

Expand Only Top 3

↓

Repeat
```

Very popular because it balances quality and efficiency.

# 13. ReAct vs ToT

| ReAct | Tree of Thoughts |
|--------|------------------|
| Think + Act | Think in multiple branches |
| Uses tools | May or may not use tools |
| Sequential reasoning | Parallel reasoning paths |
| Dynamic execution | Search-based reasoning |

---

# 14. Plan-and-Execute vs ToT

| Plan-and-Execute | Tree of Thoughts |
|------------------|------------------|
| One execution plan | Multiple candidate plans |
| Execute chosen plan | Compare several plans first |
| Structured workflow | Search and evaluation workflow |

# 18. Interview Questions

## Q1. What is Tree of Thoughts?

**Answer:**

Tree of Thoughts is a reasoning framework where an AI generates multiple reasoning paths, evaluates them, expands the most promising ones, and selects the best solution.

---

## Q2. Why is ToT better than CoT?

**Answer:**

CoT follows a single reasoning path, whereas ToT explores multiple alternative paths, reducing the chance of getting stuck with a poor solution.

---

## Q3. Does ToT always use tools?

**Answer:**

No.

Tree of Thoughts is primarily a reasoning strategy. It can be combined with tool usage but does not require it.

---

## Q4. Which search strategies are commonly used in ToT?

**Answer:**

- Breadth-First Search (BFS)
- Depth-First Search (DFS)
- Beam Search

---

## Q5. When should we use Tree of Thoughts?

**Answer:**

Use it for complex problems with multiple possible solutions, such as planning, coding, optimization, research, and strategic decision-making.

---

# 19. Common Interview Mistakes

❌ **"Tree of Thoughts is just a longer Chain of Thought."**

Wrong.

It explores **multiple branches**, not just more reasoning steps.

---

❌ **"Every branch is executed."**

Wrong.

Weak branches are evaluated and pruned.

---

❌ **"ToT replaces ReAct."**

Wrong.

ToT is a reasoning strategy. ReAct is a reasoning + action framework. They can be combined.
