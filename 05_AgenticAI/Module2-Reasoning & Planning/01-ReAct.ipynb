# 1. Why ReAct?

Let's start with a problem.

Suppose you ask an LLM:

```
Who is the current CEO of Microsoft?
```

If the model's knowledge is outdated, it may answer incorrectly.

Why?

Because a normal LLM can only **reason** using its training data.

It **cannot verify** the answer.

---

Another example:

```
What's today's weather in Pune?
```

The LLM cannot know today's weather.

It needs to **use a Weather API**.

This is where **ReAct** comes in.

---

# 2. What is ReAct?

**ReAct** stands for:

```
Reason + Act
```

Instead of only reasoning,

the model also performs actions using tools.

---

## Interview Definition

> ReAct is an agent reasoning framework where the LLM alternates between reasoning about the problem and taking actions using external tools until the final answer is obtained.

---

# 3. Traditional LLM vs ReAct

## Traditional LLM

```text
Question → Think → Answer
```

Only one step.

---

## ReAct Agent

```text
Question → Think → Act → Observe → Think → Act → Observe → Answer
```

Notice the repeated loop.

---

# 4. ReAct Lifecycle

```text
User Question
      ↓
Think (Reason)
      ↓
Choose Tool
      ↓
Execute Tool
      ↓
Observe Result
      ↓
Need More Information?
      ├── Yes → Think Again
      └── No  → Final Answer
```

This loop continues until the goal is achieved.

---

# 5. Core Components

| Step | Description |
|------|-------------|
| Think | Analyze the problem |
| Act | Select and execute a tool |
| Observe | Read the tool output |
| Repeat | Continue until sufficient information is collected |
| Answer | Generate the final response |

# 10. Real-world Example

Goal

```
Find the cheapest iPhone.
```

ReAct Execution

```text
Think
→ Need prices

Act
→ Search Amazon

Observe
→ ₹74,000

Think
→ Compare another website

Act
→ Search Flipkart

Observe
→ ₹72,500

Think
→ Cheapest found

Answer
→ Flipkart offers the lowest price.
```

# 11. Internal Reasoning

A ReAct Agent typically performs this internally:

```
Thought:
Need today's weather.

Action:
Weather Tool

Observation:
27°C

Thought:
Enough information.

Final Answer:
Today's weather is 27°C.
```

> **Note:** In production systems, these internal thoughts are often hidden from users for safety and privacy reasons.

# 14. ReAct vs Normal Tool Calling

Normal Tool Calling

```text
Question
→ Tool
→ Answer
```

Only one tool call.

---

ReAct

```text
Question
→ Think
→ Tool
→ Observe
→ Think
→ Tool
→ Observe
→ Answer
```

Multiple reasoning cycles.

---

# 15. ReAct vs Workflow

Workflow

```text
Task
→ Step 1
→ Step 2
→ Step 3
→ Done
```

Fixed.

---

ReAct

```text
Task
→ Think
→ Decide
→ Execute
→ Observe
→ Decide Again
→ Done
```

Dynamic.

# 18. Interview Questions

## Q1. What is ReAct?

**Answer:**

ReAct (Reason + Act) is an AI Agent framework in which the LLM alternates between reasoning about the problem and taking actions using external tools until it has enough information to produce the final answer.

---

## Q2. Why was ReAct introduced?

**Answer:**

Traditional LLMs rely only on their training data. ReAct enables LLMs to interact with external tools, retrieve live information, and make better decisions based on observations.

---

## Q3. What are the main steps in ReAct?

**Answer:**

```text
Think → Act → Observe → Repeat → Final Answer
```

---

## Q4. Can ReAct use multiple tools?

**Answer:**

Yes. During different reasoning steps, the agent may call multiple tools such as web search, SQL, Python, or APIs before generating the final response.

---

## Q5. Which frameworks use ReAct?

**Answer:**

The ReAct pattern is used directly or conceptually in:

- LangChain Agents
- LangGraph
- CrewAI
- AutoGen
- OpenAI Agents SDK

---

# 19. Common Interview Mistakes

❌ **"ReAct means calling one tool."**

Wrong.

ReAct may involve **multiple reasoning and tool execution cycles**.

---

❌ **"ReAct replaces the LLM."**

Wrong.

The LLM is still the reasoning engine.

ReAct defines **how** the LLM reasons and interacts with tools.

---

❌ **"ReAct always uses web search."**

Wrong.

ReAct can use any tool:

- SQL
- Python
- APIs
- File Readers
- RAG
- Search
- Email
- Calculator

---

# 20. Quick Revision

|------|----------------|

| Act | Execute the selected tool |
| Observe | Analyze tool output |
| Repeat | Continue if needed |
| Answer | Return final response |

# Interview Summary (30-Second Answer)

> **ReAct (Reason + Act) is a reasoning framework where an LLM alternates between thinking and using external tools. After each tool execution, it observes the result, reasons again, and repeats the cycle until it has enough information to generate an accurate final answer.**

---

# Key Takeaway

> **LLM = Think Once**
>
> **ReAct = Think → Act → Observe → Think Again**

This iterative reasoning loop is the foundation of modern AI Agents.
