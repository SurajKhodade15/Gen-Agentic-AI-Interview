# 1. Why Do We Need Graph of Thoughts?

Imagine you're building a business plan.

You need to think about:

- Market
- Pricing
- Competitors
- Customers
- Marketing

While analyzing **pricing**, you may also need competitor data.

Later, while planning **marketing**, you again need competitor data.

In a **Tree**, competitor analysis may be repeated.

In a **Graph**, it can be reused.

---

# 2. What is Graph of Thoughts?

Graph of Thoughts (GoT) is a reasoning framework where thoughts are represented as interconnected nodes instead of a strict tree, allowing information to be shared, reused, and combined across multiple reasoning paths.

---

## Interview Definition

> Graph of Thoughts is an advanced reasoning framework in which thoughts form a graph rather than a tree, enabling multiple reasoning paths to merge, reuse intermediate results, and solve complex problems more efficiently.

---

# 3. Evolution of Reasoning

```text
LLM
→ One Answer

↓

Chain of Thought
→ One Reasoning Path

↓

Tree of Thoughts
→ Multiple Independent Branches

↓

Graph of Thoughts
→ Multiple Connected Branches
```

---

# 4. Tree vs Graph

## Tree of Thoughts

```text
Goal
├── A
│   ├── A1
│   └── A2
├── B
└── C
```

Branches never reconnect.

---

## Graph of Thoughts

```text
Goal
├── Market Analysis ──┐
├── Competitor Analysis ─┼── Business Strategy
├── Customer Analysis ──┘
```

Multiple branches can reuse the same information.

---

# 5. Core Idea

Instead of asking:

```text
What is the next thought?
```

GoT asks:

```text
Which previous thoughts can help solve this problem?
```

This reuse reduces redundant reasoning.

---

# 6. Workflow

```text
Goal
→ Generate Thoughts
→ Connect Related Thoughts
→ Reuse Existing Knowledge
→ Combine Results
→ Final Answer
```

# 10. Tree vs Graph Comparison

| Tree of Thoughts | Graph of Thoughts |
|------------------|-------------------|
| Tree structure | Graph structure |
| Independent branches | Connected branches |
| Limited reuse | High reuse |
| Simpler | More flexible |
| Easier implementation | More complex |

# 11. Graph Components

| Component | Purpose |
|-----------|---------|
| Node | A thought or intermediate result |
| Edge | Relationship between thoughts |
| Merge | Combine reasoning paths |
| Split | Create new branches |
| Reuse | Use previous results again |

# 18. Interview Questions

## Q1. What is Graph of Thoughts?

**Answer:**

Graph of Thoughts is a reasoning framework where intermediate thoughts form a graph instead of a tree, allowing reasoning paths to connect, merge, and reuse previous results.

---

## Q2. Why is GoT better than ToT?

**Answer:**

GoT avoids redundant reasoning by allowing branches to share intermediate knowledge, making it more efficient for highly interconnected problems.

---

## Q3. What is a node in GoT?

**Answer:**

A node represents a thought, intermediate result, or reasoning step within the graph.

---

## Q4. When should Graph of Thoughts be used?

**Answer:**

Use it for complex problems where multiple reasoning paths depend on shared information, such as research, enterprise planning, and knowledge-intensive workflows.

---

## Q5. Is GoT commonly used in production today?

**Answer:**

Not as widely as ReAct or Plan-and-Execute. GoT is primarily a research-oriented reasoning framework, although its concepts influence advanced agent systems.

---

# 19. Common Interview Mistakes

❌ **"Graph of Thoughts is just a bigger Tree of Thoughts."**

Wrong.

The key difference is **connections and reuse**, not size.

---

❌ **"Every node is independent."**

Wrong.

Nodes may depend on and share information with other nodes.

---

❌ **"GoT replaces ReAct."**

Wrong.

GoT is a reasoning strategy. ReAct is an action-oriented framework. They can be combined in sophisticated agents.
