# 1. Why Do We Need Chain of Thought?

Consider this question:

```
A shopkeeper buys a pen for ₹20 and sells it for ₹25.
What is the profit?
```

A normal LLM may answer correctly.

Now consider a harder problem:

```
Rahul has 5 apples.
He gives 2 to Amit.
Amit gives 1 to Neha.
Rahul buys 3 more apples.
How many apples does Rahul have?
```

This requires **multiple reasoning steps**.

Instead of answering immediately, we want the model to think step by step.

This is called **Chain of Thought (CoT)**.

---

# 2. What is Chain of Thought?

Chain of Thought is a prompting technique where the LLM **breaks a complex problem into smaller reasoning steps before producing the final answer**.

---

## Interview Definition

> Chain of Thought (CoT) is a prompting technique that encourages an LLM to perform intermediate reasoning steps before generating the final answer.

---

# 3. Standard Prompt vs Chain of Thought

## Standard Prompt

```text
Question → Answer
```

Example

```
What is 125 × 17?
```

↓

```
2125
```

---

## Chain of Thought

```text
Question
→ Step 1
→ Step 2
→ Step 3
→ Final Answer
```

The model reasons before answering.

---

# 4. Example

Question

```
A train travels 60 km in one hour.

How far will it travel in 5 hours?
```

Normal Response

```
300 km
```

---

Chain of Thought

```text
Step 1:
Speed = 60 km/hour

↓

Step 2:
Time = 5 hours

↓

Step 3:
Distance = Speed × Time

↓

60 × 5 = 300 km

↓

Answer = 300 km
```

---

# 5. Why CoT Improves Accuracy

Without CoT

```text
Question
→ Guess
→ Answer
```

---

With CoT

```text
Question
→ Reason Step 1
→ Reason Step 2
→ Reason Step 3
→ Answer
```

Breaking a problem into smaller parts reduces reasoning errors.

# 7. CoT Flow

# 8. Types of CoT

## Zero-shot CoT

Prompt:

```
Let's think step by step.
```

Example

```text
Question
↓

Let's think step by step.

↓

Reasoning

↓

Answer
```

No examples are provided.

---

## Few-shot CoT

Provide solved examples first.

Example

```text
Example 1

Question

↓

Reasoning

↓

Answer

--------------------

Example 2

Question

↓

Reasoning

↓

Answer

--------------------

Actual Question
```

The model learns the reasoning style from examples.

# 11. CoT vs ReAct

This is a very common interview question.

| Chain of Thought | ReAct |
|------------------|--------|
| Thinks step by step | Thinks and uses tools |
| Internal reasoning | Reasoning + Actions |
| No external APIs | Can call APIs |
| No observations | Observes tool results |
| Ends after reasoning | Iterates until goal is complete |

---

Example

Question

```
What is today's weather?
```

CoT

```text
Think

↓

Cannot know today's weather

↓

Answer may be outdated
```

---

ReAct

```text
Think

↓

Weather API

↓

Observe Result

↓

Answer
```

# 16. Interview Questions

## Q1. What is Chain of Thought?

**Answer:**

Chain of Thought is a prompting technique where an LLM generates intermediate reasoning steps before producing the final answer, improving performance on complex reasoning tasks.

---

## Q2. Why does CoT improve accuracy?

**Answer:**

Breaking a complex problem into smaller reasoning steps reduces mistakes and helps the model solve each part systematically.

---

## Q3. Does CoT use external tools?

**Answer:**

No.

Chain of Thought performs internal reasoning only.

It does not call APIs, databases, or other external tools.

---

## Q4. What is the difference between CoT and ReAct?

**Answer:**

CoT focuses only on reasoning, whereas ReAct combines reasoning with tool usage and observations to solve dynamic, real-world tasks.

---

## Q5. What are Zero-shot and Few-shot CoT?

**Answer:**

- **Zero-shot CoT:** Uses prompts like *"Let's think step by step"* without examples.
- **Few-shot CoT:** Provides solved examples before the actual question to demonstrate the reasoning process.

---

# 17. Common Interview Mistakes

❌ **"CoT and ReAct are the same."**

Wrong.

CoT reasons internally.

ReAct reasons **and** interacts with external tools.

---

❌ **"CoT is only for math."**

Wrong.

It is useful for coding, planning, finance, healthcare, and many other reasoning tasks.

---

❌ **"CoT can access live data."**

Wrong.

It only reasons over the information available in the prompt and model knowledge.
