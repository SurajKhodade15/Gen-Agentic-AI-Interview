## 2. Reverse a String (Without Slicing)

### Question
Reverse a string without using `[::-1]`.

```python
text = "python"

rev = ""

for ch in text:
    rev = ch + rev

print(rev)
```

## 4. Count Character Frequency

### Question
Count frequency of every character.

```python
text = "programming"

freq = {}

for c in text:
    freq[c] = freq.get(c, 0) + 1

print(freq)
```

## 5. First Non-Repeating Character

### Question
Find the first non-repeating character.

```python
text = "programming"

freq = {}

for c in text:
    freq[c] = freq.get(c, 0) + 1

for c in text:
    if freq[c] == 1:
        print(c)
        break
```

## 6. Find Duplicate Elements

### Question
Return duplicate elements.

```python
nums = [1,2,3,2,4,1,5]

seen = set()
duplicates = set()

for n in nums:
    if n in seen:
        duplicates.add(n)
    else:
        seen.add(n)

print(list(duplicates))
```

## 7. Remove Duplicates

### Question
Remove duplicate elements while preserving order.

```python
nums = [1,2,2,3,1,4]

result = []

for n in nums:
    if n not in result:
        result.append(n)

print(result)
```

## 12. Group Employees by Department

### Question
Group employees by department.

```python
employees = [
    {"name":"John","dept":"IT"},
    {"name":"Alice","dept":"HR"},
    {"name":"Bob","dept":"IT"}
]

result = {}

for emp in employees:
    result.setdefault(emp["dept"], []).append(emp["name"])

print(result)
```

## 15. Flatten Nested List

### Question
Flatten a nested list.

```python
nums = [1,[2,3],[4,[5,6]]]

def flatten(lst):

    result = []

    for item in lst:

        if isinstance(item,list):
            result.extend(flatten(item))
        else:
            result.append(item)

    return result

print(flatten(nums))
```

## 29. Timing Decorator

### Question
Write a timer decorator.

```python
import time

def timer(func):

    def wrapper(*args,**kwargs):

        start = time.time()

        result = func(*args,**kwargs)

        print(time.time()-start)

        return result

    return wrapper
```
