**MCP (Model Context Protocol)** is one of the hottest topics in Agentic AI. Since Anthropic introduced it, companies building enterprise AI agents are increasingly adopting it.

---

# 1. What is MCP?

### Interview Answer

> **Model Context Protocol (MCP)** is an open standard that allows AI models to securely communicate with external tools, databases, APIs, and enterprise systems through a standardized protocol. Instead of writing custom integrations for every application, the LLM communicates with MCP servers using a common interface.

Think of it as:

> **"USB-C for AI applications."**

Just as USB-C lets many devices connect through one standard, MCP lets AI models connect to many tools through one protocol.

---

# 2. Why do we need MCP?

### Without MCP

If your AI agent needs to access:

- Salesforce
- Jira
- GitHub
- SQL Database
- SharePoint
- Slack

You write **separate integrations** for each.

```text
LLM
 │
 ├── Salesforce API
 ├── Jira API
 ├── SQL API
 ├── GitHub API
 └── SharePoint API
```

Lots of custom code.

---

### With MCP

```text
LLM
 │
 ▼
MCP Client
 │
 ├── GitHub MCP Server
 ├── SQL MCP Server
 ├── Jira MCP Server
 ├── SharePoint MCP Server
 └── Slack MCP Server
```

The LLM speaks one protocol, and every tool exposes an MCP server.

# 4. Components

| Component | Purpose |
|------------|---------|
| **MCP Client** | Used by the AI application to communicate with MCP servers |
| **MCP Server** | Exposes tools, resources, and prompts to AI models |
| **MCP Tool** | Executable function (e.g., query database, create Jira ticket) |
| **MCP Resource** | Read-only data (documents, files, configs) |
| **MCP Prompt** | Reusable prompt templates |

Resources are **read-only context**.

Examples

- PDF
- Word document
- SharePoint page
- Wiki
- Configuration file

The LLM can read these without executing actions.

# 9. MCP Prompt

Reusable prompts.

Example

```
Summarize this invoice.

Extract total amount.

Extract due date.
```

Instead of rewriting prompts every time, they are published through the MCP server.

# 10. MCP vs REST API

| MCP | REST API |
|------|----------|
| AI-native protocol | General web protocol |
| Tool discovery built in | No discovery |
| Standardized for LLMs | Generic HTTP endpoints |
| Resources, tools, prompts | Endpoints only |
| Designed for Agentic AI | Designed for applications |
| Dynamic capabilities | Fixed endpoints |

### Example

REST

```text
POST /searchCustomer
```

MCP

```
Available Tools:

✓ Search Customer

✓ Create Ticket

✓ Send Email
```

The LLM can discover available tools automatically.

---

# 11. MCP vs Function Calling

| Function Calling | MCP |
|-----------------|-----|
| Tools defined in application code | Tools exposed by MCP servers |
| Local to one application | Standardized across systems |
| Manual registration | Automatic discovery |
| Limited scope | Enterprise-wide integration |

# ⭐ Interview Questions

### Q1. What is MCP?

> MCP is an open standard that enables AI models to communicate with external tools, data sources, and enterprise systems through a common protocol.

---

### Q2. Why use MCP instead of REST APIs?

> REST APIs expose endpoints, but MCP exposes AI-friendly capabilities such as tools, resources, and prompts with built-in discovery. This makes it much easier for AI agents to integrate with enterprise systems.

---

### Q3. What are the main components of MCP?

- MCP Client
- MCP Server
- Tools
- Resources
- Prompts

---

### Q4. Is MCP replacing REST APIs?

> **No.** MCP complements REST APIs rather than replacing them. The MCP server often **calls REST APIs internally**, but exposes them to AI agents in a standardized, AI-friendly way.

---

# ⭐ 30-Second Interview Answer

> "Model Context Protocol (MCP) is an open standard for connecting AI models with external systems. Instead of integrating separately with each API, the AI communicates with an MCP client, which discovers and interacts with MCP servers exposing tools, resources, and prompts. This standardization simplifies enterprise AI integrations, improves interoperability, and reduces custom development effort. In many cases, an MCP server wraps existing REST APIs and presents them in a format that's easier for AI agents to understand and use."
