# 1. What is an MCP Server?

An **MCP Server** is a server that exposes **tools, resources, and prompts** to AI applications using the **Model Context Protocol (MCP)**.

Instead of writing custom integrations for every application,

you expose them once through an MCP Server.

---

# Interview Definition ⭐⭐⭐⭐⭐

> **An MCP Server is a standardized server that exposes tools, resources, and prompts to AI agents and LLM applications using the Model Context Protocol (MCP).**

---

```text
ChatGPT

↓

Custom Integration

↓

Database

----------------

Claude

↓

Custom Integration

↓

Database

----------------

VS Code

↓

Custom Integration

↓

Database
```

---

With MCP

```text
ChatGPT

↓

Claude

↓

VS Code

↓

Cursor

↓

MCP Server

↓

Database
```

One integration works for all MCP-compatible clients.

# 3. MCP Server Responsibilities

An MCP Server can expose:

- Tools
- Resources
- Prompts

It does **not** run the LLM.

# 8. Small Python Example

```python
from mcp.server.fastmcp import FastMCP

mcp = FastMCP("Weather Server")

@mcp.tool
def get_weather(city: str):
    return f"The weather in {city} is sunny."

mcp.run()
```

This exposes

```
get_weather()
```

to any MCP-compatible client.

# 13. MCP Server vs Function Calling

| MCP Server | Function Calling |
|------------|------------------|
| Provides tools | Uses tools |
| Runs externally | Inside AI workflow |
| Can expose many tools | Calls one selected tool |
| Shared across clients | Usually application-specific |

# 16. Common Interview Questions

### Q1. What is an MCP Server?

**Answer**

An MCP Server exposes tools, resources, and prompts to AI applications using the Model Context Protocol.

---

### Q2. Does an MCP Server contain an LLM?

**Answer**

No.

It provides capabilities to LLMs but does not run the language model itself.

---

### Q3. What can an MCP Server expose?

**Answer**

- Tools
- Resources
- Prompts

---

### Q4. Why is MCP important?

**Answer**

It standardizes how AI applications access external tools and data, eliminating the need for custom integrations for each client.

---

### Q5. Can multiple AI applications use the same MCP Server?

**Answer**

Yes.

Any MCP-compatible client can connect to the same MCP Server.

# 18. Real Interview Scenario

**Question:**

> Your company has a CRM API that needs to be accessed by ChatGPT, Claude, Cursor, and an internal AI agent. How would you avoid building four separate integrations?

**Answer:**

I would build a single **MCP Server** that exposes the CRM operations as MCP tools. Each AI client would connect through its MCP client, allowing all of them to use the same standardized interface. This reduces duplication, improves maintainability, and enables consistent access control.
