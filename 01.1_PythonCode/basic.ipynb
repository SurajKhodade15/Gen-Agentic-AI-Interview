```python
# Two Sum
def two_sum(nums, target):
    seen = set()
    pairs = []

    for num in nums:
        complement = target - num

        if complement in seen:
            pairs.append((complement, num))

        seen.add(num)
        print(seen)

    return pairs


nums = [2, 7, 8, 1, 11, 15]
target = 9

print(two_sum(nums, target))
```

```python
## Second Largest Element in an Array
def second_largest(nums):
    first = second = float('-inf')

    for num in nums:
        if num > first:
            second = first
            first = num
        elif first > num > second:
            second = num

    return second


nums = [10, 5, 20, 8, 15,15]
print(second_largest(nums))
```

```python
## Missing Number
def missing_number(nums):
    nums = set(nums)

    expected = 1

    while expected in nums:
        expected += 1

    return expected


nums = [1,2,3,5]
print(missing_number(nums))
```

```python
## Group Anagrams
from collections import defaultdict

def group_anagrams(words):
    groups = defaultdict(list)

    for word in words:
        key = ''.join(sorted(word))
        groups[key].append(word)
    print(groups)
    return list(groups.values())


words = ["eat", "tea", "tan", "ate", "nat", "bat"]

print(group_anagrams(words))
```

```python
## Merge two sorted lists
def merge_sorted_lists(list1, list2):
    i = j = 0
    result = []

    while i < len(list1) and j < len(list2):
        if list1[i] < list2[j]:
            result.append(list1[i])
            i += 1
        else:
            result.append(list2[j])
            j += 1
    print(result)
    result.extend(list1[i:])
    result.extend(list2[j:])

    return result


list1 = [1, 3, 5]
list2 = [2, 4, 6]

print(merge_sorted_lists(list1, list2))
```

```python
## Fibbonaci Sequence with iteration
def fibonacci(n):
    a, b = 0, 1

    for _ in range(n):
        print(a, end=" ")
        a, b = b, a + b


fibonacci(8)
```

```python
## Fibbonacci Series with recursion
def fib(n):
    if n == 0:
        return [0]
    if n == 1:
        return [0, 1]

    series = fib(n - 1)
    series.append(series[-1] + series[-2])
    return series

print(fib(7))
```

```python
## REverse a string
def reverse_string(s):
    rev = ""

    for char in s:
        rev = char + rev

    return rev

print(reverse_string("hello Suraj"))
```

```python
## Character Count
def char_count(s):
    count = {}

    for char in s:
        if char in count:
            count[char] += 1
        else:
            count[char] = 1

    return count

print(char_count("hello Suraj"))
```

```python
## Reverse a words
##Input : "I love python"
##Output: "python love I"

def reverse_words(sentence):
    words = sentence.split()
    return " ".join(words[::-1])

print(reverse_words("I love python"))
```

```python
## Revers words and characters
def reverse_words_and_chars(sentence):
    words = sentence.split()
    result = []

    for word in words[::-1]:
        result.append(word[::-1])

    return " ".join(result)


print(reverse_words_and_chars("I love python"))
```

```python
## Merged two intervals
def merge_intervals(intervals):
    intervals.sort()

    merged = []

    for interval in intervals:
        if not merged or merged[-1][1] < interval[0]:
            merged.append(interval)
        else:
            merged[-1][1] = max(merged[-1][1], interval[1])

    return merged


intervals = [[1,3], [2,6], [8,10], [9,13], [15,18]]

print(merge_intervals(intervals))
```

```python
## Sliding Window Maximum
def max_sum_subarray(nums, k):
    window_sum = sum(nums[:k])
    print(f"Initial window sum: {window_sum}")
    max_sum = window_sum

    for i in range(k, len(nums)):
        window_sum += nums[i] - nums[i-k]
        max_sum = max(max_sum, window_sum)
        print(f"Window sum at index {i}: {window_sum}")

    return max_sum


nums = [2, 1, 5, 1, 3, 2]
print(max_sum_subarray(nums, 4))
```

```python
## Frequency Counter
from collections import Counter

def top_k_elements(nums, k):
    freq = Counter(nums)
    return [num for num, count in freq.most_common(k)]


nums = [1, 1, 1, 2, 2, 3]
print(top_k_elements(nums, 2))
```

```python
## Flatten Dict
def flatten_dict(data, parent_key="", sep="."):
    result = {}

    for key, value in data.items():
        new_key = f"{parent_key}{sep}{key}" if parent_key else key

        if isinstance(value, dict):
            result.update(flatten_dict(value, new_key, sep))
        else:
            result[new_key] = value

    return result


data = {
    "name": "Suraj",
    "address": {
        "city": "Pune",
        "location": {
            "area": "Chinchwad",
            "pincode": 411033
        }
    }
}

print(flatten_dict(data))
```

```python
## Aggregate by department
from collections import defaultdict

def aggregate_by_department(employees):
    result = defaultdict(lambda: {"headcount": 0, "names": []})

    for employee in employees:
        department = employee["department"]
        result[department]["headcount"] += 1
        result[department]["names"].append(employee["name"])

    return dict(result)


employees = [
    {"name": "John", "department": "IT"},
    {"name": "Alice", "department": "HR"},
    {"name": "Bob", "department": "IT"},
    {"name": "David", "department": "Finance"},
    {"name": "Emma", "department": "HR"}
]

print(aggregate_by_department(employees))
```
