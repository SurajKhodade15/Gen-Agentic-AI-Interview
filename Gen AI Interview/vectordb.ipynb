# Vector Databases - Top 30 Interview Questions & Answers

---

## Q1. What is a Vector Database?

### Answer

A Vector Database is a specialized database designed to store, index, and search high-dimensional vector embeddings efficiently.

Unlike traditional databases that search exact values, vector databases search based on **semantic similarity**.

**Examples:**
- FAISS
- ChromaDB
- Pinecone
- Weaviate
- Qdrant
- Milvus

---

## Q2. Why do we need a Vector Database?

### Answer

Embedding models convert text into vectors, but millions of vectors cannot be searched efficiently using SQL databases.

Vector databases provide:

- Fast similarity search
- Scalable storage
- Approximate Nearest Neighbor (ANN) search
- Metadata filtering
- Top-k retrieval

---

## Q3. What is stored inside a Vector Database?

### Answer

A vector database typically stores:

- Embedding vector
- Original document/chunk
- Metadata
- Unique ID

Example:

```
ID: 101

Embedding:
[0.23, -0.81, ...]

Document:
"RAG combines retrieval with generation."

Metadata:
{
    source:"pdf",
    page:12
}
```

---

## Q4. Traditional Database vs Vector Database

### Answer

| Traditional DB | Vector DB |
|----------------|-----------|
| Stores rows | Stores embeddings |
| Exact matching | Semantic matching |
| SQL search | Similarity search |
| B-Tree Index | ANN Index |
| Structured data | Unstructured data |

---

## Q5. What is Similarity Search?

### Answer

Similarity search finds vectors that are closest to the query vector.

Instead of matching exact text, it retrieves semantically similar documents.

Example:

Query:
```
AI Engineer Jobs
```

Returns:

```
Machine Learning Engineer

Generative AI Engineer

LLM Developer
```

---

## Q6. What is Semantic Search?

### Answer

Semantic Search retrieves documents based on meaning rather than exact keywords.

Example:

Search:
```
Doctor
```

Can return

```
Physician
Medical Practitioner
Healthcare Professional
```

---

## Q7. What is Top-k Retrieval?

### Answer

Top-k retrieval returns the **k most similar vectors**.

Example:

```
Top-3

↓

Document A

Document B

Document C
```

These documents are passed to the LLM in a RAG pipeline.

---

## Q8. What is Nearest Neighbor Search?

### Answer

Nearest Neighbor Search finds vectors closest to the query vector.

The "nearest" vectors represent the most semantically similar documents.

---

## Q9. What is Approximate Nearest Neighbor (ANN) Search?

### Answer

ANN is a fast search technique that returns approximately nearest vectors instead of exact ones.

Benefits:

- Extremely fast
- Scalable
- Slight trade-off in accuracy
- Used in production systems

---

## Q10. Why is ANN preferred over Exact Search?

### Answer

Exact search compares the query against every vector.

For millions of vectors, this becomes too slow.

ANN dramatically improves search speed while maintaining high accuracy.

---

## Q11. What is Cosine Similarity?

### Answer

Cosine similarity measures the angle between two vectors.

Range:

- 1 → Identical
- 0 → Unrelated
- -1 → Opposite

It is the most common similarity metric used in vector databases.

---

## Q12. What is Euclidean Distance?

### Answer

Euclidean distance measures the straight-line distance between two vectors.

Smaller distance indicates greater similarity.

It is less commonly used for text embeddings.

---

## Q13. What is Dot Product Similarity?

### Answer

Dot product measures similarity using both vector direction and magnitude.

Some embedding models are optimized specifically for dot product search.

---

## Q14. What is Metadata?

### Answer

Metadata is additional information stored with embeddings.

Example:

```
{
    file:"Resume.pdf",
    page:5,
    author:"Suraj",
    department:"AI"
}
```

---

## Q15. Why is Metadata important?

### Answer

Metadata enables filtering.

Example:

Retrieve only:

- PDFs
- HR documents
- Finance documents
- Page > 20

without searching the entire dataset.

---

## Q16. What is Metadata Filtering?

### Answer

Metadata filtering limits search results using metadata before or after similarity search.

Example:

```
Department = HR

AND

Similarity > 0.8
```

---

## Q17. What is Indexing in a Vector Database?

### Answer

Indexing organizes vectors for fast retrieval.

Without indexes, every vector would need to be compared, making searches inefficient.

---

## Q18. What is HNSW?

### Answer

HNSW (Hierarchical Navigable Small World) is an ANN indexing algorithm.

Advantages:

- Very fast
- High accuracy
- Low latency
- Widely used in production

---

## Q19. What is IVF?

### Answer

IVF (Inverted File Index) groups vectors into clusters.

During search, only relevant clusters are searched, reducing search time.

---

## Q20. What is Product Quantization (PQ)?

### Answer

PQ compresses vectors to reduce memory usage while maintaining efficient similarity search.

Useful for very large datasets.

---

## Q21. What is FAISS?

### Answer

FAISS (Facebook AI Similarity Search) is an open-source vector search library developed by Meta.

Features:

- Local storage
- High performance
- GPU support
- No managed cloud service

Best suited for local deployments.

---

ChromaDB is an open-source vector database designed for LLM and RAG applications.

Features:

- Lightweight
- Persistent storage
- Popular with LangChain

## Q28. How does a Vector Database work in RAG?

### Answer

```
Documents
      ↓
Chunking
      ↓
Embeddings
      ↓
Vector Database
      ↓
Similarity Search
      ↓
Top-k Chunks
      ↓
LLM
      ↓
Final Answer
```

The vector database retrieves the most relevant chunks for the LLM.

---

## Q29. What are common use cases of Vector Databases?

### Answer

- Retrieval-Augmented Generation (RAG)
- Semantic Search
- Chatbots
- Recommendation Systems
- Image Search
- Similarity Search
- Duplicate Detection
- Personalized Search

---

## Q30. How do you choose a Vector Database?

### Answer

Consider:

- Dataset size
- Search latency
- Scalability
- Cloud vs Local deployment
- Metadata filtering support
- Cost
- Integration with LangChain/LlamaIndex

Examples:

- Development → ChromaDB
- Local high-performance → FAISS
- Production cloud → Pinecone
- Open-source production → Qdrant or Weaviate

---

# ⭐ Interview Tips

- Embeddings are stored inside a Vector Database.
- Vector Databases perform semantic search, not keyword search.
- ANN makes large-scale similarity search practical.
- Metadata filtering improves retrieval precision.
- FAISS is a library, while Pinecone, Weaviate, ChromaDB, and Qdrant are vector database solutions.
- Every production RAG pipeline relies on a vector database for efficient retrieval.

## Q9. What is stored in a Qdrant collection?

### Answer

Each record typically contains:

- Unique ID
- Embedding vector
- Metadata (payload)
- Original text or document reference

Example:

[0.31, -0.18, ...]

  "page": 5,
  "department": "HR"
}

## Q10. What is a Collection in Qdrant?

### Answer

A collection is similar to a table in a relational database.

It stores vectors that share:

- Same embedding dimension
- Same distance metric
- Same indexing configuration

Example:

employees_collection

documents_collection

policies_collection

## Q14. How do you insert data into Qdrant?

### Answer

The steps are:

1. Generate embeddings.
2. Create a collection.
3. Upsert vectors with IDs and payload.
4. Build indexes automatically.
5. Perform similarity search.

## Q15. If your project grows from 10,000 to 10 million vectors, how does Qdrant handle it?

### Interview Answer

Qdrant is designed to scale by using efficient ANN indexing (HNSW), memory optimization, and support for distributed deployments. As the dataset grows, we can tune indexing parameters, shard collections, and deploy multiple nodes to maintain low-latency search.
