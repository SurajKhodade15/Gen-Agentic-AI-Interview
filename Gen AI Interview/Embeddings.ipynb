# Embeddings - Top 30 Interview Questions & Answers

---

## Q1. What are Embeddings?

### Answer

Embeddings are dense numerical vector representations of text, images, or other data that capture their semantic meaning.

Instead of representing text as words, embeddings convert text into vectors of floating-point numbers.

Example:

```
Dog → [0.24, -0.81, 0.52, ...]
Cat → [0.26, -0.79, 0.48, ...]
```

Since Dog and Cat have similar meanings, their vectors are close in vector space.

---

## Q2. Why do we need Embeddings?

### Answer

LLMs cannot understand raw text.

Embeddings convert text into numerical vectors that preserve semantic meaning, allowing machines to compare, search, and retrieve similar information.

Applications:
- Semantic Search
- RAG
- Recommendation Systems
- Clustering
- Similarity Search

---

## Q3. What is an Embedding Vector?

### Answer

An embedding vector is a list of floating-point numbers representing the meaning of a piece of text.

Example:

```
"The capital of India is Delhi"

↓

[0.34, -0.71, 0.29, ...]
```

## Q4. What is the difference between Tokens and Embeddings?

### Answer

| Tokens | Embeddings |
|---------|------------|
| Pieces of text | Numerical vectors |
| Human-readable | Machine-readable |
| Produced by tokenizer | Produced by embedding model |
| Input to embedding model | Used for similarity search |

## Q6. What is Semantic Similarity?

### Answer

Semantic similarity measures how similar two pieces of text are in meaning rather than exact words.

Example:

```
Doctor
Physician
```

These words have different spellings but similar meanings, so their embeddings are close.

---

## Q7. What is Vector Space?

### Answer

Vector space is a mathematical space where embeddings are represented as points.

Semantically similar embeddings appear closer together.

---

## Q8. Why are similar meanings located close together?

### Answer

Embedding models are trained so that semantically related texts have similar vector representations.

Example:

```
Dog  -------- Cat

Apple -------- Orange

Dog ---------- Car
```

Dog and Cat are closer than Dog and Car.

---

## Q9. What is Cosine Similarity?

### Answer

Cosine similarity measures the angle between two embedding vectors.

Range:

- 1 → Exactly similar
- 0 → Unrelated
- -1 → Opposite direction

It is the most commonly used similarity metric in RAG systems.

---

## Q10. Why is Cosine Similarity preferred?

### Answer

Because it measures direction instead of magnitude.

Two vectors with different lengths but similar meanings still have a high cosine similarity.

---

## Q11. Difference between Cosine Similarity and Euclidean Distance?

### Answer

| Cosine Similarity | Euclidean Distance |
|-------------------|--------------------|
| Measures angle | Measures distance |
| Ignores magnitude | Uses magnitude |
| Preferred for text | Preferred for geometry |

---

## Q12. What is Dot Product Similarity?

### Answer

Dot product measures similarity using both direction and magnitude.

Some vector databases support dot product instead of cosine similarity.

---

## Q13. What is Embedding Dimension?

### Answer

Embedding dimension is the number of values in an embedding vector.

Examples:

- 384
- 768
- 1024
- 1536
- 3072

Higher dimensions usually capture richer semantic information but require more storage.

---

## Q14. Does a higher embedding dimension always mean better quality?

### Answer

No.

Higher dimensions may improve representation but also increase:

- Storage
- Memory usage
- Search latency
- Cost

Choose the dimension based on your application.

---

## Q15. What are Dense Embeddings?

### Answer

Dense embeddings are vectors where most values are non-zero.

Modern embedding models produce dense embeddings.

---

## Q16. What are Sparse Embeddings?

### Answer

Sparse embeddings contain mostly zero values.

Example:

TF-IDF vectors are sparse embeddings.

---

## Q17. Dense vs Sparse Embeddings?

### Answer

| Dense | Sparse |
|--------|---------|
| Small vectors | Large vectors |
| Mostly non-zero | Mostly zeros |
| Capture semantics | Capture keywords |
| Used in LLMs | Used in traditional NLP |

## Q19. What is Sentence Embedding?

### Answer

Sentence embedding represents an entire sentence as a single vector.

Example:

```
"I love AI."

↓

One vector representing the whole sentence.
```

---

## Q20. What is Document Embedding?

### Answer

A document embedding represents an entire document or chunk as one vector.

Used in RAG and semantic search.

---

## Q21. Why don't we embed an entire book as one vector?

### Answer

Because:

- Too much information is lost.
- Retrieval quality decreases.
- Context becomes less precise.

Instead, books are split into chunks before embedding.

---

## Q22. Why are Embeddings important in RAG?

### Answer

Embeddings enable semantic search.

Workflow:

```
Documents
      ↓
Chunking
      ↓
Embeddings
      ↓
Vector Database
      ↓
Similarity Search
      ↓
Relevant Chunks
      ↓
LLM
```

Without embeddings, RAG cannot retrieve semantically relevant documents.

---

## Q24. Can two different sentences have similar embeddings?

### Answer

Yes.

Example:

```
How are you?

How do you feel today?
```

Different words, similar meaning → similar embeddings.

---

## Q25. Can the same word have different embeddings?

### Answer

Yes.

Modern embedding models generate contextual embeddings.

Example:

```
Apple (fruit)

Apple (company)
```

Different contexts produce different embeddings.

---

## Q26. What are contextual embeddings?

### Answer

Contextual embeddings change depending on surrounding words.

Example:

```
Bank of river

Bank account
```

The word "Bank" has different meanings, so it gets different embeddings.

### Answer

```
Raw Text
     ↓
Tokenizer
     ↓
Tokens
     ↓
Embedding Model
     ↓
Dense Vector
     ↓
Vector Database
     ↓
Similarity Search
     ↓
Top-k Results
     ↓
LLM
     ↓
Final Response
```
