# 1. Why FastAPI instead of Flask?

### Interview Answer

- FastAPI supports **async programming** natively.
- Automatic **Swagger/OpenAPI** documentation.
- Built-in **request validation** using Pydantic.
- Better performance (built on Starlette + Uvicorn).
- Type hint support.

### Example

```python
from fastapi import FastAPI

app = FastAPI()

@app.get("/")
def home():
    return {"message": "Hello"}
```

# 2. Async API

### Interview Answer

Use `async` for **I/O-bound operations** like database queries, API calls, and LLM requests.

### Example

```python
from fastapi import FastAPI
import asyncio

app = FastAPI()

@app.get("/chat")
async def chat():
    await asyncio.sleep(2)
    return {"message": "Done"}
```

---

# 3. Dependency Injection

### Interview Answer

Dependency Injection (DI) allows reusable logic (authentication, DB connections, validation) to be injected into endpoints.

### Example

```python
from fastapi import Depends

def get_user():
    return "Suraj"

@app.get("/profile")
def profile(user=Depends(get_user)):
    return {"user": user}
```

# 4. Middleware

### Interview Answer

Middleware executes **before and after** every request.

Used for:
- Logging
- Authentication
- CORS
- Request timing

### Example

```python
@app.middleware("http")
async def log_request(request, call_next):
    print("Before")
    response = await call_next(request)
    print("After")
    return response
```

# 5. JWT Authentication

### Interview Answer

JWT (JSON Web Token) is used for **stateless authentication**.

Flow:

```text
Login
↓

JWT Token

↓

Client stores token

↓

Authorization: Bearer <token>

↓

Server validates token
```

# 7. Background Tasks

### Interview Answer

Used to execute tasks **after sending the response**.

Examples:
- Send email
- Save logs
- Generate reports

### Example

```python
from fastapi import BackgroundTasks

def send_email():
    print("Email Sent")

@app.get("/")
def home(background_tasks: BackgroundTasks):
    background_tasks.add_task(send_email)
    return {"message": "Processing"}
```

---

# 8. Streaming Response

### Interview Answer

Streaming sends data **chunk by chunk** instead of waiting for the complete response.

Used for:
- ChatGPT responses
- File downloads
- LLM token streaming

### Example

```python
from fastapi.responses import StreamingResponse

def data():
    yield "Hello "
    yield "World"

@app.get("/stream")
def stream():
    return StreamingResponse(data())
```

# 9. WebSockets

### Interview Answer

WebSockets provide **two-way real-time communication**.

Used for:
- Chat applications
- Notifications
- Live dashboards
- AI streaming

### Example

```python
from fastapi import WebSocket

@app.websocket("/ws")
async def websocket(ws: WebSocket):
    await ws.accept()
    await ws.send_text("Hello")
```

# 10. Docker Deployment

### Interview Answer

Docker packages the application and its dependencies into a container, ensuring it runs consistently across environments.

### Dockerfile

```dockerfile
FROM python:3.12

WORKDIR /app

COPY . .

RUN pip install -r requirements.txt

CMD ["uvicorn", "main:app", "--host", "0.0.0.0", "--port", "8000"]
```

Run

```bash
docker build -t fastapi-app .
docker run -p 8000:8000 fastapi-app
```

# REST API from Memory

```python
from fastapi import FastAPI

app = FastAPI()

@app.get("/")
def home():
    return {"message": "Hello"}

@app.get("/users/{id}")
def get_user(id: int):
    return {"id": id}

@app.post("/users")
def create_user(user: dict):
    return user

@app.put("/users/{id}")
def update_user(id: int, user: dict):
    return {"id": id, "user": user}

@app.delete("/users/{id}")
def delete_user(id: int):
    return {"message": "Deleted"}
```
