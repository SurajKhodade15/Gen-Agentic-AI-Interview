# 1. Tell me about your GenAI project ⭐⭐⭐⭐⭐

### Interview Answer

> One of my recent projects was **AutoShift**, an enterprise Agentic AI platform for workforce shift automation. The business problem was that operations teams were manually processing hundreds of shift requests received through emails, which was time-consuming and error-prone.
>
> I designed an end-to-end Agentic AI workflow using **Python, LangGraph, AWS Bedrock Claude Sonnet, FastAPI, Qdrant, RabbitMQ, Docker, and LangSmith**.
>
> The application extracts shift information from emails, validates employee qualifications, resolves locations using semantic search, and invokes downstream REST APIs to create or update shifts automatically.
>
> When the AI confidence is low, the workflow enters a Human-in-the-Loop stage where an operations user reviews and corrects the output. Those corrections are stored and reused to improve future predictions.
>
> We also implemented LangSmith tracing, RabbitMQ-based event processing, Docker deployment, and production monitoring. Overall, the solution reduced manual effort by approximately **65%** while significantly improving operational accuracy.

# 2. Biggest Production Issue ⭐⭐⭐⭐⭐

## Scenario

During production rollout,

Operations reported that

```text
Some shift requests were being created for the wrong location.
```

---

### Root Cause

The RAG retriever was retrieving multiple locations with similar names.

Example

```text
New York Office

New York Warehouse

New York Remote Hub
```

LLM selected the wrong one.

---

### How I Solved It

Instead of only vector search,

I implemented

- Metadata filtering
- Hybrid Search
- Reranking
- Confidence score
- Human-in-the-Loop approval below 85% confidence

We also logged every retrieval using LangSmith.

---

### Result

- Wrong shift creation reduced by around **90%**
- Hallucinations significantly reduced
- Production accuracy improved
- Support tickets dropped

# 3. How did you reduce hallucinations? ⭐⭐⭐⭐⭐

### Interview Answer

I don't rely on prompt engineering alone.

I use multiple techniques:

- Hybrid Search (BM25 + Vector)
- Semantic Chunking
- Parent-Child Retrieval
- Query Rewriting
- Context Compression
- Reranking
- Low temperature (0–0.2)
- Prompt: *Answer only from provided context.*
- Human-in-the-Loop for low-confidence responses
- Continuous evaluation using RAGAS and DeepEval

This reduced hallucinations considerably in our HR assistant.
