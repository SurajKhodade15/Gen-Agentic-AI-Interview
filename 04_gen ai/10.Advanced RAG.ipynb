# What is RAG?

**RAG (Retrieval-Augmented Generation)** combines an LLM with external knowledge.

Instead of relying only on the model's training data, it retrieves relevant documents before generating an answer.

---

## Basic RAG Flow

```text
User Question
      │
      ▼
Embedding
      │
      ▼
Vector Database
      │
 Top-K Chunks
      │
      ▼
LLM
      │
      ▼
Answer
```

---

# What is Advanced RAG?

Advanced RAG improves the **accuracy**, **relevance**, and **quality** of retrieved information.

Instead of only:

```
Question
↓

Vector Search

↓

LLM
```

It performs multiple optimization steps.

---

# Production Advanced RAG Flow

```text
User Question
      │
      ▼
Query Rewriting
      │
      ▼
Embedding
      │
      ▼
Hybrid Search
(Vector + Keyword)
      │
      ▼
Metadata Filtering
      │
      ▼
Top 20 Chunks
      │
      ▼
Reranker
      │
      ▼
Top 5 Chunks
      │
      ▼
Context Compression
      │
      ▼
LLM
      │
      ▼
Grounded Answer
```

# Common Interview Questions

### Q1. Why Hybrid Search?

Vector search may miss exact keywords.

Keyword search may miss semantic meaning.

Together

↓

Best accuracy.

---

### Q2. Why Reranking?

Initial vector search

is approximate.

Reranking

improves relevance before sending context to the LLM.

---

### Q3. Why Context Compression?

LLMs have context window limits.

Compression reduces:

- Cost
- Latency
- Noise

---

### Q4. How do you reduce hallucination?

- Better retrieval
- Prompt engineering
- Guardrails
- Reranking
- Metadata filtering
- Ground responses in retrieved context
- Return "I don't know" when context is insufficient

---

### Q5. Difference between RAG and Fine-Tuning?

| RAG | Fine-Tuning |
|------|-------------|
| Uses external knowledge | Changes model weights |
| Easy to update | Requires retraining |
| Ideal for frequently changing data | Ideal for changing model behavior or style |
| Lower operational cost | Higher training cost |

# EPAM Interview Answer (2–3 Minutes)

> "Basic RAG retrieves documents from a vector database and passes them to the LLM. In production, I prefer an Advanced RAG architecture. First, I rewrite the user's query if needed to improve retrieval. Then I perform hybrid search using vector similarity and keyword search, followed by metadata filtering to narrow the search space. I retrieve the top candidate chunks, rerank them with a cross-encoder, and optionally compress the context to reduce token usage. The retrieved context is passed to the LLM with prompts that instruct it to answer only from the provided information and acknowledge when the answer is unavailable. For complex enterprise workflows, I extend this into Agentic RAG, where an orchestrator can decide whether to retrieve documents, execute SQL queries, call external APIs, or use other tools before generating the final response. This architecture improves accuracy, reduces hallucinations, lowers latency and cost, and provides better scalability for production systems."
