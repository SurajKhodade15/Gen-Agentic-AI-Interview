# 1. What is a Model?

A **Model** is the component responsible for generating responses based on the input provided.

In LangChain, the model acts as the "brain" of your AI application.

```text
User
  │
  ▼
Prompt
  │
  ▼
Model
  │
  ▼
Generated Response
```

Without a model, LangChain cannot generate any output.

---

# 2. Types of Models in LangChain

LangChain mainly works with two types of models.

## 1. LLM (Legacy)

Input

```
String
```

Output

```
String
```

Example:

```python
response = llm.invoke("What is AI?")
```

---

## 2. Chat Model (Modern)

Input

```
Messages
```

Output

```
AIMessage
```

Example:

```python
response = chat.invoke("What is AI?")
```

Almost all modern LLMs (GPT-4o, Claude, Gemini, Llama 3, etc.) are exposed as **chat models**.

# 3. LLM vs Chat Model

| Feature | LLM | Chat Model |
|----------|-----|------------|
| Input | Plain text | Messages |
| Output | Text | AIMessage |
| Roles | ❌ | ✅ (System, Human, AI) |

> **Interview Tip:** In LangChain v1.x, you'll almost always use **Chat Models**.

# 6. Creating Your First Chat Model

```python
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(
    model="gpt-4o-mini",
    temperature=0
)
```

### Explanation

- `ChatOpenAI` → OpenAI wrapper
- `model` → Which LLM to use
- `temperature` → Controls randomness

---

# 7. Invoking the Model

```python
response = llm.invoke("What is Artificial Intelligence?")

print(response.content)
```

Output:

```
Artificial Intelligence (AI) is the simulation...
```

# 8. What Does `invoke()` Return?

It returns an `AIMessage` object.

```python
response = llm.invoke("Hello")

print(type(response))
```

Output:

```python
AIMessage
```

To access the actual text:

```python
print(response.content)
```

# 9. Model Parameters

## temperature

Controls randomness.

```python
ChatOpenAI(
    model="gpt-4o-mini",
    temperature=0
)
```

| Temperature | Behavior |
|-------------|----------|
| 0.0 | Deterministic |
| 0.3 | Slight variation |
| 0.7 | Creative |
| 1.0 | Highly creative |

### Interview Tip

Use **temperature = 0** for:

- SQL
- RAG
- Agents
- Coding
- Document Q&A

Use higher values for creative writing.

---

## max_tokens

Limits the maximum output length.

```python
ChatOpenAI(
    model="gpt-4o-mini",
    max_tokens=500
)
```

---

## top_p

Controls token sampling.

Usually keep the default unless you have a specific need.

---

## frequency_penalty

Reduces repeated words.

---

## presence_penalty

Encourages introducing new topics.

# 10. Batch Processing

Instead of one request:

```python
response = llm.invoke("Hello")
```

Process multiple prompts together:

```python
responses = llm.batch([
    "What is AI?",
    "What is ML?",
    "What is NLP?"
])

for res in responses:
    print(res.content)
```

Useful for bulk inference.

---

# 11. Streaming

Normally:

```
Wait

↓

Entire response
```

With streaming:

```
Token

↓

Token

↓

Token

↓

...
```

Example:

```python
for chunk in llm.stream("Explain AI"):
    print(chunk.content, end="")
```

Streaming improves user experience for long responses.

---

# 12. Asynchronous Invocation

Instead of blocking:

```python
response = llm.invoke(prompt)
```

Use async:

```python
response = await llm.ainvoke(prompt)
```

Useful in FastAPI and high-concurrency applications.

# 17. Common Interview Questions

## Q1. What is the difference between an LLM and a Chat Model?

**Answer:**

An LLM accepts plain text as input and returns plain text. A Chat Model works with structured messages (System, Human, AI), making it suitable for conversations, tool calling, and agent workflows.

---

## Q2. Why does LangChain recommend Chat Models?

**Answer:**

Because modern providers expose conversational APIs. Chat Models support roles, conversation history, tool calling, and multi-turn interactions.

---

## Q3. What does `invoke()` return?

**Answer:**

An `AIMessage` object containing metadata and the generated response. The actual text is accessed using `.content`.

---

## Q4. When should you use `batch()`?

**Answer:**

When processing multiple independent prompts efficiently in a single operation.

---

## Q5. When should you use `stream()`?

**Answer:**

For long responses where showing tokens incrementally improves user experience.

---

## Q6. What is the purpose of `temperature`?

**Answer:**

It controls the randomness of model outputs. Lower values produce more deterministic responses, while higher values encourage creativity.

---

# 18. Common Mistakes

❌ Using high temperature for SQL generation.

❌ Printing the entire `AIMessage` object instead of `.content`.

❌ Using synchronous calls in highly concurrent APIs.

❌ Assuming all providers have identical model names.
