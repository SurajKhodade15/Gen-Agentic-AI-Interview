# Part 1 : Vector Stores

# 2. What is a Vector Store?

A Vector Store is a specialized database designed to store embeddings and efficiently perform similarity searches.

Unlike SQL databases, it retrieves data based on **meaning**, not exact text matches.

---

# Interview Definition ⭐⭐⭐⭐⭐

> **A Vector Store is a database optimized for storing embedding vectors and performing fast semantic similarity searches using nearest-neighbor algorithms.**

---

# Part 2 : Retrievers

# 12. What is a Retriever?

A Retriever is a LangChain abstraction that fetches the most relevant documents for a query.

It hides the implementation details of the underlying storage.

---

# Interview Definition ⭐⭐⭐⭐⭐

> **A Retriever accepts a query and returns the most relevant documents, regardless of whether they come from a vector database, keyword search, hybrid search, or another retrieval system.**

---

# 13. Why Not Query the Vector Store Directly?

You can.

But a Retriever provides

- Common interface
- Reusable API
- Search configuration
- Metadata filtering
- Future flexibility

---

# 24. Interview Questions

## Q1. What is a Vector Store?

**Answer:**

A Vector Store is a specialized database that stores embedding vectors and performs efficient semantic similarity searches.

---

## Q2. What is a Retriever?

**Answer:**

A Retriever is an abstraction that accepts a query and returns the most relevant documents from a retrieval system such as a vector database.

---

## Q3. What is Top-K Retrieval?

**Answer:**

Top-K retrieval returns only the K most relevant chunks instead of the entire document collection, reducing cost and improving response quality.

---

## Q4. Why use metadata filtering?

**Answer:**

Metadata filtering restricts the search space, improving retrieval accuracy and reducing unnecessary results.

---

## Q5. What is MMR?

**Answer:**

Maximum Marginal Relevance balances relevance with diversity, preventing retrieval of multiple nearly identical chunks.

---

## Q6. Difference between Vector Store and Retriever?

**Answer:**

The Vector Store is responsible for storing and indexing embeddings, while the Retriever provides a standardized interface to fetch relevant documents from that storage.

# 26. Real Interview Scenario (Based on Qdrant)

**Question:**

> Your RAG chatbot stores 500,000 document chunks in Qdrant. Users report that responses are becoming slower and occasionally irrelevant. How would you improve retrieval quality?

**Answer:**

I would first inspect the retrieval stage rather than the LLM. Specifically, I would:
1. Tune the `k` value to avoid sending too many or too few chunks.
2. Enable metadata filtering to reduce the search space (for example, by department or document type).
3. Use MMR if duplicate chunks dominate the results.
4. Verify that the same embedding model is used for indexing and querying.
5. Remove duplicate or low-quality chunks and adjust chunk size/overlap if necessary.
6. Evaluate retrieval metrics such as Recall@K and Precision@K before changing the LLM.

# 30-Second Interview Answer

> **A Vector Store is a specialized database for storing embedding vectors and performing semantic similarity search. A Retriever is a higher-level abstraction that queries the Vector Store and returns the most relevant documents for a user's question. In a RAG system, documents are embedded and stored in the Vector Store during indexing, while the Retriever performs Top-K semantic search during inference and supplies relevant context to the LLM for answer generation.**

---

# Key Takeaway

> **The LLM can only answer based on the context it receives. In a RAG system, retrieval quality—driven by the Vector Store and Retriever—is often the biggest factor determining answer quality. Improving retrieval usually has a greater impact than switching to a larger LLM.**
