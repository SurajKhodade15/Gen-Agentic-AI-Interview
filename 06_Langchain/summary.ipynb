# Agentic AI + LangChain Quick Revision Cheat Sheet

| Topic | One-Line Definition | Code Example (if applicable) |
|--------|----------------------|------------------------------|
| Agentic AI | AI systems that reason, plan, use tools, and take actions autonomously. | ❌ |
| AI Agent | LLM + Memory + Tools + Reasoning + Planning + Actions. | ❌ |
| LLM | Predicts the next token to generate human-like text. | `llm.invoke("Hello")` |
| Chatbot | Conversational interface built using an LLM. | ❌ |
| Workflow | Fixed sequence of predefined steps. | ❌ |
| Components of AI Agent | Brain (LLM), Memory, Tools, Planning, Actions, Environment. | ❌ |
| Agent Lifecycle | Perceive → Think → Plan → Act → Observe → Repeat. | ❌ |
| Agent Architecture | User → Agent → Tool → Observation → Response. | ❌ |
| Single Agent | One agent performs the complete task. | ❌ |
| Multi-Agent | Multiple specialized agents collaborate to solve tasks. | ❌ |
| ReAct | Reason first, then choose and execute a tool. | ❌ |
| Chain of Thought (CoT) | Step-by-step reasoning before answering. | ❌ |
| Reflexion | Agent learns from previous mistakes and retries. | ❌ |
| Plan-and-Execute | Planner creates tasks; executor completes them. | ❌ |
| Tree of Thoughts (ToT) | Explore multiple reasoning paths before selecting the best. | ❌ |
| Graph of Thoughts (GoT) | Reasoning represented as interconnected graphs instead of trees. | ❌ |
| Short-Term Memory | Stores recent conversation context. | ❌ |
| Long-Term Memory | Stores persistent user knowledge across sessions. | ❌ |
| Semantic Memory | Stores facts and knowledge. | ❌ |
| Episodic Memory | Stores previous interactions and experiences. | ❌ |
| Memory Management | Decide what to store, retrieve, summarize, or forget. | ❌ |
| Tools | External functions an LLM can call. | `@tool` |
| Function Calling | LLM generates structured tool invocation. | `llm.bind_tools()` |
| Tool Selection | LLM chooses the appropriate tool automatically. | ❌ |
| API Integration | Agents call external REST APIs. | `requests.get()` |
| Python Tool | Execute Python functions as tools. | `@tool` |
| SQL Tool | Query relational databases using SQL. | `SQLDatabaseToolkit` |
| Web Search Tool | Retrieve real-time information from the web. | `TavilySearchResults()` |
| File System Tool | Read/write files using tools. | `FileManagementToolkit()` |
| LangChain | Framework for building LLM applications. | `pip install langchain` |
| LangChain Architecture | Models + Prompts + Chains + Tools + Memory + Agents. | ❌ |
| Models | Interface for chat models and LLMs. | `ChatOpenAI()` |
| Prompt Templates | Dynamic prompt generation using variables. | `ChatPromptTemplate()` |
| Output Parsers | Convert model output into structured data. | `StrOutputParser()` |
| Messages | System, Human, AI, Tool message objects. | `HumanMessage()` |
| LCEL | Pipe (`|`) syntax to compose AI workflows. | `prompt \| llm \| parser` |
| Chains | Sequential execution of LangChain components. | `prompt \| llm` |
| Runnables | Standard interface supporting invoke, batch, stream. | `.invoke()` |
| @tool | Convert Python functions into AI tools. | `@tool` |
| Tool Calling | LLM decides when and how to invoke tools. | `llm.bind_tools()` |
| LangChain Agent | AI system that reasons and dynamically selects tools. | `create_agent()` |
| Agent Executor | Executes tool calls and manages the reasoning loop. | `AgentExecutor()` |
| Callbacks | Event hooks for monitoring execution. | `BaseCallbackHandler` |
| Memory (LangChain) | Maintains conversation context across requests. | `RunnableWithMessageHistory()` |
| Document Loader | Loads PDFs, DOCX, TXT, CSV, etc. into Document objects. | `PyPDFLoader()` |
| Text Splitter | Splits large documents into smaller chunks. | `RecursiveCharacterTextSplitter()` |
| Chunk Size | Maximum size of each text chunk. | `chunk_size=500` |
| Chunk Overlap | Shared text between consecutive chunks. | `chunk_overlap=100` |
| Embeddings | Convert text into semantic vectors. | `OpenAIEmbeddings()` |
| Vector Store | Stores embeddings for similarity search. | `QdrantVectorStore()` |
| Similarity Search | Finds semantically similar vectors. | `.similarity_search()` |
| Retriever | Retrieves relevant chunks from a vector store. | `.as_retriever()` |
| Top-K Retrieval | Returns the K most relevant chunks. | `k=5` |
| MMR | Balances relevance and diversity during retrieval. | `search_type="mmr"` |
| Metadata Filtering | Restricts retrieval using metadata. | `filter={...}` |
| RAG | Retrieve relevant documents before LLM generation. | ❌ |
| RAG Indexing | Load → Split → Embed → Store. | ❌ |
| RAG Retrieval | Embed Query → Retrieve → Prompt → LLM. | ❌ |
| Conversation History | Previous chat messages sent with each request. | `RunnableWithMessageHistory()` |
| Session ID | Identifies a user's conversation history. | `session_id="123"` |
| Structured Output | Return validated JSON or typed objects. | `llm.with_structured_output()` |
| Pydantic Schema | Defines expected structured output. | `class User(BaseModel)` |
| Streaming | Streams tokens while generating responses. | `llm.stream()` |
| Error Handling | Detect, recover, and report failures gracefully. | `try...except` |
| Retry | Retry failed operations automatically. | `max_retries=3` |
| Exponential Backoff | Increase wait time after each retry. | 1s → 2s → 4s |
| Timeout | Stop requests that take too long. | `timeout=30` |
| Fallback Model | Backup model if primary model fails. | ❌ |
| Circuit Breaker | Temporarily stop requests to failing services. | ❌ |
| Graceful Degradation | Return a safe response instead of crashing. | ❌ |
| LangSmith | Observability platform for LangChain apps. | `Client()` |
| Trace | Complete execution record of one request. | ❌ |
| Evaluation Dataset | Ground-truth dataset for testing AI quality. | ❌ |
| Token Monitoring | Track prompt, completion, and total tokens. | ❌ |
| End-to-End Agent | Production pipeline combining Agent + RAG + Tools + Memory + Monitoring. | `create_agent()` |

# End-to-End RAG Pipeline

```text
Documents
    ↓
Document Loader
    ↓
Text Splitter
    ↓
Embeddings
    ↓
Vector Store (Qdrant)
    ↓
Retriever
    ↓
Prompt Template
    ↓
LLM / Agent
    ↓
Structured Output
    ↓
Streaming
    ↓
User
    ↓
LangSmith Trace
```

---

# End-to-End Agent Flow

```text
User
   ↓
Conversation History
   ↓
Agent
   ↓
Reason
   ↓
Select Tool
   ↓
Execute Tool
   ↓
Observe Result
   ↓
Need More Tools?
   ↓
Final Answer
```
