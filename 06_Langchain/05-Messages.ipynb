# 1. Why Do We Need Messages?

Earlier LLMs accepted only plain text.

Example:

```text
Explain Artificial Intelligence.
```

Modern LLMs (GPT-4o, Claude, Gemini, Llama 3, etc.) are **chat-based**.

Instead of a single string, they receive a list of messages.

```text
[
    System Message,
    Human Message,
    AI Message,
    Human Message
]
```

This enables:

- Conversations
- Memory
- Tool Calling
- Better context handling

---

# Interview Definition ⭐⭐⭐⭐⭐

> **Messages are structured communication units used by chat models. Each message has a specific role, such as System, Human, AI, or Tool, allowing the model to understand instructions, conversation history, and external tool outputs.**

---

# 2. Message Architecture

```text
Conversation

│

├── System Message

├── Human Message

├── AI Message

├── Human Message

├── Tool Message

└── AI Message
```

Unlike plain text prompts, messages preserve the role of each participant.

# 4. System Message

The System Message tells the model **how it should behave**.

Example:

```
You are a Senior Python Instructor.

Always explain concepts with examples.
```

It is usually the **first message** in the conversation.

---

## Coding Example

```python
from langchain_core.messages import SystemMessage

system = SystemMessage(

    content="""
You are an AI Interviewer.
Answer professionally.
"""

)
```

# 14. Messages vs PromptTemplate

A very common interview question.

| PromptTemplate | Messages |
|----------------|----------|
| Creates prompts | Represents conversations |
| Before model call | Input to chat model |
| Generates text/messages | Stores structured dialogue |

They work together.

```text
PromptTemplate

↓

Messages

↓

Chat Model
```

# 15. Messages vs Memory

Another common interview question.

| Messages | Memory |
|-----------|--------|
| Current conversation | Stores conversation history |
| Temporary | Persistent or managed |
| Input to LLM | Source of previous messages |

Memory often **retrieves previous messages** and appends them before the next model call.

# 18. Best Practices

✅ Always use a System Message.

✅ Keep system prompts short and explicit.

✅ Pass conversation history when needed.

✅ Avoid conflicting system instructions.

✅ Prefer structured messages over concatenated strings.

---

# 19. Common Mistakes

❌ Passing one long string instead of structured messages.

❌ Forgetting the System Message.

❌ Creating AIMessage manually for model responses.

❌ Mixing user instructions into the System Message.

# 20. Interview Questions

## Q1. Why are Messages used in LangChain?

**Answer:**

Messages provide a structured representation of conversations, allowing chat models to distinguish between system instructions, user input, assistant responses, and tool outputs.

---

## Q2. What is the purpose of a System Message?

**Answer:**

A System Message defines the assistant's behavior, role, and overall instructions for the conversation.

---

## Q3. What does `invoke()` return?

**Answer:**

It returns an `AIMessage` object containing the model's response and metadata.

---

## Q4. What is a ToolMessage?

**Answer:**

A ToolMessage represents the output returned by an external tool and is added to the conversation before the model generates its final response.

---

## Q5. How are Messages related to Memory?

**Answer:**

Memory stores previous conversation history and supplies relevant messages back to the model during future interactions.

# 22. Complete Coding Example

```python
from langchain_openai import ChatOpenAI
from langchain_core.messages import (
    SystemMessage,
    HumanMessage
)

llm = ChatOpenAI(
    model="gpt-4o-mini",
    temperature=0
)

messages = [

    SystemMessage(
        content="You are an expert AI tutor."
    ),

    HumanMessage(
        content="Explain LangChain in simple words."
    )

]

response = llm.invoke(messages)

print(response.content)
```

# 23. Real-World Interview Scenario

**Question:**

> Why doesn't LangChain simply send one big prompt instead of using messages?

**Answer:**

Because modern chat models are trained on structured conversations. Messages preserve the role of each participant (System, Human, AI, Tool), enabling better instruction following, conversation history, tool calling, and multi-turn interactions. This structure also makes agent workflows more reliable and maintainable than concatenating everything into a single text prompt.

---

# 30-Second Interview Answer

> **Messages are the fundamental communication format used by modern chat models in LangChain. Instead of passing a single text prompt, applications send structured messages such as SystemMessage, HumanMessage, AIMessage, and ToolMessage. This enables role-based instructions, conversation history, tool integration, and multi-turn interactions, making chat-based AI applications more robust and scalable.**
