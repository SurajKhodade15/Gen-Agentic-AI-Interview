# 1. What are Callbacks?

A **Callback** is a mechanism that allows you to execute custom code whenever a specific event occurs during the execution of a LangChain application.

Think of it as an **event listener**.

Example:

```
LLM starts

↓

Trigger Callback

↓

LLM finishes

↓

Trigger Callback
```

---

# Interview Definition ⭐⭐⭐⭐⭐

> **Callbacks are event hooks that allow developers to monitor, log, trace, or customize the execution of LangChain components such as LLMs, chains, tools, retrievers, and agents.**

---

# 2. Why Do We Need Callbacks?

Suppose your application is:

```
Prompt

↓

LLM

↓

Tool

↓

Retriever

↓

Parser
```

How do you know:

- Which tool was called?
- How many tokens were used?
- How long did the LLM take?
- Which retriever returned documents?
- Which step failed?

Callbacks provide these answers.

# 9. Callback Handlers

A **Callback Handler** is a class that responds to callback events.

Example

```python
from langchain_core.callbacks import BaseCallbackHandler

class MyHandler(BaseCallbackHandler):

    def on_llm_start(
        self,
        serialized,
        prompts,
        **kwargs
    ):
        print("LLM Started")
```

Whenever the LLM starts, this method executes.

---

# 10. Custom Callback Example

```python
from langchain_core.callbacks import BaseCallbackHandler

class TokenLogger(BaseCallbackHandler):

    def on_llm_start(
        self,
        serialized,
        prompts,
        **kwargs
    ):
        print("Starting LLM")

    def on_llm_end(
        self,
        response,
        **kwargs
    ):
        print("Finished LLM")
```

---

# 11. Registering Callbacks

```python
llm.invoke(

    "Explain AI",

    config={

        "callbacks":[

            TokenLogger()

        ]

    }

)
```

Now the callback is executed automatically.

# 18. Best Practices

✅ Keep callback logic lightweight.

✅ Avoid blocking operations inside callbacks.

✅ Log errors separately.

✅ Use callbacks for observability, not business logic.

✅ Integrate with LangSmith or monitoring tools.

---

# 19. Common Mistakes

❌ Performing heavy database operations inside callbacks.

❌ Throwing exceptions from callback handlers.

❌ Mixing business logic with monitoring.

❌ Ignoring callback failures.

# 21. Interview Questions

## Q1. What are Callbacks?

**Answer:**

Callbacks are event-driven hooks that allow developers to observe, monitor, and customize the execution of LangChain components such as LLMs, tools, retrievers, chains, and agents.

---

## Q2. Why are Callbacks useful?

**Answer:**

They provide visibility into execution, enabling logging, debugging, monitoring, token tracking, latency measurement, and error handling.

---

## Q3. What is a Callback Handler?

**Answer:**

A Callback Handler is a class that implements callback methods like `on_llm_start()` or `on_tool_end()` and executes custom logic when those events occur.

---

## Q4. Are Callbacks the same as Logging?

**Answer:**

No. Callbacks are framework events, while logging is one possible action performed in response to those events.

---

## Q5. How does LangSmith use Callbacks?

**Answer:**

LangSmith captures callback events to build execution traces, visualize workflows, measure performance, and simplify debugging.

---

## Q6. Can Callbacks modify the LLM response?

**Answer:**

They are primarily intended for observation and monitoring rather than modifying execution results. Business logic should generally remain outside callback handlers.

# 23. Real Interview Scenario

**Question:**

> Your AI application is slow, and users complain about high latency. How can Callbacks help?

**Answer:**

I would use callback handlers to measure execution time for each stage, including LLM calls, retriever operations, and tool executions. By logging these events or sending them to LangSmith, I can identify the bottleneck—whether it is a slow model, an external API, or document retrieval—and optimize that specific component.

# 30-Second Interview Answer

> **Callbacks in LangChain are event-driven hooks that provide observability into AI application execution. They allow developers to monitor LLM calls, tool usage, retriever performance, agent actions, token usage, and errors. By implementing custom callback handlers or integrating with LangSmith, developers can debug, monitor, and optimize production AI systems without changing the core application logic.**

# Key Takeaway

> **Callbacks are the observability layer of LangChain. They don't perform the AI work—they tell you what happened, when it happened, and how well it performed, making them essential for debugging and production monitoring.**
