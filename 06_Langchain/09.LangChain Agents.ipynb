# 1. Why Do We Need Agents?

Consider the question:

```
What is the capital of France?
```

A simple LLM is enough.

---

Now consider:

```
Find today's weather in Pune and email it to my manager.
```

The AI needs to:

- Search weather
- Understand the result
- Send an email
- Return confirmation

A simple chain is not enough because it has a **fixed workflow**.

An Agent decides what to do dynamically.

---

# Interview Definition ⭐⭐⭐⭐⭐

> **An Agent is an AI system that can reason about a task, decide which tools to use, execute those tools, observe the results, and continue until it reaches a final answer.**

---

# 2. Chain vs Agent

| Chain | Agent |
|--------|--------|
| Fixed workflow | Dynamic workflow |
| Steps are predefined | Steps decided at runtime |
| No reasoning | Performs reasoning |
| Limited flexibility | Highly flexible |
| Best for predictable tasks | Best for open-ended tasks |

---

### Example

#### Chain

```text
Question

↓

Prompt

↓

LLM

↓

Answer
```

Always follows the same path.

---

#### Agent

```text
Question

↓

Reason

↓

Choose Tool

↓

Execute Tool

↓

Observe Result

↓

Need Another Tool?

↓

Yes → Repeat

↓

No

↓

Final Answer
```

# 3. Agent Architecture

```text
               User

                 │

                 ▼

          Agent (LLM Brain)

                 │

      ┌──────────┴──────────┐

      ▼                     ▼

   Reason              Choose Tool

      │                     │

      └──────────┬──────────┘

                 ▼

            Execute Tool

                 │

                 ▼

            Observe Result

                 │

        More Work Needed?

          │            │

         Yes          No

          │            │

          ▼            ▼

      Continue     Final Answer
```

# 5. The Agent Loop

The most important concept.

```text
Thought

↓

Action

↓

Observation

↓

Thought

↓

Action

↓

Observation

↓

Final Answer
```

This continues until the task is complete.

# 20. Interview Questions

## Q1. What is an Agent?

**Answer:**

An Agent is an AI system that reasons about a task, selects appropriate tools, executes them, observes the results, and repeats the process until it produces the final answer.

---

## Q2. What is the difference between a Chain and an Agent?

**Answer:**

A Chain follows a predefined sequence of steps, whereas an Agent dynamically decides which actions to take based on the current task and intermediate results.

---

## Q3. What is the Agent Loop?

**Answer:**

The Agent Loop consists of repeated cycles of reasoning, tool selection, tool execution, observation, and decision-making until the objective is achieved.

---

## Q4. Does every Agent use Tools?

**Answer:**

Most practical agents do because tools extend the model's capabilities, but an agent can still perform reasoning without tools for purely text-based tasks.

---

## Q5. When should you use an Agent instead of a Chain?

**Answer:**

Use an Agent when the sequence of actions cannot be predetermined and depends on the user's request or intermediate results.

---

## Q6. Why are Agents more expensive than Chains?

**Answer:**

Agents often require multiple reasoning steps, multiple LLM calls, and several tool executions, increasing latency and cost.

# 22. Real Interview Scenario

**Question:**

> A user asks:
>
> "Search the latest AI news, summarize it, and email it to me."

How would a LangChain Agent solve this?

**Answer:**

1. Understand the user's goal.
2. Decide to use the Search Tool.
3. Execute the Search Tool.
4. Observe the retrieved news.
5. Use the LLM to summarize it.
6. Decide to call the Email Tool.
7. Send the email.
8. Return a confirmation to the user.

The workflow is generated dynamically rather than hardcoded.

# 30-Second Interview Answer

> **A LangChain Agent is an AI system that combines an LLM with external tools to solve complex tasks. Unlike a fixed Chain, an Agent reasons about the problem, selects tools at runtime, executes them, observes the results, and continues this Reason → Act → Observe loop until it reaches the final answer. This makes Agents ideal for dynamic, multi-step workflows such as web search, SQL querying, API integration, and automation.**

# Key Takeaway

> **An Agent is a reasoning engine, not just a text generator. It decides *what to do next* based on the current situation, making it the core building block for autonomous AI applications.**
