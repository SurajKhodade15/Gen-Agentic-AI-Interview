# 1. What is Conversation History?

Conversation History is the collection of previous messages exchanged between the user and the AI assistant.

Example

```
User:
Hi

Assistant:
Hello

User:
Explain LangChain

Assistant:
...

User:
Give another example.
```

Without history, the model doesn't know that "another example" refers to LangChain.

---

# Interview Definition ⭐⭐⭐⭐⭐

> **Conversation History is the ordered sequence of previous messages that is supplied to the LLM so it can generate context-aware responses across multiple turns of a conversation.**

# 3. How Conversation History Works

The LLM **does not store history**.

The application stores messages and sends them with every request.

```text
History

+

Current Question

↓

Prompt

↓

LLM

↓

Answer

↓

Update History
```

---

# 4. Message Structure

LangChain represents conversations using messages.

```text
SystemMessage

↓

HumanMessage

↓

AIMessage

↓

HumanMessage

↓

AIMessage
```

Each new message is appended to the history.

# 6. Conversation History vs Memory

One of the most common interview questions.

| Conversation History | Memory |
|----------------------|--------|
| Stores previous messages | Broader concept |
| Ordered conversation | Can include summaries, entities, vectors |
| Usually short-term | Can be short-term or long-term |
| Used for chat continuity | Used for personalization and context |

Think of it this way:

```
Conversation History

⊂

Memory
```

Conversation history is one type of memory.

---

# 7. Conversation History in LangChain

Modern LangChain uses **message history** instead of legacy memory classes.

The recommended approach is to manage chat history explicitly.

---

# 8. RunnableWithMessageHistory

LangChain provides a wrapper to automatically maintain chat history.

```python
from langchain_core.runnables.history import (
    RunnableWithMessageHistory
)
```

This wrapper:

- Retrieves previous messages
- Sends them to the model
- Saves new messages
- Associates them with a session

---

# 9. Basic Example

Suppose we already have a chain.

```python
chain = prompt | llm
```

Wrap it

```python
chain_with_history = RunnableWithMessageHistory(
    runnable=chain,
    get_session_history=get_history
)
```

Now the history is managed automatically.

# 12. Conversation History in RAG

Suppose

User

```
Explain vector databases.
```

Later

```
How are they different from SQL?
```

The retrieval system should understand that "they" refers to vector databases.

Flow

```text
History

+

Question

↓

Retriever

↓

Relevant Documents

↓

LLM
```

Without history, retrieval quality may decrease because the question is ambiguous.

---

# 13. Multi-Turn RAG

```text
User

↓

Conversation History

↓

Question Rewriting (Optional)

↓

Retriever

↓

Relevant Chunks

↓

Prompt

↓

LLM

↓

Answer
```

Many production systems rewrite follow-up questions into standalone queries before retrieval.

Example

```
How does it work?
```

↓

```
How does Retrieval-Augmented Generation work?
```

This improves retrieval accuracy.

# 17. Best Practices

✅ Store history per session.

✅ Limit history size.

✅ Summarize long conversations.

✅ Remove sensitive information when appropriate.

✅ Persist history if users should continue conversations later.

---

# 18. Common Mistakes

❌ Sharing history across users.

❌ Sending every message forever.

❌ Confusing history with RAG.

❌ Ignoring session management.

# 19. Interview Questions

## Q1. What is Conversation History?

**Answer:**

Conversation History is the ordered sequence of previous messages exchanged between the user and the assistant. It is included in subsequent requests to maintain conversational context.

---

## Q2. Does the LLM remember previous conversations?

**Answer:**

No. The application stores the conversation history and sends relevant messages with each request.

---

## Q3. What is RunnableWithMessageHistory?

**Answer:**

It is a LangChain wrapper that automatically manages chat history by loading previous messages, passing them to the runnable, and saving new messages after each interaction.

---

## Q4. Why is Session ID important?

**Answer:**

The Session ID uniquely identifies a conversation, ensuring each user's history is isolated and preventing conversations from mixing.

---

## Q5. Why rewrite follow-up questions in conversational RAG?

**Answer:**

Follow-up questions often contain pronouns or incomplete references. Rewriting them into standalone questions improves retrieval quality.

---

## Q6. Difference between Conversation History and Memory?

**Answer:**

Conversation History is the sequence of chat messages, while Memory is a broader concept that may also include summaries, entities, preferences, and long-term information.

# 21. Real Interview Scenario

**Question:**

> A user asks:
>
> "Explain vector databases."
>
> Then asks:
>
> "How are they different from SQL?"
>
> Your chatbot gives an incorrect answer because it searches only for "they". How would you fix this?

**Answer:**

I would use conversational RAG. Before retrieval, I would combine the conversation history with the latest question or rewrite the follow-up into a standalone query such as "How are vector databases different from SQL databases?". This gives the retriever enough context to fetch the correct documents, resulting in a more accurate response.

# 30-Second Interview Answer

> **Conversation History is the sequence of previous chat messages supplied to the LLM to maintain context across multiple interactions. Since LLMs are stateless, the application stores and retrieves this history, often using `RunnableWithMessageHistory` in LangChain. In production systems, history is maintained per session, summarized when necessary, and combined with retrieval to support conversational RAG.**

---

# Key Takeaway

> **Conversation History is the bridge between independent LLM calls. It allows users to have natural, multi-turn conversations by providing the model with the necessary context, while the application—not the LLM—remains responsible for storing and managing that history.**
