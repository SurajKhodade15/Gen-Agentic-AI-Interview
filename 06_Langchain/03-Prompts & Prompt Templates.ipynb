# 1. What is a Prompt?

A **Prompt** is the instruction given to an LLM to perform a task.

Example:

```
Explain Artificial Intelligence in simple words.
```

The quality of the prompt directly affects the quality of the response.

---

# 2. What is a Prompt Template?

A **Prompt Template** is a reusable prompt with placeholders (variables).

Instead of hardcoding values:

```
Explain Python.
Explain Java.
Explain C++.
```

Create one reusable template:

```
Explain {topic}.
```

Then replace `{topic}` dynamically.

---

# Interview Definition ⭐⭐⭐⭐⭐

> **A Prompt Template is a reusable prompt containing variables that are filled at runtime, allowing developers to generate dynamic prompts without rewriting the prompt text.**

---

# 3. Why Do We Need Prompt Templates?

Without Prompt Template

```python
prompt = "Explain Python"

prompt = "Explain Java"

prompt = "Explain SQL"

prompt = "Explain AI"
```

Repeated code.

---

With Prompt Template

```python
template = "Explain {topic}"
```

Runtime

```
topic = Python

↓

Explain Python
```

```
topic = Java

↓

Explain Java
```

Much cleaner.

# 5. PromptTemplate

Used for simple text prompts.

Example

```python
from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate.from_template(
    "Explain {topic} in simple words."
)

formatted_prompt = prompt.invoke(
    {"topic": "Machine Learning"}
)

print(formatted_prompt.text)
```

Output

```
Explain Machine Learning in simple words.
```

---

# 6. Multiple Variables

Templates can contain multiple placeholders.

```python
prompt = PromptTemplate.from_template(

"""
Explain {topic}
for a {level} student.
"""

)
```

Runtime

```python
formatted = prompt.invoke({

    "topic": "AI",
    "level": "beginner"

})
```

Output

```
Explain AI
for a beginner student.
```

# 7. ChatPromptTemplate

Modern chat models work with **messages**, not plain text.

Instead of generating one long string,

they generate structured messages.

```text
System Message

↓

Human Message

↓

AI Response
```

---

Example

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages(

    [

        ("system", "You are an AI tutor."),

        ("human", "{question}")

    ]

)
```

Runtime

```python
messages = prompt.invoke(

    {

        "question": "Explain AI"

    }

)
```

# 8. PromptTemplate vs ChatPromptTemplate

| PromptTemplate | ChatPromptTemplate |
|----------------|--------------------|
| Plain text | Chat messages |
| Legacy style | Modern style |
| One string | Multiple messages |
| Limited | Supports System/Human/AI roles |
| Less used | Recommended |

# 9. Message Roles

Modern LLMs understand different message types.

### System Message

Defines behavior.

Example

```
You are an expert Python teacher.
```

---

### Human Message

Actual user request.

Example

```
Explain decorators.
```

---

### AI Message

Previous assistant response.

Useful in conversations.

# 11. Partial Variables

Sometimes a variable never changes.

Instead of passing it every time,

store it inside the template.

Example

```python
prompt = PromptTemplate(

    template="""
You are a {role}.

Explain {topic}.
""",

    input_variables=["topic"],

    partial_variables={

        "role": "Senior AI Engineer"

    }

)
```

Only provide

```
topic
```

at runtime.

# 15. Best Practices

✅ Keep prompts reusable.

✅ Separate prompts from business logic.

✅ Use variables instead of string concatenation.

✅ Prefer ChatPromptTemplate for new projects.

✅ Keep system prompts concise and explicit.

---

# 16. Common Mistakes

❌ Hardcoding prompts everywhere.

❌ Building prompts using string concatenation.

❌ Putting business logic inside prompts.

❌ Ignoring system prompts.

# 17. Coding Example

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(
    model="gpt-4o-mini",
    temperature=0
)

prompt = ChatPromptTemplate.from_messages(

    [

        ("system", "You are a helpful AI tutor."),

        ("human", "Explain {topic}.")

    ]

)

chain = prompt | llm

response = chain.invoke(

    {

        "topic": "LangChain"

    }

)

print(response.content)
```

# 20. Interview Questions

## Q1. What is a Prompt Template?

**Answer:**

A Prompt Template is a reusable prompt containing placeholders that are replaced with runtime values, enabling dynamic prompt generation.

---

## Q2. Why use PromptTemplate instead of string formatting?

**Answer:**

PromptTemplate provides reusable, validated, and maintainable prompt construction, reducing manual string manipulation and improving code organization.

---

## Q3. What is ChatPromptTemplate?

**Answer:**

ChatPromptTemplate creates structured chat messages (System, Human, AI) for modern chat models, making it the preferred approach in LangChain.

---

## Q4. Which should you use in modern LangChain projects?

**Answer:**

ChatPromptTemplate, because modern LLM APIs are chat-based and support roles, conversation history, and tool calling.

---

## Q5. What are partial variables?

**Answer:**

Partial variables are template variables whose values are predefined and do not need to be supplied each time the template is invoked.
