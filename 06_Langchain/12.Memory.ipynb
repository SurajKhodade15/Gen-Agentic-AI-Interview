# 1. What is Memory?

By default, an LLM is **stateless**.

Every request is independent.

Example

Request 1

```
My name is Suraj.
```

↓

LLM replies

```
Nice to meet you.
```

---

Request 2

```
What's my name?
```

↓

Without previous context

```
I don't know.
```

The model forgot because it only received the second request.

---

# Interview Definition ⭐⭐⭐⭐⭐

> **Memory is the mechanism that stores and supplies previous conversation context to the LLM so it can generate context-aware responses across multiple interactions.**

# 3. How Memory Works

The LLM itself **does not remember**.

Your application sends previous messages.

```text
Conversation History

↓

Current Question

↓

LLM

↓

Answer
```

The LLM simply processes everything it receives.

# 6. Types of Memory

| Memory Type | Purpose |
|--------------|----------|
| Conversation Buffer | Stores complete conversation |
| Window Memory | Stores last N messages |
| Summary Memory | Stores summarized history |
| Entity Memory | Stores important entities |
| Vector Memory | Retrieves relevant memories |

Stores everything.

```text
Human

↓

AI

↓

Human

↓

AI

↓

Human

↓

AI
```

Advantages

- Full context

Disadvantages

# 12. Modern LangChain Memory

Earlier LangChain versions had classes like:

- ConversationBufferMemory
- ConversationSummaryMemory
- ConversationBufferWindowMemory

Example

### Modern Approach

Modern LangChain encourages managing **message history explicitly** using chat history and LangGraph state instead of relying heavily on legacy memory classes.

This provides:

- Better flexibility
- Better scalability
- Easier debugging

# 21. Interview Questions

## Q1. Does an LLM remember previous conversations?

**Answer:**

No. LLMs are stateless. Memory is managed by the application, which stores previous messages and includes relevant history in future requests.

---

## Q2. What is Memory in LangChain?

**Answer:**

Memory is a mechanism for storing and supplying previous conversation context to the LLM, enabling context-aware multi-turn interactions.

---

## Q3. What is the difference between Buffer Memory and Window Memory?

**Answer:**

Buffer Memory stores the entire conversation, whereas Window Memory stores only the most recent N messages, reducing token usage.

---

## Q4. What is Summary Memory?

**Answer:**

Summary Memory compresses long conversations into concise summaries so that important context is preserved while minimizing token consumption.

---

## Q5. How is modern LangChain memory different from older versions?

**Answer:**

Older versions relied on predefined memory classes such as `ConversationBufferMemory`. Modern LangChain encourages explicit management of chat history and state, especially with LangGraph, providing greater flexibility and scalability.

---

## Q6. What is the difference between Memory and RAG?

**Answer:**

Memory stores conversation history and user-specific context, while RAG retrieves relevant information from external knowledge sources such as documents or vector databases.

# 23. Real Interview Scenario

**Question:**

> A chatbot becomes slow and expensive after long conversations because every previous message is sent to the LLM. How would you solve this?

**Answer:**

I would avoid sending the entire conversation. Instead, I would use a sliding window for recent interactions, summarize older conversations, store long-term information separately (for example, in a database or vector store), and only include relevant history within the model's context window. This reduces token usage while preserving important context.

# 30-Second Interview Answer

> **LLMs are stateless and do not remember previous conversations. Memory in LangChain is implemented by the application, which stores conversation history and supplies relevant messages with each request. Different memory strategies, such as buffer, window, summary, entity, and vector memory, balance context retention with token efficiency. Modern LangChain favors explicit chat history and state management over legacy memory classes.**

---

# Key Takeaway

> **Memory is not a feature of the LLM—it is a responsibility of the application. Effective memory management ensures conversational continuity while keeping latency, cost, and context size under control.**
