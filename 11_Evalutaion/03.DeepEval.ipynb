# 1. What is DeepEval?

## Definition

DeepEval is an open-source LLM evaluation framework used to automatically test and evaluate GenAI applications using predefined and custom evaluation metrics.

It works similarly to unit testing in software engineering.

---

## Interview Answer

> DeepEval is an LLM evaluation framework that enables automated testing of AI applications, including RAG systems, AI agents, tool calling, and conversational assistants. It provides metrics such as Answer Relevancy, Faithfulness, Hallucination, Toxicity, Bias, Task Completion, and Tool Correctness, making it suitable for production AI quality assurance.

# 15. Simple Example

```python
# ==========================================================
# STEP 1 : Import Libraries
# ==========================================================

from deepeval.test_case import LLMTestCase

from deepeval.metrics import (
    AnswerRelevancyMetric,
    FaithfulnessMetric
)

from deepeval import evaluate


# ==========================================================
# STEP 2 : Create Test Case
# ==========================================================

test_case = LLMTestCase(

    input="How many annual leave days are provided?",

    actual_output="Employees receive 20 annual leave days.",

    expected_output="Employees receive 20 annual leave days."
)


# ==========================================================
# STEP 3 : Create Metrics
# ==========================================================

metrics = [

    AnswerRelevancyMetric(),

    FaithfulnessMetric()
]


# ==========================================================
# STEP 4 : Evaluate
# ==========================================================

evaluate(

    test_cases=[test_case],

    metrics=metrics
)
```

---

# 16. Example with Hallucination

```python
from deepeval.metrics import HallucinationMetric

hallucination = HallucinationMetric()

evaluate(

    test_cases=[test_case],

    metrics=[hallucination]
)
```

# 25. CI/CD Pipeline Integration

One of DeepEval's biggest strengths is automated testing.

```text
Developer

↓

GitHub

↓

GitHub Actions / Azure DevOps / CodePipeline

↓

Run Unit Tests

↓

Run DeepEval Tests

↓

All Metrics Above Threshold?

      │
 ┌────┴────┐
 │         │
Yes        No
 │         │
 ▼         ▼
Deploy   Block Deployment
```

Example policy:

- Faithfulness ≥ **0.90**
- Answer Relevancy ≥ **0.90**
- Hallucination ≤ **0.10**

If any threshold fails, the deployment is rejected.

This makes DeepEval valuable in enterprise environments because it acts like **automated quality gates** for AI applications.

# 27. Common Interview Questions

### Q1. What is DeepEval?

A framework for automated testing and evaluation of LLM applications, including RAG systems, AI agents, and tool-calling workflows.

---

### Q2. How is DeepEval different from RAGAS?

RAGAS focuses on **retrieval quality** in RAG systems.

DeepEval evaluates **complete AI applications**, including agents, tools, safety, and task completion.

---

### Q3. Can DeepEval evaluate Agentic AI?

Yes.

It supports evaluating:

- Tool selection
- Task completion
- Hallucination
- Multi-step workflows
- Agent behavior

---

### Q4. Can DeepEval run in CI/CD?

Yes.

It is commonly integrated into GitHub Actions, Azure DevOps, AWS CodePipeline, and Jenkins to prevent regressions before deployment.

---

### Q5. Do I still need LangSmith if I use DeepEval?

Yes.

- **LangSmith** explains **what happened** by providing traces and observability.
- **DeepEval** tells you **how well it performed** by assigning quality metrics and pass/fail results.

They complement each other rather than replacing one another.
