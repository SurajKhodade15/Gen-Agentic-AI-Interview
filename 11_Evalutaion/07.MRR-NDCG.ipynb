# Phase 8 — Evaluation
# MRR (Mean Reciprocal Rank) & NDCG (Normalized Discounted Cumulative Gain) ⭐⭐⭐⭐⭐

These are **advanced retrieval evaluation metrics** used to evaluate **how well your retriever ranks documents**.

Most engineers know **Precision@K** and **Recall@K**, but senior AI/ML and GenAI interviews (EPAM, Microsoft, Amazon, Google, Deloitte, etc.) often ask about **MRR** and **NDCG** because ranking quality directly impacts RAG performance.

---

Both retrievers retrieved the correct document.

The second one.

Why?

Because it ranked the correct document earlier.

That is what

measure.

# 3. MRR (Mean Reciprocal Rank)

## Definition

MRR measures

> **How early the first relevant document appears in the ranked results.**

Higher MRR means

The correct document appears closer to Rank 1.

---

## Interview Definition

> Mean Reciprocal Rank measures the average reciprocal of the rank of the first relevant document across multiple queries. It rewards retrieval systems that return the correct document as early as possible.

---

# Formula

For one query

```text
Reciprocal Rank

=

1 / Rank
```

MRR

=

Average of all reciprocal ranks.

# 5. NDCG (Normalized Discounted Cumulative Gain)

This is one of the most important ranking metrics.

Unlike MRR,

NDCG considers

ALL relevant documents

not just the first one.

---

## Definition

NDCG measures

> **How well the retriever ranks all relevant documents, giving higher importance to documents appearing earlier in the ranking.**

---

## Interview Definition

> NDCG measures ranking quality by rewarding highly relevant documents appearing near the top of the search results. It supports graded relevance instead of simple relevant/not relevant labels.

---

# Why "Discounted"?

Suppose

Rank

```text
1
```

Very valuable.

Rank

```text
10
```

Less valuable.

NDCG applies a

Discount

to lower ranks.

---

# Example

Suppose

Question

```text
Leave Policy
```

Retrieved

```text
Rank1 Leave Policy ⭐⭐⭐

Rank2 Carry Forward ⭐⭐

Rank3 Medical Leave ⭐

Rank4 Payroll
```

This is excellent ranking.

High NDCG.

---

Bad Ranking

```text
Rank1 Payroll

Rank2 Insurance

Rank3 Travel

Rank4 Leave Policy ⭐⭐⭐
```

Low NDCG.

# 10. MRR vs NDCG

| MRR | NDCG |
|------|------|
| Looks at first relevant document | Looks at all relevant documents |
| Simple | More comprehensive |
| Binary relevance | Supports graded relevance |
| Faster | More informative |
| Good for QA | Good for Search Engines |

---

# 11. MRR vs Precision@K

| Precision@K | MRR |
|--------------|-----|
| Measures accuracy of retrieved set | Measures position of first relevant result |
| Ignores ranking order | Ranking is critical |
| Top-K evaluation | Rank evaluation |

---

# 12. NDCG vs Recall@K

| Recall@K | NDCG |
|-----------|------|
| Measures completeness | Measures ranking quality |
| Ignores ordering | Rewards better ordering |
| Binary relevance | Graded relevance |

# 16. Common Interview Questions

### Q1. What is MRR?

MRR measures how early the **first relevant document** appears in the ranked search results.

---

### Q2. What is Reciprocal Rank?

It is simply:

```text
1 / Rank
```

Examples:

- Rank 1 → 1.0
- Rank 2 → 0.5
- Rank 5 → 0.2

---

### Q3. What is NDCG?

NDCG measures the quality of the **entire ranked list**, rewarding highly relevant documents that appear near the top.

---

### Q4. Which metric is more comprehensive?

NDCG.

Because it evaluates all relevant documents and their ranking positions.

---

### Q5. Which metric is more useful for RAG?

Both are useful:

- **MRR** is valuable for question-answering systems where finding the first correct document quickly is important.
- **NDCG** is valuable when multiple relevant documents contribute to the final answer.

# 19. EPAM Senior Answer (3–4 Minutes)

> "MRR and NDCG are ranking evaluation metrics used to measure the quality of document retrieval in RAG systems. MRR, or Mean Reciprocal Rank, focuses on the position of the first relevant document. If the first relevant document appears at rank one, the score is 1.0; if it appears at rank three, the score is 1/3. This makes MRR especially useful for question-answering systems where finding the first correct document quickly is critical. NDCG, or Normalized Discounted Cumulative Gain, evaluates the entire ranked list rather than just the first relevant document. It rewards highly relevant documents appearing near the top and supports graded relevance scores, making it a more comprehensive ranking metric. In production RAG systems on AWS using Amazon OpenSearch and Bedrock—or on Azure using Azure AI Search and Azure OpenAI—I use Precision@K and Recall@K to measure retrieval accuracy and completeness, MRR and NDCG to evaluate ranking quality, and Faithfulness and Hallucination metrics to evaluate the quality of the final LLM-generated answer."
