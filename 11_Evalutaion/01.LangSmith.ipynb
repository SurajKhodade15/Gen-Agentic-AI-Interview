# 1. What is LangSmith?

## Definition

LangSmith is an **LLM application development and observability platform** developed by LangChain.

It helps developers:

- Trace every LLM call
- Debug LangChain/LangGraph applications
- Evaluate RAG quality
- Compare prompts
- Monitor production AI systems
- Measure latency and cost

---

## Interview Answer

> LangSmith is an observability, debugging, and evaluation platform for LangChain and LangGraph applications. It provides execution traces, prompt inspection, token usage, latency analysis, dataset management, and automated evaluation, enabling developers to build reliable production-grade AI applications.

# 9. Simple LangChain Example

```python
# ==========================================================
# STEP 1 : Enable LangSmith Tracing
# ==========================================================

import os

os.environ["LANGCHAIN_TRACING_V2"] = "true"

os.environ["LANGCHAIN_API_KEY"] = "YOUR_API_KEY"

os.environ["LANGCHAIN_PROJECT"] = "Enterprise-RAG"


# ==========================================================
# STEP 2 : Create Bedrock LLM
# ==========================================================

from langchain_aws import ChatBedrockConverse

llm = ChatBedrockConverse(
    model="anthropic.claude-3-5-sonnet-20241022-v2:0",
    region_name="us-east-1"
)


# ==========================================================
# STEP 3 : Invoke Model
# ==========================================================

response = llm.invoke(
    "Explain the annual leave policy."
)

print(response.content)
```

Every invocation is automatically traced in LangSmith.

# 16. Best Practices

✅ Enable tracing in development and staging.

✅ Keep tracing enabled in production with appropriate sampling if traffic is high.

✅ Use meaningful project names.

✅ Store API keys securely.

✅ Tag traces with application version and deployment environment.

---

# 17. Common Mistakes

❌ Logging secrets or sensitive user data in traces.

❌ Ignoring failed traces.

❌ Not comparing prompt versions.

❌ Monitoring only latency and not answer quality.

# 21. Common Interview Questions

### Q1. What is LangSmith?

An observability and evaluation platform for LangChain and LangGraph applications.

---

### Q2. Why use LangSmith?

To trace, debug, evaluate, and monitor LLM applications.

---

### Q3. Does LangSmith replace CloudWatch?

No.

- **CloudWatch** monitors infrastructure and AWS services.
- **LangSmith** monitors LLM workflows, prompts, retrieval, tools, and agent execution.

---

### Q4. Can LangSmith trace LangGraph?

Yes.

It traces every node, edge, tool call, retriever, and LLM invocation.

---

### Q5. Can LangSmith evaluate RAG?

Yes.

It supports evaluations for retrieval quality, answer quality, hallucination, faithfulness, and custom evaluators.

---

### Q6. Is LangSmith only for LangChain?

It integrates most deeply with LangChain and LangGraph, but it can also trace custom Python code through decorators and APIs.
