# 2. Latency

Latency is the **total time taken to generate a response** from the moment the user sends a request until the response is returned.

# 16. Common Interview Questions

### Q1. What is latency?

The total time required to process a user request and return a response.

---

### Q2. What contributes to latency?

- API processing
- Retrieval
- Tool execution
- LLM inference
- Network communication

---

### Q3. What are prompt tokens?

The tokens sent as input to the LLM, including the system prompt, user query, conversation history, and retrieved context.

---

### Q4. What are completion tokens?

The tokens generated by the LLM as the response.

---

### Q5. How do you reduce token usage?

- Context Compression
- Better chunking
- Reranking
- Query Rewriting
- Lower Top-K
- Prompt optimization

---

### Q6. How do you reduce inference cost?

- Cache responses with Redis
- Use smaller models where appropriate
- Optimize prompts
- Compress context
- Remove duplicate retrieved content
- Monitor usage with LangSmith and cloud monitoring tools

---

### Q7. Which tools monitor these metrics?

| Metric | Tool |
|----------|------|
| Latency | LangSmith, CloudWatch, Application Insights |
| Cost | LangSmith, Bedrock/Azure billing dashboards |
| Token Usage | LangSmith, LLM provider usage metrics |

# 18. EPAM Senior Answer (3–4 Minutes)

> "In production AI systems, I monitor three key operational metrics: latency, cost, and token usage. Latency measures the total time required to process a request, including retrieval, tool execution, and LLM inference. Cost represents the expense of embeddings, vector search, model inference, storage, and infrastructure. Token usage consists of prompt tokens, completion tokens, and total tokens, which directly influence both latency and cost. To optimize these metrics, I use Hybrid Search, Semantic Chunking, Parent-Child Retrieval, reranking, Context Compression, Redis caching, prompt optimization, and appropriate model selection. On AWS, I monitor infrastructure with CloudWatch and AI workflows with LangSmith while using Amazon Bedrock and OpenSearch. On Azure, I use Application Insights together with LangSmith, Azure OpenAI, and Azure AI Search. This combination provides complete visibility into both infrastructure health and AI application performance, enabling reliable and cost-efficient enterprise deployments."
