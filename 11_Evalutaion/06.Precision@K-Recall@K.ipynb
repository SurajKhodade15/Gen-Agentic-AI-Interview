# Precision@K and Recall@K (Enterprise Retrieval Metrics ⭐⭐⭐⭐⭐)

These are **core Information Retrieval (IR) metrics** used to evaluate the **retriever**, not the LLM.

They are among the most frequently asked interview topics because every RAG system depends on retrieval quality.

Typical interview questions:

- What is Precision@K?
- What is Recall@K?
- Difference between Precision and Recall?
- Which metric is more important in RAG?
- How do you improve Precision@K and Recall@K?

---

# 3. Precision@K

## Definition

Precision@K measures

> **Out of the Top K retrieved documents, how many are actually relevant?**

---

## Formula

```text
Precision@K

=

Relevant Retrieved Documents

----------------------------

K
```

---

## Interview Definition

> Precision@K measures the percentage of retrieved documents in the top K results that are actually relevant.

# 4. Recall@K

## Definition

Recall@K measures

> **Out of all relevant documents available in the database, how many did the retriever successfully retrieve?**

---

## Formula

```text
Recall@K

=

Relevant Retrieved Documents

------------------------------------

Total Relevant Documents
```

---

## Interview Definition

> Recall@K measures the percentage of all relevant documents that were successfully retrieved within the top K results.

---

# Example

Suppose

Database contains

```text
10 Relevant Documents
```

Retriever found

```text
7
```

Calculation

```text
Recall

=

7 / 10

=

0.70
```

70%

---

Interpretation

Three important documents were missed.

# 18. Precision@K vs Recall@K

| Feature | Precision@K | Recall@K |
|----------|-------------|----------|
| Measures | Accuracy | Completeness |
| Formula | Relevant Retrieved / Retrieved | Relevant Retrieved / Total Relevant |
| Goal | Fewer irrelevant documents | Fewer missed documents |
| High Value Means | Better ranking | Better coverage |

---

# 19. Common Interview Questions

### Q1. What is Precision@K?

The proportion of the top K retrieved documents that are relevant.

---

### Q2. What is Recall@K?

The proportion of all relevant documents that appear in the top K retrieved results.

---

### Q3. Which metric is more important?

It depends on the application.

- Search engines often prioritize Precision.
- Healthcare and legal systems often prioritize Recall.
- Enterprise RAG systems usually optimize for a balance.

---

### Q4. Does increasing K always improve Recall?

Usually yes, because more documents are retrieved.

However, the exact improvement depends on whether additional retrieved documents are actually relevant.

---

### Q5. Does increasing K always improve Precision?

No.

Precision often decreases as K increases because more irrelevant documents are likely to be included.

---

# 20. Precision@K vs Recall@K vs Faithfulness

| Metric | Measures |
|---------|----------|
| Precision@K | Retriever accuracy |
| Recall@K | Retriever completeness |
| Faithfulness | Whether the LLM stayed grounded in the retrieved context |

Notice:

- **Precision@K** and **Recall@K** evaluate the **retriever**.
- **Faithfulness** evaluates the **generated answer**.

# 22. EPAM Senior Answer (3–4 Minutes)

> "Precision@K and Recall@K are retrieval evaluation metrics used to measure the quality of a RAG system's retriever. Precision@K measures how many of the top K retrieved documents are actually relevant, making it an indicator of retrieval accuracy. Recall@K measures how many of all relevant documents in the knowledge base were successfully retrieved within the top K results, making it an indicator of retrieval completeness. There is often a trade-off between the two: increasing K generally improves recall but may reduce precision by introducing more irrelevant documents. In production, I improve Precision@K using Hybrid Search, metadata filtering, reranking, and Query Rewriting, while I improve Recall@K using Multi-Query Retrieval, Semantic Chunking, Parent-Child Retrieval, and appropriate Top-K tuning. These metrics evaluate the retriever, whereas metrics such as Faithfulness and Hallucination evaluate the LLM's generated answer."
