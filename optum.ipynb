# 1. Model Serving

## What is Model Serving?

Model Serving is the process of making a trained ML/LLM model available to other applications through an API so it can perform inference in real time or batch mode.

Simply:

```
Training → Save Model → Deploy → API → Prediction
```

---

## Why do we need Model Serving?

Without serving:

- Model only works in Jupyter Notebook
- Nobody can access it
- Cannot integrate with applications

Serving enables

- REST APIs
- High availability
- Authentication
- Monitoring
- Auto Scaling
- Versioning

---

## Typical Architecture

```
                Client
                   │
                   ▼
             API Gateway
                   │
             Load Balancer
                   │
         -------------------
         │                 │
     FastAPI Pod      FastAPI Pod
         │                 │
         └────────┬────────┘
                  │
            LLM / ML Model
                  │
         Vector Database
                  │
              PostgreSQL
```

# 2. Model Drift

## What is Model Drift?

Model Drift occurs when production data changes over time and model accuracy decreases.

---

## Types

### Data Drift

Input data changes.

Example

Training

Age = 25–40

Production

Age = 55–80

Distribution changed.

---

### Concept Drift

Relationship changes.

Example

Fraud patterns evolve.

Old fraud detection no longer works.

---

### Prediction Drift

Prediction distribution changes.

Example

Model predicts

95% Positive

Earlier

50% Positive

Something is wrong.

---

## Detection

- Accuracy
- Precision
- Recall
- F1 Score
- PSI
- KS Test
- Embedding Drift
- Feature Distribution

---

## Tools

# Common Production Interview Questions

### Why Docker?

**Answer:**
Docker packages the application with all its dependencies, ensuring consistent execution across development, testing, and production environments while simplifying deployment and scaling.

---

### Why Kubernetes?

**Answer:**
Kubernetes automates deployment, scaling, load balancing, rolling updates, and self-healing of containerized applications, making it suitable for production-grade AI services.

---

### Why EKS instead of EC2?

**Answer:**
EC2 requires manual server management. EKS abstracts Kubernetes control plane management and provides built-in orchestration, auto scaling, rolling deployments, and self-healing, making it more suitable for scalable AI workloads.

---

### How would you deploy a FastAPI AI application?

**Answer:**
1. Build the FastAPI application.
2. Containerize it with Docker.
3. Push the image to Amazon ECR.
4. Deploy on Amazon EKS (or ECS for simpler workloads).
5. Expose it via an Application Load Balancer.
6. Enable Horizontal Pod Autoscaler.
7. Monitor using CloudWatch and Prometheus/Grafana.
8. Use CI/CD (e.g., GitHub Actions) for automated deployments.

---

### How do you make an AI application scalable?

**Answer:**
- Stateless API services.
- Horizontal scaling with Kubernetes.
- Load balancing.
- Redis for caching.
- SQS/EventBridge for asynchronous jobs.
- Optimize LLM inference with batching and streaming.
- Auto scale based on CPU, memory, or request latency.

---

### How do you make an AI application reliable?

**Answer:**
- Multi-AZ deployment.
- Health checks and readiness probes.
- Automatic retries with exponential backoff.
- Circuit breakers for dependent services.
- Queue-based processing for long-running tasks.
- Centralized logging and monitoring.
- Regular backups and disaster recovery planning.

---

### How do you deploy an LLM application on AWS?

**Answer:**
For a production GenAI application, I would expose a FastAPI service on EKS, integrate with Amazon Bedrock (or a self-hosted model if required), use Amazon S3 for documents, a vector database for embeddings, Amazon RDS for metadata, Redis for caching, EventBridge/SQS for asynchronous processing, and CloudWatch with Prometheus/Grafana for monitoring. CI/CD would build Docker images, push them to ECR, and deploy updates to EKS using rolling or blue-green deployments. This architecture provides scalability, resilience, observability, and security suitable for enterprise healthcare workloads.
