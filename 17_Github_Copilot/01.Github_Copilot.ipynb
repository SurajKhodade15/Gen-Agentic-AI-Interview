# GitHub Copilot Enterprise

**GitHub Copilot Enterprise** is GitHub's enterprise-focused AI coding assistant capability. It extends GitHub Copilot with organization-level controls and the ability to provide more relevant assistance using an organization's codebase and development context.

For your Altimetrik JD, focus on:

```text
GitHub Copilot Enterprise
        │
        ├── AI-assisted development
        ├── Enterprise codebase context
        ├── Custom instructions
        ├── Organization governance
        ├── Code generation
        ├── Code explanation
        ├── Refactoring
        ├── Test generation
        └── Developer productivity
```

---

# 1. What is GitHub Copilot?

At a basic level:

```text
Developer
   ↓
GitHub Copilot
   ↓
AI assistance
   ↓
Code / explanation / tests / suggestions
```

It can assist with:

- Writing code
- Completing code
- Explaining code
- Refactoring
- Generating tests
- Debugging
- Documentation
- Code review assistance

Example:

```python
def calculate_salary(base_salary, bonus):
    # Ask Copilot to calculate total compensation
```

Copilot can suggest:

```python
def calculate_salary(base_salary, bonus):
    return base_salary + bonus
```

# 4. Custom Instructions ⭐⭐⭐⭐⭐

This is one of the most important interview topics.

Organizations can define instructions that guide Copilot's responses.

For example:

```text
Project standards:

- Use Python 3.11
- Follow PEP 8
- Use type hints
- Use Pydantic for request models
- Use FastAPI
- Write unit tests using pytest
- Follow repository naming conventions
- Do not expose secrets
```

Then when a developer asks:

> "Create an API for employee details."

Copilot can generate code aligned with the organization's conventions.

Conceptually:

```text
Developer Request
       +
Custom Instructions
       +
Repository Context
       ↓
GitHub Copilot
       ↓
Enterprise-aligned Code
```

# 8. Code Generation

Example prompt:

```text
Create a FastAPI endpoint that accepts an employee ID
and returns the employee's leave balance.
Use Pydantic and type hints.
```

Copilot can generate:

```python
from fastapi import FastAPI
from pydantic import BaseModel

app = FastAPI()


class EmployeeRequest(BaseModel):
    employee_id: str


@app.post("/leave-balance")
def get_leave_balance(request: EmployeeRequest):
    return {
        "employee_id": request.employee_id,
        "leave_balance": 12
    }
```

The important enterprise practice is:

> **Review and validate generated code; don't blindly accept it.**

# 9. Unit Test Generation

Given:

```python
def add(a: int, b: int) -> int:
    return a + b
```

You can ask Copilot:

> "Generate pytest unit tests for this function."

Potential output:

```python
def test_add():
    assert add(2, 3) == 5


def test_add_negative():
    assert add(-2, 3) == 1
```

This accelerates development while keeping testing responsibility with the developer.

For enterprise adoption, consider:

```text
Developer
   ↓
GitHub Copilot
   ↓
Organization Policies
   ↓
Repository / Code Context
```

# 18. Copilot vs ChatGPT

A useful conceptual distinction:

| GitHub Copilot | ChatGPT |
|---|---|
| Primarily developer workflow | General-purpose AI assistant |
| Deeply integrated with coding environments/GitHub workflows | Broad conversational capabilities |
| Code completion/generation | Coding + analysis + many other tasks |
| Repository/development context | Context depends on the product/workflow |
| Enterprise GitHub governance | OpenAI workspace/product governance |
| Developer productivity | Broader use cases |

Don't present this as an absolute product boundary because capabilities evolve.

```text
GitHub Copilot
      ↓
AI-assisted software development
```

```text
Agentic AI
      ↓
AI system that reasons/plans
      ↓
Uses tools
      ↓
Executes workflows
```
