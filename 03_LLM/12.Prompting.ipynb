# 18. System Prompt ⭐⭐⭐⭐⭐

## What is a System Prompt?

A **System Prompt** defines the **role, behavior, rules, and personality** of the LLM.

It acts as **global instructions** for the conversation.

---

## Interview Definition

> **A System Prompt is a high-priority instruction that defines how the model should behave throughout the conversation.**

---

## Example

```text
You are a Senior Python Developer.

Always answer in markdown.

Explain concepts with examples.

Never generate harmful code.
```

Now User asks

```
Explain Python List.
```

Response

```
Markdown format

+

Python examples

+

Professional explanation
```

---

## Why Use System Prompt?

- Define AI role
- Control response style
- Enforce rules
- Maintain consistency

---

## Small Coding Example

```python
from openai import OpenAI

client = OpenAI()

response = client.responses.create(
    model="gpt-4.1",
    instructions="""
    You are a Senior Python Trainer.
    Explain with examples.
    """,
    input="Explain decorators."
)

print(response.output_text)
```

---

## Best Practices

✅ Define role clearly

✅ Keep instructions concise

✅ Specify output format

---

## Common Mistakes

❌ Writing vague instructions

❌ Conflicting instructions

# 19. User Prompt ⭐⭐⭐⭐

## What is a User Prompt?

A **User Prompt** is the actual request or question sent to the LLM.

---

## Interview Definition

> **A User Prompt is the input provided by the user that specifies the task the LLM should perform.**

---

## Example

System Prompt

```text
You are an SQL Expert.
```

User Prompt

```text
Write an SQL query to find the top 5 highest-paid employees.
```

---

## Good Prompt

```text
Explain LangGraph.

Use simple language.

Give one example.

Limit answer to 200 words.
```

---

## Poor Prompt

```text
Explain.
```

---

## Small Coding Example

```python
response = client.responses.create(
    model="gpt-4.1",
    input="Explain Multi-Head Attention in simple words."
)
```

---

## Best Practices

✅ Be specific

✅ Provide context

✅ Mention output format

---

## Common Mistakes

❌ Asking multiple unrelated questions

❌ No context

# 20. Structured Output (JSON Mode) ⭐⭐⭐⭐⭐

## What is Structured Output?

Instead of generating free-form text,

the model generates **structured JSON**.

---

## Interview Definition

> **Structured Output forces the model to return responses in a predefined schema, making them easy for applications to parse programmatically.**

---

## Why Is It Needed?

Without JSON

```text
John is 25 years old.
```

Application cannot reliably parse it.

---

With JSON

```json
{
  "name": "John",
  "age": 25
}
```

Easy to consume by backend applications.

---

## Architecture

```text
User Prompt

↓

LLM

↓

JSON Schema

↓

Structured Response

↓

Backend
```

---

## Small Coding Example

```python
from pydantic import BaseModel
from openai import OpenAI

class Person(BaseModel):
    name: str
    age: int

client = OpenAI()

response = client.responses.parse(
    model="gpt-4.1",
    input="John is 25 years old.",
    text_format=Person
)

print(response.output_parsed)
```

Output

```python
Person(
    name="John",
    age=25
)
```

---

## Real-World Uses

- Resume Parsing
- Invoice Extraction
- Chat APIs
- AI Agents
- RAG Pipelines

---

## Best Practices

✅ Always define schema

✅ Validate response

---

## Common Mistakes

❌ Parsing plain text manually

❌ Not validating JSON

# 21. Function Calling (Overview) ⭐⭐⭐⭐⭐

## What is Function Calling?

Function Calling allows an LLM to decide

> **when to call an external function instead of answering directly.**

---

## Interview Definition

> **Function Calling enables an LLM to identify when an external tool or function is required and return structured arguments for that function instead of generating a direct answer.**

---

## Example

User

```
What's the weather in Pune?
```

Instead of guessing,

LLM returns

```json
{
  "tool": "get_weather",
  "arguments": {
    "city": "Pune"
  }
}
```

Application

↓

Calls Weather API

↓

Returns

```
30°C
Cloudy
```

---

## Architecture

```text
User

↓

LLM

↓

Should Tool Be Called?

↓

Yes

↓

Function Arguments

↓

Tool/API

↓

Result

↓

LLM

↓

Final Response
```

---

## Small Coding Example

```python
from openai import OpenAI

client = OpenAI()

response = client.responses.create(
    model="gpt-4.1",
    input="What's the weather in Pune?",
    tools=[
        {
            "type": "function",
            "name": "get_weather",
            "description": "Get current weather",
            "parameters": {
                "type": "object",
                "properties": {
                    "city": {
                        "type": "string"
                    }
                },
                "required": ["city"]
            }
        }
    ]
)
```

---

## Real-World Uses

- Weather Bots
- SQL Generation
- Database Queries
- Web Search
- AI Agents
- LangGraph
- OpenAI Agents SDK

---

## Best Practices

✅ Define tool schema clearly

✅ Validate arguments

✅ Keep functions focused

---

## Common Mistakes

❌ Expecting the LLM to execute functions

❌ Not validating tool inputs

> **Remember:** The LLM decides **which function to call**, but **your application executes it**.

# Common Interview Questions

### Q1. Difference between System Prompt and User Prompt?

**Answer**

- System Prompt defines the model's behavior.
- User Prompt defines the current task.

---

### Q2. Why use Structured Output?

**Answer**

To return predictable, machine-readable responses that backend systems can process without manual parsing.

---

### Q3. What is Function Calling?

**Answer**

Function Calling allows the LLM to generate structured arguments for external tools instead of answering directly.

---

### Q4. Does the LLM execute functions?

**Answer**

No.

The LLM only decides **which function to call** and generates the arguments.

Your application executes the function.

---

### Q5. Difference between Structured Output and Function Calling?

**Answer**

Structured Output returns JSON as the **final response**.

Function Calling returns JSON **to invoke an external tool**, after which the tool result is typically sent back to the model to generate the final answer.

# 30-Second Interview Answer

> **Prompting is the process of instructing an LLM to perform a task. The System Prompt defines the model's role and behavior, while the User Prompt specifies the actual request. Structured Output ensures responses follow a predefined JSON schema for reliable application integration, and Function Calling enables the model to invoke external tools by generating structured function arguments. Together, these techniques form the foundation of production-grade GenAI and Agentic AI applications.**
