# 27. Hallucination ⭐⭐⭐⭐⭐

## What is Hallucination?

A **Hallucination** occurs when an LLM generates information that sounds correct but is **factually incorrect, fabricated, or unsupported**.

---

## Interview Definition

> **Hallucination is the generation of incorrect, fabricated, or misleading information by an LLM while presenting it as if it were factual.**

---

## Example

Prompt

```
Who invented Python?
```

Correct

```
Guido van Rossum
```

Hallucinated Response

```
James Gosling
```

The answer is fluent but incorrect.

---

## Why Does Hallucination Happen?

- Missing knowledge
- Ambiguous prompts
- Weak retrieval
- High Temperature
- Probabilistic text generation

---

## Types

### 1. Factual Hallucination

```
The moon has two suns.
```

---

### 2. Citation Hallucination

```
According to IEEE Paper 2027...
```

(No such paper exists.)

---

### 3. Tool Hallucination

```
Weather API says 42°C.
```

(No API call was made.)

---

## How to Reduce Hallucination?

✅ Use RAG

✅ Use low Temperature

✅ Verify outputs

✅ Use trusted data sources

✅ Use tool/function calling

---

## Small Coding Example

```python
response = client.responses.create(
    model="gpt-4.1",
    input="Answer only using the provided context.",
    temperature=0
)
```

---

## Interview Question

### Q. Can Hallucination be completely eliminated?

**Answer**

No.

It can only be reduced through techniques such as RAG, grounding, tool use, guardrails, and verification.

# 28. KV Cache ⭐⭐⭐⭐

## What is KV Cache?

KV Cache (Key-Value Cache) stores previously computed **Key and Value vectors** during generation.

Instead of recomputing attention for every previous token, the model reuses cached values.

---

## Interview Definition

> **KV Cache is an inference optimization that stores previously computed Key and Value tensors, enabling faster autoregressive generation by avoiding repeated attention computations.**

---

## Without KV Cache

```text
Token 1

↓

Recompute

↓

Token 2

↓

Recompute

↓

Token 3

↓

Recompute
```

Every step repeats old calculations.

---

## With KV Cache

```text
Token 1

↓

Store K & V

↓

Token 2

↓

Reuse Previous K & V

↓

Token 3

↓

Reuse Cache
```

Only the new token needs computation.

---

## Benefits

✅ Faster inference

✅ Lower latency

✅ Better throughput

---

## Why Is It Important?

ChatGPT generates one token at a time.

Without KV Cache,

generation would become increasingly slower.

---

## Interview Question

### Q. Does KV Cache improve training?

**Answer**

No.

KV Cache is an **inference optimization**, not a training optimization.

# 29. Ollama ⭐⭐⭐⭐⭐

## What is Ollama?

Ollama is a framework for **running open-source LLMs locally** on your own machine.

---

## Interview Definition

> **Ollama is a local inference platform that allows developers to download, manage, and run open-source LLMs such as Llama, Mistral, Gemma, and Qwen without relying on cloud APIs.**

---

## Architecture

```text
Application

↓

Ollama

↓

Local LLM

↓

Response
```

---

## Supported Models

- Llama
- Mistral
- Gemma
- Qwen
- DeepSeek
- Phi

---

## Installation

```bash
ollama run llama3
```

---

## Python Example

```python
from langchain_ollama import ChatOllama

llm = ChatOllama(
    model="llama3"
)

print(llm.invoke("Explain LangGraph."))
```

---

## Why Use Ollama?

✅ Offline

✅ No API cost

✅ Data privacy

✅ Local development

---

## Limitations

- Requires local hardware
- Slower than cloud GPUs (depending on your machine)
- Large models require more RAM/VRAM

---

## Interview Question

### Q. Why choose Ollama over OpenAI?

**Answer**

Ollama is ideal for local deployment, privacy-sensitive applications, offline environments, and avoiding recurring API costs.

# 30. vLLM (Overview) ⭐⭐⭐⭐⭐

## What is vLLM?

vLLM is a **high-performance inference engine** for serving LLMs efficiently in production.

It is designed to maximize throughput and reduce latency.

---

## Interview Definition

> **vLLM is an optimized LLM serving engine that improves inference performance using techniques such as PagedAttention and efficient request batching.**

---

## Architecture

```text
Users

↓

vLLM Server

↓

Optimized GPU Memory

↓

LLM

↓

Responses
```

---

## Key Features

- High Throughput
- Low Latency
- Efficient GPU Memory
- Continuous Batching
- OpenAI-Compatible API

---

## Why Is vLLM Faster?

It uses

- PagedAttention
- KV Cache optimization
- Continuous batching
- Efficient memory management

---

## Small Example

```bash
python -m vllm.entrypoints.openai.api_server \
--model meta-llama/Llama-3-8B
```

Now applications can use the server like an OpenAI-compatible endpoint.

---

## When Should You Use vLLM?

- Production APIs
- High traffic
- Multiple concurrent users
- GPU servers

---

## Ollama vs vLLM

| Ollama | vLLM |
|---------|------|
| Local development | Production serving |
| Easy setup | High-performance inference |
| CPU/GPU | Primarily GPU optimized |
| Single developer | Multi-user deployment |
| Offline usage | API serving |

---

## Interview Question

### Q. Why is vLLM popular?

**Answer**

Because it significantly increases inference throughput while reducing latency and GPU memory usage, making it well suited for production-scale LLM serving.
