# 1. What is a Token?

A **Token** is the smallest unit of text that an LLM processes.

A token **is not always a word**.

A token can be:

- A word
- Part of a word
- A punctuation mark
- A number
- A special symbol

---

# Interview Definition ⭐⭐⭐⭐⭐

> **A Token is the smallest unit of text processed by an LLM. Tokens may represent complete words, subwords, punctuation, or special characters.**

---

# 2. What is Tokenization?

**Tokenization** is the process of converting raw text into tokens before sending it to the LLM.

LLMs **cannot understand plain text** directly.

They first convert text into tokens.

---

# 3. Why Do LLMs Use Tokens Instead of Words?

Consider the word:

```
unbelievable
```

Instead of storing every possible word,

the tokenizer may split it into:

```text
un

believe

able
```

This helps the model:

- Handle new words
- Reduce vocabulary size
- Support multiple languages
- Improve efficiency

---

# 4. Tokenization Flow

```text
Input Text

↓

Tokenizer

↓

Tokens

↓

Token IDs

↓

LLM
```

---

# 5. Example

Input

```
I love Python.
```

Possible Tokens

```text
"I"

" love"

" Python"

"."
```

Then converted into IDs

```text
[40, 1842, 13325, 13]
```

> **Note:** Token IDs vary depending on the tokenizer and model.

---

# 6. Types of Tokenization

| Type | Example |
|------|---------|
| Character | `H`, `e`, `l`, `l`, `o` |
| Word | `Hello`, `World` |
| Subword ⭐ | `un`, `believe`, `able` |
| Sentence | Entire sentence |

Modern LLMs mainly use **Subword Tokenization**.

---

# 7. Why Subword Tokenization?

Unknown word

```
ChatGPTEngineer
```

Instead of failing,

Tokenizer splits it

```text
Chat

GPT

Engineer
```

The model can still understand it.

---

# 8. Tokenization Architecture

```text
Raw Text

↓

Tokenizer

↓

Tokens

↓

Token IDs

↓

Embedding Layer

↓

Transformer
```

Notice:

**Tokenization happens before the Transformer.**

---

# 9. Small Coding Example

Using OpenAI's tokenizer (`tiktoken`)

```python
import tiktoken

encoding = tiktoken.get_encoding("cl100k_base")

tokens = encoding.encode(
    "Hello World"
)

print(tokens)
```

Possible Output

```python
[9906, 4435]
```

Decode back

```python
text = encoding.decode(tokens)

print(text)
```

Output

```
Hello World
```

# 14. Interview Questions

## Q1. What is a Token?

**Answer:**

A Token is the smallest unit of text processed by an LLM. It can represent a whole word, part of a word, punctuation, or special characters.

---

## Q2. What is Tokenization?

**Answer:**

Tokenization is the process of converting raw text into tokens that can be processed by the LLM.

---

## Q3. Why don't LLMs process words directly?

**Answer:**

LLMs use subword tokenization to reduce vocabulary size, handle unseen words, and support multiple languages more efficiently.

---

## Q4. Does every model use the same tokenizer?

**Answer:**

No. Different models use different tokenizers, so the same text may produce different token IDs.

---

## Q5. Why are tokens important in OpenAI APIs?

**Answer:**

API pricing, context limits, latency, and response generation are all based on the number of tokens rather than the number of words.

# 16. Real Interview Scenario

**Question:**

> Your application sends a very large prompt to GPT-4. Why is understanding tokenization important?

**Answer:**

The prompt is first tokenized before being processed by the model. If the number of tokens exceeds the model's context window, the request may fail or require truncation. Additionally, API pricing and response latency are based on token count, so understanding tokenization helps optimize both cost and performance.
