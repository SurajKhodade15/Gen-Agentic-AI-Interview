# 1. Why Do We Need QKV?

In the previous topic, we learned:

> **Every token looks at every other token.**

But a question remains:

> **How does the model decide which tokens are important?**

The answer is:

**Query (Q), Key (K), and Value (V).**

---

# 2. Real-Life Analogy

Imagine you're in a library.

You want books about **Python**.

### You

```
↓

Query
```

### Library Index

```
↓

Keys
```

### Books

```
↓

Values
```

You compare your query with the available keys.

The best matching books (values) are returned.

This is exactly how Self-Attention works.

---

# 3. What is Query (Q)?

A **Query** represents:

> **"What information am I looking for?"**

Every token generates its own Query vector.

---

# Interview Definition

> **Query is a vector that represents what information the current token is searching for from other tokens.**

---

# 4. What is Key (K)?

A **Key** represents:

> **"What information do I contain?"**

Every token generates its own Key vector.

---

# Interview Definition

> **Key is a vector that represents the information a token offers to other tokens.**

---

# 5. What is Value (V)?

A **Value** contains the actual information that will be used if the token is considered important.

---

# Interview Definition

> **Value is the semantic information associated with a token that is combined to produce the final contextual representation.**

---

# 6. QKV Flow

```text
Each Token

↓

Create

Query

Key

Value

↓

Compare

Query

with

Keys

↓

Attention Scores

↓

Weighted Values

↓

Context-Aware Output
```

This is the heart of Self-Attention.

# 16. Interview Questions

## Q1. What are Query, Key, and Value?

**Answer:**

Query, Key, and Value are vector representations generated for every token. Query represents what the token is looking for, Key represents what information the token offers, and Value contains the information that contributes to the final contextual representation.

---

## Q2. Why do we need Query, Key, and Value?

**Answer:**

They allow the model to determine which tokens are most relevant to the current token by comparing Queries with Keys and then combining the corresponding Values.

---

## Q3. Which vectors are compared during Self-Attention?

**Answer:**

The Query vector of the current token is compared with the Key vectors of all tokens to compute attention scores.

---

## Q4. What is the purpose of the Value vector?

**Answer:**

The Value vector contains the semantic information that is weighted and combined based on the attention scores to produce the final output representation.

---

## Q5. Does every token have its own Q, K, and V?

**Answer:**

Yes. Every token generates its own Query, Key, and Value vectors.

# 18. Real Interview Scenario

**Question:**

> In the sentence "Virat scored a century because he played well," how does the Transformer know that "he" refers to "Virat"?

**Answer:**

The token "he" generates a Query vector, which is compared with the Key vectors of all other tokens. Since "Virat" has the highest contextual similarity, its Value vector receives the highest attention weight. As a result, the model correctly understands that "he" refers to "Virat."

By comparing Queries with Keys and weighting the corresponding Values, the Transformer builds context-aware representations that enable accurate language understanding.**
