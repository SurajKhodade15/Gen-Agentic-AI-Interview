# 1. What is a Transformer?

A **Transformer** is a deep learning architecture introduced by Google in 2017 in the paper:

> **"Attention Is All You Need"**

It processes text using the **Attention Mechanism**, allowing it to understand relationships between words in parallel.

---

# Interview Definition ⭐⭐⭐⭐⭐

> **A Transformer is a neural network architecture that processes text using self-attention, enabling efficient understanding of long-range relationships and parallel computation.**

---

# 2. Why Was the Transformer Introduced?

Before Transformers, NLP models mainly used:

- RNN
- LSTM

Problems:

❌ Sequential processing

❌ Slow training

❌ Difficulty remembering long contexts

Transformers solved these issues using **Attention**.

# 7. Why Is It Called "Attention"?

Suppose the sentence is:

```
The cat sat on the mat because it was tired.
```

When processing the word:

```
it
```

The model should understand that **"it" refers to "cat"**, not **"mat"**.

Attention helps the model focus on the most relevant words.

# 14. Interview Questions

## Q1. What is a Transformer?

**Answer:**

A Transformer is a deep learning architecture that uses self-attention to process text efficiently and understand relationships between words.

---

## Q2. Why did Transformers replace RNNs and LSTMs?

**Answer:**

Because Transformers process tokens in parallel, capture long-range dependencies better, and train much faster than sequential models like RNNs and LSTMs.

---

## Q3. Does the Transformer process raw text?

**Answer:**

No. Raw text is first tokenized and converted into embeddings. The Transformer processes embedding vectors.

---

## Q4. What is the most important component of a Transformer?

**Answer:**

The Attention Mechanism, especially Self-Attention, because it helps the model determine which words are most relevant to each other.

---

## Q5. Are GPT and Llama based on the Transformer?

**Answer:**

Yes. GPT, Llama, Mistral, Qwen, DeepSeek, and most modern LLMs are built on the Transformer architecture.

**Question:**
