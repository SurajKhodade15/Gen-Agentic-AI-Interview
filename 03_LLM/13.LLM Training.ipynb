# 22. Pre-training ⭐⭐⭐⭐

## What is Pre-training?

Pre-training is the **first phase** of training an LLM.

The model learns general language patterns from massive text datasets by performing **Next Token Prediction**.

---

## Interview Definition

> **Pre-training is the process of training a language model on massive unlabeled text data using next-token prediction to learn grammar, facts, reasoning patterns, and language representations.**

---

## Training Data

Examples

- Wikipedia
- Books
- Research Papers
- GitHub Code
- News Articles
- Public Web Pages

---

## Objective

Input

```
Python is a programming
```

Target

```
language
```

The model repeatedly predicts the next token millions or billions of times.

---

## Result

```
Raw Data

↓

Pre-training

↓

Base LLM
```

Examples

- GPT Base
- Llama Base

---

## Small Example

Dataset

```text
Python is a programming language.

AI is transforming industries.

The capital of France is Paris.
```

The model continuously learns

```
Next Token Prediction
```

---

## Interview Question

### Q. What does Pre-training teach the model?

**Answer**

General language understanding, grammar, reasoning, coding patterns, and factual knowledge.

---

# 23. Fine-Tuning ⭐⭐⭐⭐

## What is Fine-Tuning?

Fine-Tuning trains an already pre-trained model on a **smaller domain-specific dataset**.

---

## Interview Definition

> **Fine-Tuning adapts a pre-trained LLM to a specific domain or task using labeled data.**

---

## Example

Base LLM

↓

Medical Dataset

↓

Medical Chatbot

---

Another Example

Base LLM

↓

Legal Documents

↓

Legal Assistant

---

## Architecture

```text
Base Model

↓

Domain Dataset

↓

Fine-Tuning

↓

Specialized Model
```

---

## Small Example

General Model

```
What is diabetes?
```

↓

Medical Fine-Tuning

↓

Detailed medical response.

---

## Use Cases

- Healthcare
- Finance
- Legal
- Customer Support
- Enterprise Chatbots

---

## Interview Question

### Q. Why Fine-Tune instead of training from scratch?

**Answer**

Pre-training requires enormous compute and data. Fine-Tuning is faster, cheaper, and adapts an existing model to a specific domain.

---

# 24. Instruction Tuning ⭐⭐⭐⭐⭐

## What is Instruction Tuning?

Instruction Tuning teaches the model

> **How to follow human instructions.**

---

## Interview Definition

> **Instruction Tuning is a supervised fine-tuning process where an LLM learns to follow natural language instructions using prompt-response pairs.**

---

## Example

Training Data

```text
Instruction:

Summarize this article.

↓

Expected Output:

Article Summary
```

---

Another Example

```text
Translate to French.

↓

French Translation
```

The model learns to respond correctly to different instruction styles.

---

## Architecture

```text
Base Model

↓

Instruction Dataset

↓

Instruction-Tuned Model
```

---

## Result

Instead of only predicting text,

the model now understands commands like

- Explain
- Summarize
- Translate
- Generate Code

---

## Interview Question

### Q. Difference between Fine-Tuning and Instruction Tuning?

**Answer**

Fine-Tuning specializes the model for a domain or task.

Instruction Tuning teaches the model to follow human instructions effectively.

---

# 25. RLHF (Reinforcement Learning from Human Feedback) ⭐⭐⭐⭐⭐

## What is RLHF?

RLHF improves the model using **human preferences**.

Humans compare multiple model responses and indicate which one is better.

---

## Interview Definition

> **RLHF aligns an LLM with human preferences by using human feedback to optimize response quality, safety, and helpfulness.**

---

## Workflow

```text
Prompt

↓

LLM

↓

Response A

Response B

↓

Human Chooses Better Response

↓

Reward Model

↓

LLM Improves
```

---

## Example

Prompt

```
How do I learn Python?
```

Response A

```
Practice regularly and build projects.
```

Response B

```
Python is a snake.
```

Human selects

```
Response A
```

The model learns that A is preferable.

---

## Why RLHF?

Without RLHF

- Less helpful
- Less safe
- More inconsistent

With RLHF

- Helpful
- Safe
- Better aligned with user expectations

---

## Interview Question

### Q. Why is RLHF important?

**Answer**

RLHF aligns model behavior with human preferences, improving helpfulness, safety, and response quality.

---

# 26. LoRA (High Level) ⭐⭐⭐⭐

## What is LoRA?

**LoRA (Low-Rank Adaptation)** is an efficient Fine-Tuning technique.

Instead of updating **all model parameters**, it trains only a **small number of additional parameters**.

---

## Interview Definition

> **LoRA is a parameter-efficient fine-tuning technique that adapts an LLM by training a small set of additional weights while keeping the original model frozen.**

---

## Traditional Fine-Tuning

```text
Base Model

↓

Update All Parameters

↓

Very Expensive
```

---

## LoRA

```text
Base Model

↓

Freeze Original Model

↓

Train Small LoRA Layers

↓

Fast & Efficient
```

---

## Why LoRA?

Advantages

- Lower GPU memory
- Faster training
- Lower cost
- Easier deployment

---

## Use Cases

- Enterprise LLM customization
- Domain adaptation
- Local LLM fine-tuning
- Open-source models (Llama, Mistral, Qwen)

---

## Interview Question

### Q. Why is LoRA popular?

**Answer**

Because it enables efficient fine-tuning of large models with significantly lower computational and memory requirements compared to updating all model parameters.

|-----------|---------|---------------|--------|
| Pre-training | Learn language | Massive unlabeled data | Base LLM |

# Common Interview Questions

### Q1. What is Pre-training?

**Answer**

Training a model on massive unlabeled text using next-token prediction to learn general language understanding.

---

### Q2. Why is Fine-Tuning needed?

**Answer**

To adapt a general-purpose model to a specific domain or task using a smaller labeled dataset.

---

### Q3. What is Instruction Tuning?

**Answer**

A supervised fine-tuning process that teaches an LLM to follow natural language instructions using prompt-response examples.

---

### Q4. Why is RLHF important?

**Answer**

RLHF aligns the model with human preferences, making responses more helpful, safe, and reliable.

---

### Q5. What is the main advantage of LoRA?

**Answer**

LoRA fine-tunes only a small number of additional parameters, reducing training cost and memory while preserving the original model.

---

### Q6. Can LoRA be used on GPT-4?

**Answer**

Generally, **No** for hosted proprietary models like GPT-4 because users do not have access to the model weights. LoRA is primarily used with open-source models such as Llama, Mistral, Qwen, and Gemma.
