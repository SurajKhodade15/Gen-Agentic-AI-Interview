# 1. What is a Context Window?

The **Context Window** is the maximum number of **tokens** an LLM can process in a single request.

It includes:

- Input Prompt
- Conversation History
- Retrieved Context (RAG)
- Generated Output

---

# Interview Definition ⭐⭐⭐⭐⭐

> **The Context Window is the maximum number of input and output tokens that an LLM can consider during a single inference.**

---

# 2. Why Do We Need a Context Window?

An LLM cannot remember an unlimited amount of text.

It has a fixed memory capacity during inference.

Think of it as the model's **working memory**.

---

# 3. Context Window Flow

```text
User Prompt

+

Conversation History

+

Retrieved Documents

+

Output Tokens

↓

Must Fit

↓

Context Window
```

---

Within the limit ✅

# 5. Input Tokens vs Output Tokens

```text
Context Window

↓

Input Tokens

+

Output Tokens
```

Example

```
Context Window

=

16K

Input

=

10K

Maximum Output

=

6K
```

You cannot exceed the total limit.

# 7. What Happens If the Limit Is Exceeded?

Suppose

```
Context Window

=

8K Tokens
```

Input

```
9K Tokens
```

Possible outcomes

- Request rejected
- Old messages removed
- Documents truncated
- Prompt shortened

The model **cannot process more tokens than its context window**.

# 8. Small Coding Example

Using the OpenAI Responses API

```python
from openai import OpenAI

client = OpenAI()

response = client.responses.create(
    model="gpt-4.1",
    input="Explain Context Window",
    max_output_tokens=300
)

print(response.output_text)
```

Here:

- **Input Tokens** → `"Explain Context Window"`
- **Output Tokens** → Limited to `300`

Both count toward the model's context window.

# 12. Common Mistakes

❌ Ignoring conversation history.

❌ Forgetting retrieved documents consume tokens.

❌ Believing output tokens are separate from the context limit.

# 14. Interview Questions

## Q1. What is a Context Window?

**Answer:**

The Context Window is the maximum number of input and output tokens an LLM can process in a single request.

---

## Q2. What contributes to the Context Window?

**Answer:**

The context window includes the system prompt, user messages, conversation history, retrieved documents, tool outputs, and the generated response.

---

## Q3. What happens if the Context Window is exceeded?

**Answer:**

The request may fail, or the application may need to truncate or summarize the input so that the total number of tokens fits within the model's context window.

---

## Q4. Why is Context Window important in RAG?

**Answer:**

Retrieved documents consume context window space. Retrieving too many or overly large documents can leave insufficient room for the model to generate a high-quality response.

---

## Q5. How can you manage long conversations?

**Answer:**

Use conversation summarization, memory, or selective retrieval to reduce the number of tokens while preserving important context.

# 16. Real Interview Scenario

**Question:**

> Your RAG application retrieves 20 large documents, and the model returns an error stating that the context length has been exceeded. How would you solve this?

**Answer:**

I would reduce the number of retrieved documents, retrieve only the most relevant chunks, summarize long documents, or use a model with a larger context window. The goal is to ensure that the prompt, retrieved context, and expected output all fit within the model's token limit.
