# 1. Why Do We Need Multi-Head Attention?

In the previous topic, we learned that **Self-Attention** helps a token find important words.

But what if a sentence contains **multiple relationships**?

Example:

```
Suraj works at OpenAI and lives in Pune.
```

The model should understand:

- Suraj → works at → OpenAI
- Suraj → lives in → Pune

One attention mechanism may focus on only one relationship.

We need **multiple attention heads** to learn different relationships simultaneously.

---

# 2. What is Multi-Head Attention?

Instead of using **one Self-Attention mechanism**, the Transformer runs **multiple Self-Attention mechanisms in parallel**.

Each attention head learns different patterns from the same sentence.

---

# Interview Definition ⭐⭐⭐⭐⭐

> **Multi-Head Attention is a Transformer mechanism that runs multiple independent Self-Attention operations in parallel, allowing the model to capture different types of relationships within the same input sequence.**

# 14. Interview Questions

## Q1. What is Multi-Head Attention?

**Answer:**

Multi-Head Attention runs multiple Self-Attention mechanisms in parallel, allowing the Transformer to learn different relationships within the same input sequence.

---

## Q2. Why do we need multiple heads?

**Answer:**

A single attention head may capture only one type of relationship. Multiple heads enable the model to learn grammatical, semantic, positional, and contextual relationships simultaneously.

---

## Q3. Does every head receive the same input?

**Answer:**

Yes. Every head processes the same input embeddings but learns different attention patterns.

---

## Q4. What happens after all heads finish processing?

**Answer:**

Their outputs are concatenated and passed through a linear projection before moving to the Feed Forward Network.

---

## Q5. Is Multi-Head Attention used in GPT and Llama?

**Answer:**

Yes. Multi-Head Attention is a fundamental component of modern Transformer-based LLMs such as GPT, Llama, Gemini, Claude, and Mistral.

# 16. Real Interview Scenario

**Question:**

> Why doesn't a Transformer use just one Self-Attention head?

**Answer:**

A single attention head is limited in the types of relationships it can capture. Multi-Head Attention allows different heads to focus on different aspects of the input, such as grammar, semantics, entity relationships, and long-range dependencies. Combining these perspectives produces a richer contextual representation.

---

# 17. 30-Second Interview Answer

> **Multi-Head Attention extends Self-Attention by running multiple independent attention mechanisms in parallel. Each attention head learns different relationships within the input sequence, such as semantic meaning, grammatical structure, or entity relationships. The outputs of all heads are combined to create a richer contextual representation, making Transformers more effective than using a single attention mechanism.**
