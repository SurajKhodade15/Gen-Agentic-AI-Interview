## 1. Two sum\n\n### Problem statement\nReturn indices of two values that add to `target`; assume exactly one solution.


```python
def two_sum(nums, target):
    index = {}
    for i, value in enumerate(nums):
        if target - value in index:
            return [index[target - value], i]
        index[value] = i

two_sum([2, 7, 11, 15], 9)
```

### Complexity\n- **Time:** O(n)\n- **Space:** O(n)\n\n### Interview tip\nStore the complement lookup after checking, which handles a duplicated value correctly.\n\n### Follow-up questions\n- What if all pairs are needed? What if the input is sorted?


## 2. Product except self\n\n### Problem statement\nReturn an array where each position is the product of all other elements, without division.


```python
def product_except_self(nums):
    out = [1] * len(nums)
    prefix = 1
    for i, value in enumerate(nums):
        out[i] = prefix; prefix *= value
    suffix = 1
    for i in range(len(nums) - 1, -1, -1):
        out[i] *= suffix; suffix *= nums[i]
    return out

product_except_self([1, 2, 3, 4])
```

### Complexity\n- **Time:** O(n)\n- **Space:** O(1) auxiliary (excluding output)\n\n### Interview tip\nCall out that the output array carries prefix products.\n\n### Follow-up questions\n- How do zero values affect a division-based alternative?


## 3. Merge intervals\n\n### Problem statement\nMerge overlapping closed intervals represented as `[start, end]`.


```python
def merge_intervals(intervals):
    if not intervals: return []
    intervals.sort(key=lambda x: x[0])
    merged = [intervals[0][:]]
    for start, end in intervals[1:]:
        if start <= merged[-1][1]:
            merged[-1][1] = max(merged[-1][1], end)
        else:
            merged.append([start, end])
    return merged

merge_intervals([[1,3], [2,6], [8,10], [15,18]])
```

### Complexity\n- **Time:** O(n log n)\n- **Space:** O(n)\n\n### Interview tip\nAsk whether touching intervals such as `[1,2]` and `[2,3]` should merge.\n\n### Follow-up questions\n- Can this be done online as intervals arrive?


## 4. Top K frequent\n\n### Problem statement\nReturn the `k` most frequent integers without sorting all distinct values.


```python
from collections import Counter

def top_k_frequent(nums, k):
    counts = Counter(nums)
    buckets = [[] for _ in range(len(nums) + 1)]
    for value, count in counts.items(): buckets[count].append(value)
    result = []
    for count in range(len(buckets) - 1, 0, -1):
        result.extend(buckets[count])
        if len(result) >= k: return result[:k]

top_k_frequent([1,1,1,2,2,3], 2)
```

### Complexity\n- **Time:** O(n)\n- **Space:** O(n)\n\n### Interview tip\nCompare bucket sort with a size-k heap; choose based on the data shape.\n\n### Follow-up questions\n- How would you maintain this for an unbounded event stream?
