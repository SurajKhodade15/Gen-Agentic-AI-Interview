## 1. Longest unique substring\n\n### Problem statement\nReturn the length of the longest substring of `s` with no repeated characters.


```python
def longest_unique(s):
    left = best = 0
    seen = {}
    for right, ch in enumerate(s):
        if ch in seen and seen[ch] >= left:
            left = seen[ch] + 1
        seen[ch] = right
        best = max(best, right - left + 1)
    return best

longest_unique('abcaabcd')
```

### Complexity\n- **Time:** O(n)\n- **Space:** O(min(n, alphabet))\n\n### Interview tip\nName the invariant: the current window is always duplicate-free.\n\n### Follow-up questions\n- How would you return the substring too? What changes for a stream of characters?


## 2. Group anagrams\n\n### Problem statement\nGroup strings that are anagrams. Preserve all occurrences.


```python
from collections import defaultdict

def group_anagrams(words):
    groups = defaultdict(list)
    for word in words:
        groups[''.join(sorted(word))].append(word)
    return list(groups.values())

group_anagrams(['eat', 'tea', 'tan', 'ate', 'nat', 'bat'])
```

### Complexity\n- **Time:** O(n * k log k)\n- **Space:** O(n * k)\n\n### Interview tip\nMention a 26-count tuple key when inputs are lowercase English letters.\n\n### Follow-up questions\n- When does sorting become a bottleneck? How do Unicode and normalization affect this?


## 3. Minimum window substring\n\n### Problem statement\nReturn the smallest substring of `s` containing every character of `t`, including multiplicity; return an empty string if none exists.


```python
from collections import Counter

def min_window(s, t):
    need = Counter(t); missing = len(t)
    left = start = end = 0
    for right, ch in enumerate(s, 1):
        if need[ch] > 0: missing -= 1
        need[ch] -= 1
        if missing == 0:
            while left < right and need[s[left]] < 0:
                need[s[left]] += 1; left += 1
            if end == 0 or right - left < end - start:
                start, end = left, right
            need[s[left]] += 1; missing += 1; left += 1
    return s[start:end]

min_window('ADOBECODEBANC', 'ABC')
```

### Complexity\n- **Time:** O(|s| + |t|)\n- **Space:** O(alphabet)\n\n### Interview tip\nExplain why each pointer only moves forward.\n\n### Follow-up questions\n- How would you support arbitrary token sequences instead of characters?


## 4. Run-length encode\n\n### Problem statement\nCompress consecutive repeated characters as `character + count` (e.g. `aaabb` → `a3b2`).


```python
from itertools import groupby

def rle_encode(s):
    return ''.join(f'{ch}{sum(1 for _ in group)}' for ch, group in groupby(s))

rle_encode('aaabbcccc')
```

### Complexity\n- **Time:** O(n)\n- **Space:** O(n) output\n\n### Interview tip\nClarify the output format and whether digits can appear in the input.\n\n### Follow-up questions\n- Write a decoder. How would you make the format unambiguous?
