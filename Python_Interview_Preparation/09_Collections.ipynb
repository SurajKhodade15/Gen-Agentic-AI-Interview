## 1. Task scheduler\n\n### Problem statement\nGiven `(task, cooldown)` events, return a schedule length for identical tasks separated by `n` slots.


```python
from collections import Counter, deque
import heapq

def least_interval(tasks, n):
    heap = [-c for c in Counter(tasks).values()]; heapq.heapify(heap)
    wait = deque(); time = 0
    while heap or wait:
        time += 1
        if heap:
            remaining = heapq.heappop(heap) + 1
            if remaining: wait.append((time + n, remaining))
        if wait and wait[0][0] == time: heapq.heappush(heap, wait.popleft()[1])
    return time

least_interval(list('AAABBB'), 2)
```

### Complexity\n- **Time:** O(tasks log unique tasks)\n- **Space:** O(unique tasks)\n\n### Interview tip\nExplain the max-heap and cooldown queue as separate concerns.\n\n### Follow-up questions\n- Return the actual schedule, including idle slots.


## 2. Median tracker\n\n### Problem statement\nMaintain the median as numbers are added one at a time.


```python
import heapq

class MedianFinder:
    def __init__(self): self.low, self.high = [], []
    def add(self, value):
        heapq.heappush(self.low, -value)
        heapq.heappush(self.high, -heapq.heappop(self.low))
        if len(self.high) > len(self.low): heapq.heappush(self.low, -heapq.heappop(self.high))
    def median(self):
        return -self.low[0] if len(self.low) > len(self.high) else (-self.low[0] + self.high[0]) / 2
```

### Complexity\n- **Time:** O(log n) add, O(1) median\n- **Space:** O(n)\n\n### Interview tip\nName the invariants: sizes differ by at most one; all low values precede high values.\n\n### Follow-up questions\n- How do you support removals or sliding-window medians?


## 3. Bounded event buffer\n\n### Problem statement\nUse a bounded deque to retain only the last `n` events.


```python
from collections import deque

def recent_events(events, n):
    buffer = deque(maxlen=n)
    for event in events: buffer.append(event)
    return list(buffer)

recent_events(['a', 'b', 'c', 'd'], 3)
```

### Complexity\n- **Time:** O(n)\n- **Space:** O(capacity)\n\n### Interview tip\n`deque(maxlen=...)` expresses the retention rule directly.\n\n### Follow-up questions\n- How would you make this safe for concurrent producers?


## 4. Named result\n\n### Problem statement\nRepresent retrieval results with a `namedtuple` and sort by score descending.


```python
from collections import namedtuple

SearchResult = namedtuple('SearchResult', 'doc_id score text')

def rank(results):
    return sorted(results, key=lambda r: r.score, reverse=True)

rank([SearchResult('a', .2, 'x'), SearchResult('b', .9, 'y')])
```

### Complexity\n- **Time:** O(n log n)\n- **Space:** O(n)\n\n### Interview tip\nCompare namedtuple, dataclass, and Pydantic model based on validation needs.\n\n### Follow-up questions\n- What metadata is necessary for citations and observability?
