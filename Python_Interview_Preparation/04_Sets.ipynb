## 1. Longest consecutive sequence\n\n### Problem statement\nReturn the length of the longest consecutive run in an unsorted list.


```python
def longest_consecutive(nums):
    values = set(nums); best = 0
    for value in values:
        if value - 1 not in values:
            length = 1
            while value + length in values: length += 1
            best = max(best, length)
    return best

longest_consecutive([100,4,200,1,3,2])
```

### Complexity\n- **Time:** O(n) average\n- **Space:** O(n)\n\n### Interview tip\nOnly expand from sequence starts; that prevents repeated scans.\n\n### Follow-up questions\n- What are the hash-table worst-case guarantees?


## 2. Set similarity\n\n### Problem statement\nCompute Jaccard similarity of two collections of tokens, treating them as sets.


```python
def jaccard(a, b):
    left, right = set(a), set(b)
    return len(left & right) / len(left | right) if left or right else 1.0

jaccard(['llm', 'rag', 'rag'], ['rag', 'agents'])
```

### Complexity\n- **Time:** O(n + m)\n- **Space:** O(n + m)\n\n### Interview tip\nClarify whether duplicates should matter; Jaccard ignores them.\n\n### Follow-up questions\n- When would weighted Jaccard or cosine similarity be more appropriate?


## 3. Missing numbers\n\n### Problem statement\nReturn all numbers in `1..n` that are absent from `nums`.


```python
def missing_numbers(nums, n):
    return sorted(set(range(1, n + 1)) - set(nums))

missing_numbers([1, 2, 4, 6], 6)
```

### Complexity\n- **Time:** O(n)\n- **Space:** O(n)\n\n### Interview tip\nAsk if input may contain values outside the range or duplicates.\n\n### Follow-up questions\n- Can you do it in-place when mutations are allowed?


## 4. Permission evaluation\n\n### Problem statement\nReturn whether a user has every required permission, allowing roles to contribute permissions.


```python
def is_authorized(user_permissions, roles, role_permissions, required):
    effective = set(user_permissions)
    for role in roles: effective |= set(role_permissions.get(role, ()))
    return set(required) <= effective

is_authorized(['read'], ['editor'], {'editor': ['write']}, ['read', 'write'])
```

### Complexity\n- **Time:** O(total permissions)\n- **Space:** O(total effective permissions)\n\n### Interview tip\nUse `required <= effective` to make the intent unmistakable.\n\n### Follow-up questions\n- How would you model explicit deny rules and permission inheritance?
