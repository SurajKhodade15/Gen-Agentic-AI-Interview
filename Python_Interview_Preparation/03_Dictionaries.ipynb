## 1. LRU cache\n\n### Problem statement\nImplement an LRU cache with `get` and `put` in O(1) average time.


```python
from collections import OrderedDict

class LRUCache:
    def __init__(self, capacity):
        self.capacity = capacity; self.data = OrderedDict()
    def get(self, key):
        if key not in self.data: return -1
        self.data.move_to_end(key)
        return self.data[key]
    def put(self, key, value):
        if key in self.data: self.data.move_to_end(key)
        self.data[key] = value
        if len(self.data) > self.capacity: self.data.popitem(last=False)

c = LRUCache(2); c.put(1, 1); c.put(2, 2); c.get(1)
```

### Complexity\n- **Time:** O(1) average per operation\n- **Space:** O(capacity)\n\n### Interview tip\nIn production, discuss thread safety, TTL, and whether misses should be cached.\n\n### Follow-up questions\n- Implement it with a dict plus a custom doubly-linked list.


## 2. First unique event\n\n### Problem statement\nGiven an iterable of event IDs, return the first ID whose final frequency is one.


```python
from collections import Counter

def first_unique(ids):
    counts = Counter(ids)
    return next((item for item in ids if counts[item] == 1), None)

first_unique(['a', 'b', 'a', 'c', 'b', 'd'])
```

### Complexity\n- **Time:** O(n)\n- **Space:** O(n)\n\n### Interview tip\nState whether this is offline (final frequency) or online; they are different problems.\n\n### Follow-up questions\n- Design the online version with O(1) lookup and removal.


## 3. Deep dictionary merge\n\n### Problem statement\nRecursively merge `right` into `left` without mutating either input; right-side scalars win.


```python
from collections.abc import Mapping

def deep_merge(left, right):
    result = dict(left)
    for key, value in right.items():
        if isinstance(result.get(key), Mapping) and isinstance(value, Mapping):
            result[key] = deep_merge(result[key], value)
        else:
            result[key] = value
    return result

deep_merge({'model': {'temp': 0.2}, 'v': 1}, {'model': {'max_tokens': 500}, 'v': 2})
```

### Complexity\n- **Time:** O(nodes in both mappings)\n- **Space:** O(depth + output)\n\n### Interview tip\nDefine semantics for lists, `None`, and type conflicts before coding.\n\n### Follow-up questions\n- How would you express merge policy per field?


## 4. Invert an index\n\n### Problem statement\nInvert a mapping of document ID to tags into tag to sorted document IDs.


```python
from collections import defaultdict

def invert_index(doc_tags):
    index = defaultdict(list)
    for doc_id, tags in doc_tags.items():
        for tag in tags: index[tag].append(doc_id)
    return {tag: sorted(ids) for tag, ids in index.items()}

invert_index({'d1': ['ai', 'python'], 'd2': ['python'], 'd3': ['ai']})
```

### Complexity\n- **Time:** O(total tags + sorting)\n- **Space:** O(total tags)\n\n### Interview tip\nMention deduplication if a document may repeat a tag.\n\n### Follow-up questions\n- How would you index millions of documents incrementally?
