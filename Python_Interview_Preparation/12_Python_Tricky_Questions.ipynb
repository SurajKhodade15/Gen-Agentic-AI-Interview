## 1. Mutable defaults\n\n### Problem statement\nFix a function that should append an item to a fresh list unless a list is provided.


```python
def append_item(item, items=None):
    if items is None:
        items = []
    items.append(item)
    return items

append_item('a'), append_item('b')
```

### Complexity\n- **Time:** O(1) amortized\n- **Space:** O(n) for returned list\n\n### Interview tip\nDefault values are evaluated once at function definition time.\n\n### Follow-up questions\n- Name other shared-state pitfalls in class attributes and closures.


## 2. Late-bound closures\n\n### Problem statement\nCreate callbacks that each return their own loop value.


```python
def make_callbacks():
    return [lambda i=i: i for i in range(3)]

[fn() for fn in make_callbacks()]
```

### Complexity\n- **Time:** O(n)\n- **Space:** O(n)\n\n### Interview tip\nClosures capture names, not a per-iteration snapshot; bind with a default argument.\n\n### Follow-up questions\n- Show an equivalent `functools.partial` solution.


## 3. Equality and hashing\n\n### Problem statement\nCreate an immutable coordinate usable as a dictionary key.


```python
from dataclasses import dataclass

@dataclass(frozen=True)
class Point:
    x: int
    y: int

{Point(1, 2): 'origin-adjacent'}[Point(1, 2)]
```

### Complexity\n- **Time:** O(1) average lookup\n- **Space:** O(1)\n\n### Interview tip\nObjects used as dict keys must have stable hashes consistent with equality.\n\n### Follow-up questions\n- What breaks if a hashed object is mutated after insertion?


## 4. Iterator exhaustion\n\n### Problem statement\nDemonstrate safe reuse by materializing an iterator when multiple passes are required.


```python
def summarize(values):
    values = list(values)
    return {'count': len(values), 'total': sum(values)}

summarize(i for i in range(5))
```

### Complexity\n- **Time:** O(n)\n- **Space:** O(n)\n\n### Interview tip\nGenerators are single-pass; do not hide that consumption from callers.\n\n### Follow-up questions\n- How can `itertools.tee` help, and what memory cost does it hide?
