## 1. Streaming batches\n\n### Problem statement\nYield batches from a stream while retaining only one batch in memory.


```python
from itertools import islice

def batched(stream, size):
    it = iter(stream)
    while batch := list(islice(it, size)):
        yield batch

list(batched((i for i in range(7)), 3))
```

### Complexity\n- **Time:** O(n)\n- **Space:** O(batch size)\n\n### Interview tip\nGenerators compose well for ETL and model-evaluation pipelines.\n\n### Follow-up questions\n- How would you handle a slow consumer or backpressure?


## 2. Flatten nested lists\n\n### Problem statement\nYield scalar values from arbitrarily nested lists; strings remain scalars.


```python
def flatten(value):
    for item in value:
        if isinstance(item, list): yield from flatten(item)
        else: yield item

list(flatten([1, [2, [3, 4]], 5]))
```

### Complexity\n- **Time:** O(n)\n- **Space:** O(depth)\n\n### Interview tip\nState the accepted container types—over-general recursion can surprise callers.\n\n### Follow-up questions\n- How would you detect cyclic structures?


## 3. Paginated API iterator\n\n### Problem statement\nTurn a `fetch_page(cursor)` callable returning `(items, next_cursor)` into an item iterator.


```python
def iter_pages(fetch_page):
    cursor = None
    while True:
        items, cursor = fetch_page(cursor)
        yield from items
        if cursor is None: break
```

### Complexity\n- **Time:** O(total items)\n- **Space:** O(page size)\n\n### Interview tip\nDefine whether an empty page can still have a next cursor.\n\n### Follow-up questions\n- Add retries, cancellation, and checkpointing.


## 4. Running mean\n\n### Problem statement\nYield the running mean for a numeric iterable.


```python
def running_mean(values):
    total = 0
    for count, value in enumerate(values, 1):
        total += value
        yield total / count

list(running_mean([2, 4, 6]))
```

### Complexity\n- **Time:** O(n)\n- **Space:** O(1)\n\n### Interview tip\nFor high-variance floats, discuss numerical stability and Welford's algorithm.\n\n### Follow-up questions\n- How would you compute rolling-window mean instead?
