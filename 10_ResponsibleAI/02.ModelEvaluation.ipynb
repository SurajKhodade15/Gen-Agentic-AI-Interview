# Why Evaluation is Important?

Unlike traditional software, LLMs can generate different outputs for the same input.

Therefore, we continuously evaluate:

- Model Quality
- Agent Behavior
- Prompt Quality
- Failure Cases
- Production Performance

# 4. Failure-Mode Analysis ⭐⭐⭐⭐⭐

## What is it?

Study why the AI failed.

Instead of asking

```
Did it fail?
```

Ask

```
Why did it fail?
```

---

## Common Failure Modes

| Failure | Example |
|----------|----------|
| Hallucination | Invented facts |
| Wrong Tool | Selected wrong API |
| Missing Context | Poor retrieval |
| Prompt Failure | Misunderstood instruction |
| Retrieval Failure | Wrong RAG chunk |
| JSON Parsing Failure | Invalid structured output |
| Timeout | Slow API |
| Infinite Loop | Agent never finishes |

---

## Example

```
Wrong Location Selected
```

Root Cause

```
Poor Metadata Filter
```

Fix

```
Improve Metadata Filtering
```

# Common Interview Questions

### Q1. How do you evaluate an LLM?

**Answer**

Using metrics such as accuracy, relevance, completeness, hallucination rate, latency, cost, and safety, along with automated evaluation tools and human review.

---

### Q2. How do you evaluate an AI Agent?

**Answer**

By verifying that it reasons correctly, selects the appropriate tools, executes the workflow in the right order, uses memory properly, and reaches the expected outcome.

---

### Q3. What is Prompt Evaluation?

**Answer**

Prompt evaluation compares different prompts to measure their impact on accuracy, consistency, latency, token usage, and hallucination rate.

---

### Q4. What is Failure-Mode Analysis?

**Answer**

Failure-mode analysis identifies why an AI system failed—for example due to hallucinations, poor retrieval, incorrect tool selection, or invalid structured output—and uses the findings to improve the system.

---

### Q5. How do you continuously improve a GenAI application?

**Answer**

Monitor production metrics, analyze traces and logs, evaluate prompts and retrieval quality, improve embeddings and chunking, refine agent workflows, and redeploy improved versions.

# 60-Second Interview Answer

> **In production GenAI systems, evaluation is an ongoing process rather than a one-time activity. We evaluate the model for accuracy, relevance, latency, cost, and hallucination rate. For AI agents, we verify reasoning, planning, tool selection, and workflow execution. We compare prompt versions to improve response quality and reduce token usage. We perform failure-mode analysis to identify issues such as poor retrieval, incorrect tool calls, or invalid structured outputs. Using tools like LangSmith, Ragas, and DeepEval, along with production logs and human feedback, we continuously refine prompts, retrieval strategies, embeddings, and agent workflows to improve the overall performance of the system.**
